## Ejercicio 1: Gestor de Portafolio

Crea un sistema para gestionar un portafolio de inversiones usando diccionarios.

**Datos iniciales:**
```python
portafolio = {
    "AAPL": {"cantidad": 50, "precio_compra": 150},
    "GOOGL": {"cantidad": 20, "precio_compra": 2500},
    "MSFT": {"cantidad": 30, "precio_compra": 300}
}

precios_actuales = {"AAPL": 175, "GOOGL": 2800, "MSFT": 350}
```

**Tareas:**
1. Calcula el valor total del portafolio (compra y actual)
2. Calcula la ganancia/pérdida de cada acción
3. Calcula el ROI de cada inversión
4. Identifica la mejor y peor inversión
5. Crea un reporte formateado

````python
# Solución Ejercicio 1: Gestor de portafolio
portafolio = {
    "AAPL": {"cantidad": 50, "precio_compra": 150},
    "GOOGL": {"cantidad": 20, "precio_compra": 2500},
    "MSFT": {"cantidad": 30, "precio_compra": 300}
}
precios_actuales = {"AAPL": 175, "GOOGL": 2800, "MSFT": 350}

valor_compra_total = sum(info["cantidad"] * info["precio_compra"] for info in portafolio.values())
valor_actual_total = sum(portafolio[s]["cantidad"] * precios_actuales[s] for s in portafolio)
resultados = []
for simbolo, info in portafolio.items():
    v_compra = info["cantidad"] * info["precio_compra"]
    v_actual = info["cantidad"] * precios_actuales[simbolo]
    ganancia = v_actual - v_compra
    roi = (ganancia / v_compra) * 100
    resultados.append((simbolo, ganancia, roi))
resultados.sort(key=lambda x: x[2], reverse=True)
print("=== REPORTE PORTAFOLIO ===")
for s, info in portafolio.items():
    vc = info["cantidad"] * info["precio_compra"]
    va = info["cantidad"] * precios_actuales[s]
    g = va - vc
    roi = (g / vc) * 100
    print(f"{s}: Compra ${vc:,} | Actual ${va:,} | Ganancia ${g:,} | ROI {roi:.2f}%")
print(f"\nTotal compra: ${valor_compra_total:,} | Total actual: ${valor_actual_total:,}")
print(f"Mejor: {resultados[0][0]} (ROI {resultados[0][2]:.2f}%) | Peor: {resultados[-1][0]} (ROI {resultados[-1][2]:.2f}%)")

````

## Ejercicio 2: Análisis de Gastos con Comprensión de Listas

Tienes una lista de gastos del mes:
```python
gastos = [
    ("Alimentación", 350),
    ("Transporte", 120),
    ("Alimentación", 280),
    ("Servicios", 180),
    ("Entretenimiento", 150),
    ("Transporte", 90),
    ("Alimentación", 320),
    ("Servicios", 200)
]
```

**Tareas:**
1. Usa comprensión de listas para extraer solo los montos
2. Calcula el total de gastos
3. Agrupa los gastos por categoría usando un diccionario
4. Calcula el total y promedio por categoría
5. Identifica la categoría con mayor gasto

````python
# Solución Ejercicio 2: Gastos con comprensión de listas
gastos = [
    ("Alimentación", 350), ("Transporte", 120), ("Alimentación", 280),
    ("Servicios", 180), ("Entretenimiento", 150), ("Transporte", 90),
    ("Alimentación", 320), ("Servicios", 200)
]
montos = [m for _, m in gastos]
total = sum(montos)
categorias_unicas = list(set(c for c, _ in gastos))
totales = {}
for cat in categorias_unicas:
    totales[cat] = sum(m for c, m in gastos if c == cat)
counts = {cat: sum(1 for c, _ in gastos if c == cat) for cat in categorias_unicas}
promedios = {cat: totales[cat] / counts[cat] for cat in totales}
cat_mayor = max(totales, key=totales.get)
print("=== ANÁLISIS GASTOS ===")
print(f"Total: ${total:,.2f}")
for cat in totales:
    print(f"  {cat}: Total ${totales[cat]:,.2f} | Promedio ${promedios[cat]:,.2f}")
print(f"Categoría con mayor gasto: {cat_mayor} (${totales[cat_mayor]:,.2f})")

````

## Ejercicio 3: Calculadora de Rendimientos

Dada una serie de precios de cierre de una acción:
```python
precios = [100, 102, 98, 105, 103, 107, 110, 108, 112, 115]
```

**Tareas:**
1. Usa comprensión de listas para calcular rendimientos diarios (%)
2. Calcula el rendimiento promedio
3. Identifica el día con mayor ganancia y mayor pérdida
4. Crea una lista de tuplas: (día, precio, rendimiento)
5. Filtra solo los días con rendimiento positivo

````python
# Solución Ejercicio 3: Calculadora de rendimientos
precios = [100, 102, 98, 105, 103, 107, 110, 108, 112, 115]
rendimientos = [((precios[i] - precios[i-1]) / precios[i-1]) * 100 for i in range(1, len(precios))]
prom_rend = sum(rendimientos) / len(rendimientos)
tuplas = [(i+1, precios[i+1], rendimientos[i]) for i in range(len(rendimientos))]
positivos = [(d, p, r) for d, p, r in zip(range(1, len(precios)), precios[1:], rendimientos) if r > 0]
max_gan = rendimientos.index(max(rendimientos)) + 1
max_perd = rendimientos.index(min(rendimientos)) + 1
print("=== RENDIMIENTOS DIARIOS ===")
print(f"Rendimientos (%): {[round(r, 2) for r in rendimientos]}")
print(f"Rendimiento promedio: {prom_rend:.2f}%")
print(f"Día mayor ganancia: día {max_gan} ({rendimientos[max_gan-1]:.2f}%)")
print(f"Día mayor pérdida: día {max_perd} ({rendimientos[max_perd-1]:.2f}%)")
print(f"Días con rendimiento positivo: {[(d, round(r, 2)) for d, _, r in positivos]}")

````

## Ejercicio 4: Comparador de Portafolios

Tienes tres portafolios de inversión:
```python
portafolio_1 = {"AAPL", "GOOGL", "MSFT"}
portafolio_2 = {"AAPL", "MSFT", "TSLA"}
portafolio_3 = {"GOOGL", "TSLA", "AMZN"}
```

**Tareas:**
1. Encuentra las acciones comunes entre los tres portafolios
2. Encuentra las acciones únicas de cada portafolio
3. Crea un conjunto con todas las acciones diferentes
4. Determina qué portafolio tiene más diversificación
5. Encuentra acciones que están en exactamente 2 portafolios

````python
# Solución Ejercicio 4: Comparador de portafolios
portafolio_1 = {"AAPL", "GOOGL", "MSFT"}
portafolio_2 = {"AAPL", "MSFT", "TSLA"}
portafolio_3 = {"GOOGL", "TSLA", "AMZN"}
comunes = portafolio_1 & portafolio_2 & portafolio_3
unicos_1 = portafolio_1 - portafolio_2 - portafolio_3
unicos_2 = portafolio_2 - portafolio_1 - portafolio_3
unicos_3 = portafolio_3 - portafolio_1 - portafolio_2
todas = portafolio_1 | portafolio_2 | portafolio_3
en_exactamente_2 = [a for a in todas if (a in portafolio_1) + (a in portafolio_2) + (a in portafolio_3) == 2]
print("=== COMPARADOR PORTAFOLIOS ===")
print(f"Comunes a los tres: {comunes}")
print(f"Únicos P1: {unicos_1} | P2: {unicos_2} | P3: {unicos_3}")
print(f"Todas las acciones: {todas}")
print(f"Tamaños (diversificación): P1={len(portafolio_1)}, P2={len(portafolio_2)}, P3={len(portafolio_3)}")
print(f"En exactamente 2 portafolios: {en_exactamente_2}")

````

## Ejercicio 5: Análisis de Flujo de Caja

Tienes datos mensuales de ingresos y egresos:
```python
meses = ["Enero", "Febrero", "Marzo", "Abril", "Mayo", "Junio"]
ingresos = [5000, 5200, 4800, 5500, 5100, 5300]
egresos = [3500, 3800, 4000, 3600, 3700, 3900]
```

**Tareas:**
1. Usa zip() para calcular el flujo neto de cada mes
2. Calcula el flujo acumulado mes a mes
3. Identifica el mes con mayor flujo positivo
4. Identifica el mes con mayor flujo negativo
5. Crea un diccionario con: {mes: {"ingreso": X, "egreso": Y, "neto": Z}}

````python
# Solución Ejercicio 5: Análisis de flujo de caja
meses = ["Enero", "Febrero", "Marzo", "Abril", "Mayo", "Junio"]
ingresos = [5000, 5200, 4800, 5500, 5100, 5300]
egresos = [3500, 3800, 4000, 3600, 3700, 3900]
netos = [i - e for i, e in zip(ingresos, egresos)]
acum = 0
flujo_acum = []
for n in netos:
    acum += n
    flujo_acum.append(acum)
max_pos_idx = netos.index(max(netos))
max_neg_idx = netos.index(min(netos))
datos = {mes: {"ingreso": i, "egreso": e, "neto": n} for mes, i, e, n in zip(meses, ingresos, egresos, netos)}
print("=== FLUJO DE CAJA ===")
for mes, i, e, n, fa in zip(meses, ingresos, egresos, netos, flujo_acum):
    print(f"{mes}: Ingreso ${i:,} | Egreso ${e:,} | Neto ${n:,} | Acum ${fa:,}")
print(f"Mayor flujo positivo: {meses[max_pos_idx]} | Mayor negativo: {meses[max_neg_idx]}")
print("Diccionario por mes:", {m: datos[m] for m in meses[:2]}, "...")

````

## Ejercicio 6: Ranking de Fondos de Inversión

Tienes información de varios fondos:
```python
fondos = {
    "Fondo Conservador": 6.5,
    "Fondo Moderado": 9.2,
    "Fondo Agresivo": 15.8,
    "Fondo Balanceado": 8.1,
    "Fondo Crecimiento": 12.3
}
```

Los valores son rendimientos anuales (%).

**Tareas:**
1. Ordena los fondos por rendimiento (descendente)
2. Crea un ranking numerado
3. Clasifica los fondos en categorías:
   - Alto (>10%)
   - Medio (5-10%)
   - Bajo (<5%)
4. Encuentra el fondo con mejor y peor rendimiento
5. Calcula el rendimiento promedio de todos los fondos

````python
# Solución Ejercicio 6: Ranking de fondos
fondos = {
    "Fondo Conservador": 6.5, "Fondo Moderado": 9.2, "Fondo Agresivo": 15.8,
    "Fondo Balanceado": 8.1, "Fondo Crecimiento": 12.3
}
ordenados = sorted(fondos.items(), key=lambda x: x[1], reverse=True)
ranking = [(i+1, nombre, r) for i, (nombre, r) in enumerate(ordenados)]
categorias = {}
for nombre, r in fondos.items():
    if r > 10: cat = "Alto"
    elif r >= 5: cat = "Medio"
    else: cat = "Bajo"
    categorias[nombre] = cat
mejor = ordenados[0]
peor = ordenados[-1]
prom = sum(fondos.values()) / len(fondos)
print("=== RANKING FONDOS ===")
for pos, nombre, r in ranking:
    print(f"{pos}. {nombre}: {r}% ({categorias[nombre]})")
print(f"\nMejor: {mejor[0]} ({mejor[1]}%) | Peor: {peor[0]} ({peor[1]}%)")
print(f"Rendimiento promedio: {prom:.2f}%")

````
