## 8. Errores y Excepciones

### 8.1 ¿Qué son los Errores?

En Python hay dos tipos principales de errores:

#### **Errores de Sintaxis (Syntax Errors)**
- Ocurren cuando Python no puede interpretar el código
- Detectados antes de ejecutar el programa
- Deben corregirse antes de ejecutar

#### **Excepciones (Exceptions)**
- Ocurren durante la ejecución del programa
- El código es sintácticamente correcto, pero algo sale mal
- Pueden manejarse con try/except
- Ejemplos: ZeroDivisionError, ValueError, KeyError, FileNotFoundError

````python
# Ejemplo de error de sintaxis (comentado para no romper el notebook)
# x = 5
# if x > 3  # Falta los dos puntos
#     print(x)

# Sintaxis correcta
x = 5
if x > 3:
    print(x)

print("\n" + "="*50)

# Errores de sintaxis comunes:
# - Falta de dos puntos (:)
# - Paréntesis, llaves o corchetes no balanceados
# - Indentación incorrecta
# - Palabras clave mal escritas
````

### 8.2 Excepciones Comunes

Python tiene muchas excepciones predefinidas. Aquí las más comunes:

````python
# ZeroDivisionError - División por cero
try:
    resultado = 10 / 0
except ZeroDivisionError as e:
    print(f"Error: {e}")
    print("No se puede dividir por cero")

print("\n" + "="*50)

# ValueError - Valor incorrecto
try:
    numero = int("abc")
except ValueError as e:
    print(f"Error: {e}")
    print("No se puede convertir 'abc' a entero")

print("\n" + "="*50)

# KeyError - Clave no encontrada en diccionario
diccionario = {"a": 1, "b": 2}
try:
    valor = diccionario["c"]
except KeyError as e:
    print(f"Error: {e}")
    print("La clave 'c' no existe en el diccionario")

print("\n" + "="*50)

# IndexError - Índice fuera de rango
lista = [1, 2, 3]
try:
    elemento = lista[10]
except IndexError as e:
    print(f"Error: {e}")
    print("El índice 10 está fuera de rango")

print("\n" + "="*50)

# TypeError - Operación en tipo incorrecto
try:
    resultado = "texto" + 5
except TypeError as e:
    print(f"Error: {e}")
    print("No se puede concatenar string con int")

print("\n" + "="*50)

# FileNotFoundError - Archivo no encontrado
try:
    with open("archivo_inexistente.txt", "r") as f:
        contenido = f.read()
except FileNotFoundError as e:
    print(f"Error: {e}")
    print("El archivo no existe")
````

### 8.3 Manejo de Excepciones: try/except

#### Estructura Básica:
```python
try:
    # Código que puede fallar
except TipoExcepcion:
    # Qué hacer si falla
```

#### Buenas Prácticas:
- **Específico es mejor**: Captura excepciones específicas, no genéricas
- **No silencies errores**: Siempre maneja o registra los errores
- **Mensajes claros**: Proporciona información útil sobre el error

````python
# try-except básico
def dividir(a, b):
    """Divide dos números con manejo de errores."""
    try:
        resultado = a / b
        return resultado
    except ZeroDivisionError:
        print("Error: No se puede dividir por cero")
        return None
    except TypeError:
        print("Error: Los argumentos deben ser números")
        return None

print(f"10 / 2 = {dividir(10, 2)}")
print(f"10 / 0 = {dividir(10, 0)}")
print(f"'10' / 2 = {dividir('10', 2)}")

print("\n" + "="*50)

# Capturar múltiples excepciones
def procesar_numero(texto):
    """Convierte texto a número y calcula 100/número."""
    try:
        numero = float(texto)
        resultado = 100 / numero
        return resultado
    except ValueError:
        print(f"Error: '{texto}' no es un número válido")
        return None
    except ZeroDivisionError:
        print("Error: División por cero")
        return None

procesar_numero("10")
procesar_numero("abc")
procesar_numero("0")
````

### 8.4 try/except/else/finally

La estructura completa de manejo de excepciones incluye:

- **try**: Código que puede generar excepciones
- **except**: Maneja excepciones específicas
- **else**: Se ejecuta si NO hay excepciones
- **finally**: Siempre se ejecuta (con o sin excepciones)

````python
# Estructura completa try/except/else/finally
def leer_archivo_seguro(nombre):
    """Lee un archivo con manejo completo de errores."""
    try:
        with open(nombre, "r", encoding="utf-8") as archivo:
            contenido = archivo.read()
    except FileNotFoundError:
        print(f"Error: Archivo '{nombre}' no encontrado")
        return None
    except PermissionError:
        print(f"Error: Sin permiso para leer '{nombre}'")
        return None
    except Exception as e:
        print(f"Error inesperado: {type(e).__name__}: {e}")
        return None
    else:
        # Se ejecuta solo si NO hubo excepciones
        print(f"Archivo '{nombre}' leído exitosamente")
        return contenido
    finally:
        # Siempre se ejecuta (con o sin errores)
        print("Operación de lectura finalizada")

# Probar con archivo que existe
contenido = leer_archivo_seguro("ejemplo.txt")
if contenido:
    print(f"Contenido: {len(contenido)} caracteres")

print("\n" + "="*50)

# Probar con archivo que no existe
contenido2 = leer_archivo_seguro("no_existe.txt")
````

### 8.5 Lanzar Excepciones (raise)

Puedes lanzar excepciones manualmente usando `raise`. Útil para validación y errores personalizados.

````python
# Lanzar excepciones
def validar_edad(edad):
    """Valida que la edad sea razonable."""
    if not isinstance(edad, (int, float)):
        raise TypeError(f"La edad debe ser un número, recibido: {type(edad).__name__}")
    if edad < 0:
        raise ValueError(f"La edad no puede ser negativa: {edad}")
    if edad > 150:
        raise ValueError(f"La edad no puede ser mayor a 150: {edad}")
    return True

# Probar validaciones
try:
    validar_edad(25)  # Válido
    print("Edad 25: Válida")
except (ValueError, TypeError) as e:
    print(f"Error: {e}")

try:
    validar_edad(-5)  # Inválido
except ValueError as e:
    print(f"\nEdad -5: {e}")

try:
    validar_edad("veinticinco")  # Tipo incorrecto
except TypeError as e:
    print(f"\nEdad 'veinticinco': {e}")
````

### 8.6 Excepciones Personalizadas

Puedes crear tus propias excepciones heredando de la clase `Exception` o de excepciones más específicas.

````python
# Excepciones personalizadas
class SaldoInsuficienteError(Exception):
    """Excepción lanzada cuando no hay saldo suficiente."""
    pass

class MontoInvalidoError(Exception):
    """Excepción lanzada cuando el monto es inválido."""
    def __init__(self, monto, mensaje="Monto inválido"):
        self.monto = monto
        self.mensaje = mensaje
        super().__init__(f"{mensaje}: ${monto}")

def retirar(saldo, monto):
    """Retira dinero de una cuenta."""
    if monto <= 0:
        raise MontoInvalidoError(monto, "El monto debe ser positivo")
    if monto > saldo:
        raise SaldoInsuficienteError(
            f"Saldo insuficiente. Tienes ${saldo:,.2f}, intentas retirar ${monto:,.2f}"
        )
    return saldo - monto

# Probar excepciones personalizadas
try:
    nuevo_saldo = retirar(100, 150)
except SaldoInsuficienteError as e:
    print(f"Error de saldo: {e}")

try:
    nuevo_saldo = retirar(100, -50)
except MontoInvalidoError as e:
    print(f"\nError de monto: {e}")
    print(f"Monto recibido: ${e.monto}")

# Retiro exitoso
try:
    nuevo_saldo = retirar(100, 50)
    print(f"\nRetiro exitoso. Nuevo saldo: ${nuevo_saldo:,.2f}")
except (SaldoInsuficienteError, MontoInvalidoError) as e:
    print(f"Error: {e}")
````

## 9. Clases y Programación Orientada a Objetos

### 9.1 ¿Qué es la Programación Orientada a Objetos (POO)?

La **POO** es un paradigma de programación que organiza el código en "objetos" que representan entidades del mundo real. Cada objeto tiene:

- **Atributos** (datos/propiedades): Características del objeto
- **Métodos** (funciones): Acciones que el objeto puede realizar

#### Conceptos Clave:
- **Clase**: Plantilla o molde para crear objetos
- **Objeto/Instancia**: Ejemplar específico de una clase
- **Encapsulación**: Agrupar datos y métodos relacionados
- **Herencia**: Crear nuevas clases basadas en clases existentes
- **Polimorfismo**: Diferentes clases pueden tener métodos con el mismo nombre

### 9.2 Definir una Clase

#### Sintaxis:
```python
class NombreClase:
    """Docstring de la clase"""
    
    def __init__(self, parametros):
        # Constructor - inicializa el objeto
        self.atributo = valor
    
    def metodo(self, parametros):
        # Método de la clase
        return resultado
```

#### Conceptos Importantes:
- **self**: Referencia al objeto actual (similar a "this" en otros lenguajes)
- **__init__**: Método constructor, se ejecuta al crear el objeto
- **Atributos de instancia**: Variables específicas de cada objeto
- **Atributos de clase**: Variables compartidas por todos los objetos

````python
# Clase simple
class CuentaBancaria:
    """Clase que representa una cuenta bancaria."""
    
    # Atributo de clase (compartido por todas las instancias)
    banco = "Banco Python"
    
    def __init__(self, titular, saldo_inicial=0, numero_cuenta=None):
        """
        Inicializa una cuenta bancaria.
        
        Parámetros:
        -----------
        titular : str
            Nombre del titular de la cuenta
        saldo_inicial : float
            Saldo inicial de la cuenta (default: 0)
        numero_cuenta : str, optional
            Número de cuenta (se genera automáticamente si no se proporciona)
        """
        # Atributos de instancia (específicos de cada objeto)
        self.titular = titular
        self.saldo = saldo_inicial
        self.numero_cuenta = numero_cuenta or f"ACC-{id(self)}"
        self.transacciones = []  # Historial de transacciones
    
    def depositar(self, monto):
        """Deposita dinero en la cuenta."""
        if monto > 0:
            self.saldo += monto
            self.transacciones.append(("Depósito", monto))
            return True
        return False
    
    def retirar(self, monto):
        """Retira dinero de la cuenta."""
        if monto > 0 and monto <= self.saldo:
            self.saldo -= monto
            self.transacciones.append(("Retiro", -monto))
            return True
        return False
    
    def consultar_saldo(self):
        """Consulta el saldo actual."""
        return self.saldo
    
    def obtener_historial(self):
        """Obtiene el historial de transacciones."""
        return self.transacciones
    
    def __str__(self):
        """Representación en string (para print)."""
        return f"Cuenta {self.numero_cuenta} - {self.titular}: ${self.saldo:,.2f}"
    
    def __repr__(self):
        """Representación oficial (para desarrolladores)."""
        return f"CuentaBancaria(titular='{self.titular}', saldo={self.saldo})"

# Crear instancias (objetos)
cuenta1 = CuentaBancaria("Juan Pérez", 1000)
cuenta2 = CuentaBancaria("María González", 500, "ACC-12345")

print(cuenta1)
print(cuenta2)
print(f"\nBanco: {CuentaBancaria.banco}")  # Atributo de clase

# Usar métodos
cuenta1.depositar(500)
cuenta1.retirar(200)
print(f"\n{cuenta1}")
print(f"Historial: {cuenta1.obtener_historial()}")
````

### 9.3 Métodos Especiales (Magic Methods)

Los **métodos especiales** (también llamados "dunder methods" por los dobles guiones bajos) permiten definir cómo se comportan los objetos con operadores y funciones built-in.

#### Métodos Especiales Comunes:
- `__init__`: Constructor
- `__str__`: Representación legible (para usuarios)
- `__repr__`: Representación oficial (para desarrolladores)
- `__len__`: Define len(objeto)
- `__eq__`: Define ==
- `__lt__`, `__le__`, `__gt__`, `__ge__`: Comparaciones
- `__add__`, `__sub__`, etc.: Operadores aritméticos

````python
# Ejemplo de métodos especiales
class Dinero:
    """Clase para representar cantidades de dinero."""
    
    def __init__(self, cantidad, moneda="USD"):
        self.cantidad = cantidad
        self.moneda = moneda
    
    def __str__(self):
        return f"${self.cantidad:,.2f} {self.moneda}"
    
    def __repr__(self):
        return f"Dinero({self.cantidad}, '{self.moneda}')"
    
    def __add__(self, otro):
        """Permite sumar dos objetos Dinero."""
        if isinstance(otro, Dinero) and self.moneda == otro.moneda:
            return Dinero(self.cantidad + otro.cantidad, self.moneda)
        elif isinstance(otro, (int, float)):
            return Dinero(self.cantidad + otro, self.moneda)
        else:
            raise TypeError("No se puede sumar")
    
    def __eq__(self, otro):
        """Define igualdad."""
        if isinstance(otro, Dinero):
            return self.cantidad == otro.cantidad and self.moneda == otro.moneda
        return False
    
    def __lt__(self, otro):
        """Define menor que."""
        if isinstance(otro, Dinero) and self.moneda == otro.moneda:
            return self.cantidad < otro.cantidad
        raise TypeError("No se puede comparar")

# Usar la clase
d1 = Dinero(1000, "USD")
d2 = Dinero(500, "USD")

print(f"d1: {d1}")
print(f"d2: {d2}")
print(f"\nd1 + d2: {d1 + d2}")
print(f"d1 + 200: {d1 + 200}")
print(f"d1 == d2: {d1 == d2}")
print(f"d1 < d2: {d1 < d2}")
````

### 9.4 Herencia

La **herencia** permite crear una nueva clase basada en una clase existente. La nueva clase hereda todos los atributos y métodos de la clase padre.

#### Ventajas:
- **Reutilización de código**: No repetir código
- **Extensibilidad**: Agregar funcionalidad sin modificar la clase padre
- **Organización**: Jerarquía lógica de clases

#### Sintaxis:
```python
class ClaseHija(ClasePadre):
    def __init__(self, ...):
        super().__init__(...)  # Llamar al constructor padre
        # Código adicional
```

````python
# Clase base (padre)
class CuentaBancaria:
    """Clase base para cuentas bancarias."""
    
    def __init__(self, titular, saldo_inicial=0):
        self.titular = titular
        self.saldo = saldo_inicial
    
    def depositar(self, monto):
        if monto > 0:
            self.saldo += monto
            return True
        return False
    
    def retirar(self, monto):
        if monto > 0 and monto <= self.saldo:
            self.saldo -= monto
            return True
        return False
    
    def consultar_saldo(self):
        return self.saldo
    
    def __str__(self):
        return f"Cuenta de {self.titular}: ${self.saldo:,.2f}"

# Clase derivada (hija) - Herencia
class CuentaAhorros(CuentaBancaria):
    """Cuenta de ahorros con interés."""
    
    def __init__(self, titular, saldo_inicial=0, tasa_interes=0.02):
        # Llamar al constructor de la clase padre
        super().__init__(titular, saldo_inicial)
        # Agregar atributos específicos de CuentaAhorros
        self.tasa_interes = tasa_interes
    
    def aplicar_interes(self):
        """Aplica interés al saldo."""
        interes = self.saldo * self.tasa_interes
        self.saldo += interes
        return interes
    
    # Sobrescribir método padre (override)
    def __str__(self):
        return f"Cuenta Ahorros de {self.titular}: ${self.saldo:,.2f} (Tasa: {self.tasa_interes*100}%)"

# Otra clase derivada
class CuentaCorriente(CuentaBancaria):
    """Cuenta corriente con límite de sobregiro."""
    
    def __init__(self, titular, saldo_inicial=0, limite_sobregiro=500):
        super().__init__(titular, saldo_inicial)
        self.limite_sobregiro = limite_sobregiro
    
    def retirar(self, monto):  # Sobrescribir método retirar
        """Permite sobregiro hasta el límite."""
        if monto > 0:
            saldo_minimo = -self.limite_sobregiro
            if self.saldo - monto >= saldo_minimo:
                self.saldo -= monto
                return True
        return False
    
    def __str__(self):
        return f"Cuenta Corriente de {self.titular}: ${self.saldo:,.2f} (Límite: ${self.limite_sobregiro})"

# Usar las clases
cuenta_ahorros = CuentaAhorros("María González", 1000, 0.03)
print(cuenta_ahorros)
interes = cuenta_ahorros.aplicar_interes()
print(f"Interés aplicado: ${interes:,.2f}")
print(f"Nuevo saldo: ${cuenta_ahorros.consultar_saldo():,.2f}")

print("\n" + "="*50)

cuenta_corriente = CuentaCorriente("Juan Pérez", 100, 500)
print(cuenta_corriente)
cuenta_corriente.retirar(400)  # Permite sobregiro
print(f"Después de retirar 400: {cuenta_corriente}")

# Todas las clases pueden usar métodos heredados
cuenta_ahorros.depositar(200)  # Método de la clase padre
print(f"\nCuenta ahorros después de depositar: {cuenta_ahorros}")
````

### 9.5 Encapsulación y Atributos Privados

Python no tiene verdadera privacidad, pero usa convenciones:

- **Sin prefijo**: Público (accesible desde cualquier lugar)
- **Un guion bajo `_`**: "Protegido" (convención, no es privado realmente)
- **Dos guiones bajos `__`**: "Privado" (name mangling, Python cambia el nombre)

````python
class CuentaSegura:
    """Ejemplo de encapsulación."""
    
    def __init__(self, titular, saldo_inicial=0, pin="0000"):
        self.titular = titular  # Público
        self._saldo = saldo_inicial  # Protegido (convención)
        self.__pin = pin  # Privado (name mangling)
    
    def consultar_saldo(self, pin):
        """Consulta saldo con validación de PIN."""
        if pin == self.__pin:
            return self._saldo
        else:
            raise ValueError("PIN incorrecto")
    
    def cambiar_pin(self, pin_actual, pin_nuevo):
        """Cambia el PIN."""
        if pin_actual == self.__pin:
            self.__pin = pin_nuevo
            return True
        return False

cuenta = CuentaSegura("Juan", 1000, "1234")

# Acceso público
print(f"Titular: {cuenta.titular}")

# Acceso protegido (funciona, pero es una convención)
print(f"Saldo (acceso directo): {cuenta._saldo}")

# Acceso privado (name mangling - Python cambia el nombre)
# print(cuenta.__pin)  # Error: AttributeError
# Pero puedes acceder así (no recomendado):
print(f"PIN (acceso forzado): {cuenta._CuentaSegura__pin}")

# Forma correcta (usando métodos)
print(f"\nSaldo con PIN correcto: ${cuenta.consultar_saldo('1234'):,.2f}")
try:
    cuenta.consultar_saldo("0000")
except ValueError as e:
    print(f"Error con PIN incorrecto: {e}")
````

### 9.6 Métodos de Clase y Métodos Estáticos

Además de métodos de instancia, Python soporta:

- **Métodos de instancia**: Reciben `self`, operan sobre el objeto
- **Métodos de clase**: Reciben `cls`, operan sobre la clase
- **Métodos estáticos**: No reciben `self` ni `cls`, son funciones independientes

````python
class CuentaBancaria:
    """Ejemplo de diferentes tipos de métodos."""
    
    # Atributo de clase
    total_cuentas = 0
    
    def __init__(self, titular, saldo=0):
        self.titular = titular
        self.saldo = saldo
        CuentaBancaria.total_cuentas += 1  # Incrementar contador de clase
    
    # Método de instancia (recibe self)
    def depositar(self, monto):
        """Método de instancia - opera sobre el objeto."""
        self.saldo += monto
    
    # Método de clase (recibe cls)
    @classmethod
    def crear_desde_string(cls, cadena):
        """Método de clase - crea objeto desde string."""
        partes = cadena.split(",")
        return cls(partes[0], float(partes[1]))
    
    @classmethod
    def obtener_total_cuentas(cls):
        """Método de clase - retorna total de cuentas."""
        return cls.total_cuentas
    
    # Método estático (no recibe self ni cls)
    @staticmethod
    def validar_saldo(saldo):
        """Método estático - función independiente."""
        return saldo >= 0

# Usar métodos de instancia
cuenta1 = CuentaBancaria("Juan", 1000)
cuenta1.depositar(500)
print(f"Cuenta 1: {cuenta1.titular}, Saldo: ${cuenta1.saldo:,.2f}")

# Usar método de clase
cuenta2 = CuentaBancaria.crear_desde_string("María,2000")
print(f"Cuenta 2: {cuenta2.titular}, Saldo: ${cuenta2.saldo:,.2f}")

# Usar método estático
print(f"\n¿Saldo 1000 es válido? {CuentaBancaria.validar_saldo(1000)}")
print(f"¿Saldo -100 es válido? {CuentaBancaria.validar_saldo(-100)}")

# Usar método de clase para obtener total
print(f"\nTotal de cuentas creadas: {CuentaBancaria.obtener_total_cuentas()}")
````

````python
# math - Funciones matemáticas
import math

print("=== MÓDULO MATH ===")
print(f"Pi: {math.pi}")
print(f"e: {math.e}")
print(f"Raíz cuadrada de 16: {math.sqrt(16)}")
print(f"Potencia 2^3: {math.pow(2, 3)}")
print(f"Valor absoluto de -5: {math.fabs(-5)}")
print(f"Redondear arriba 4.3: {math.ceil(4.3)}")
print(f"Redondear abajo 4.7: {math.floor(4.7)}")
print(f"Logaritmo natural de 10: {math.log(10):.4f}")
print(f"Logaritmo base 10 de 100: {math.log10(100)}")
print(f"Seno de π/2: {math.sin(math.pi/2)}")
print(f"Coseno de 0: {math.cos(0)}")
print(f"Factorial de 5: {math.factorial(5)}")
````

````python
# datetime - Fechas y tiempos
from datetime import datetime, timedelta, date, time

print("=== MÓDULO DATETIME ===")
hoy = datetime.now()
print(f"Fecha y hora actual: {hoy}")
print(f"Año: {hoy.year}, Mes: {hoy.month}, Día: {hoy.day}")
print(f"Hora: {hoy.hour}:{hoy.minute}:{hoy.second}")

# Formatear fecha
print(f"\nFormato personalizado: {hoy.strftime('%Y-%m-%d %H:%M:%S')}")
print(f"Día de la semana: {hoy.strftime('%A')}")
print(f"Mes: {hoy.strftime('%B')}")

# Operaciones con fechas
mañana = hoy + timedelta(days=1)
semana_pasada = hoy - timedelta(weeks=1)
print(f"\nMañana: {mañana.date()}")
print(f"Semana pasada: {semana_pasada.date()}")

# Crear fecha específica
fecha_especifica = datetime(2024, 1, 15, 14, 30, 0)
print(f"Fecha específica: {fecha_especifica}")

# Diferencia entre fechas
diferencia = hoy - fecha_especifica
print(f"\nDiferencia: {diferencia.days} días")
````

````python
# time - Medición de tiempo
import time

print("=== MÓDULO TIME ===")

# Medir tiempo de ejecución
inicio = time.time()
suma = sum(range(1000000))
fin = time.time()
tiempo_transcurrido = fin - inicio

print(f"Tiempo de ejecución: {tiempo_transcurrido:.4f} segundos")
print(f"Resultado: {suma}")

# Tiempo actual
tiempo_actual = time.time()
print(f"\nTiempo Unix: {tiempo_actual}")

# Convertir a formato legible
tiempo_legible = time.ctime(tiempo_actual)
print(f"Tiempo legible: {tiempo_legible}")

# Pausar ejecución
print("\nEsperando 1 segundo...")
time.sleep(1)
print("¡Listo!")
````

````python
# statistics - Estadísticas
import statistics

datos = [85, 90, 78, 92, 88, 85, 87, 91, 89, 86]

print("=== MÓDULO STATISTICS ===")
print(f"Datos: {datos}")
print(f"Promedio: {statistics.mean(datos):.2f}")
print(f"Mediana: {statistics.median(datos):.2f}")
print(f"Moda: {statistics.mode(datos)}")
print(f"Desviación estándar: {statistics.stdev(datos):.2f}")
print(f"Varianza: {statistics.variance(datos):.2f}")
print(f"Mínimo: {min(datos)}")
print(f"Máximo: {max(datos)}")
````

````python
# os y os.path - Sistema operativo
import os
from pathlib import Path

print("=== MÓDULO OS ===")
print(f"Sistema operativo: {os.name}")
print(f"Directorio actual: {os.getcwd()}")
print(f"Separador de ruta: {os.sep}")

# Información de archivos
archivo = "ejemplo.txt"
if os.path.exists(archivo):
    print(f"\nArchivo '{archivo}':")
    print(f"  Existe: {os.path.exists(archivo)}")
    print(f"  Es archivo: {os.path.isfile(archivo)}")
    print(f"  Tamaño: {os.path.getsize(archivo)} bytes")

# pathlib (más moderno)
print("\n=== PATHLIB ===")
ruta = Path("ejemplo.txt")
if ruta.exists():
    print(f"Ruta: {ruta}")
    print(f"Nombre: {ruta.name}")
    print(f"Directorio padre: {ruta.parent}")
    print(f"Extensión: {ruta.suffix}")
````
