### 8.2 Excepciones

```python
# Excepciones comunes
# ZeroDivisionError
try:
    resultado = 10 / 0
except ZeroDivisionError:
    print("Error: División por cero")

# ValueError
try:
    numero = int("abc")
except ValueError:
    print("Error: No se puede convertir a entero")

# KeyError
diccionario = {"a": 1, "b": 2}
try:
    valor = diccionario["c"]
except KeyError:
    print("Error: Clave no encontrada")
```

### 8.3 Gestionando excepciones

```python
# try-except básico
def dividir(a, b):
    try:
        resultado = a / b
        return resultado
    except ZeroDivisionError:
        print("No se puede dividir por cero")
        return None

print(dividir(10, 2))
print(dividir(10, 0))
```

```python
# Múltiples excepciones
def procesar_numero(texto):
    try:
        numero = float(texto)
        resultado = 100 / numero
        return resultado
    except ValueError:
        print("Error: No es un número válido")
        return None
    except ZeroDivisionError:
        print("Error: División por cero")
        return None

procesar_numero("10")
procesar_numero("abc")
procesar_numero("0")
```

```python
# try-except-else-finally
def leer_archivo(nombre):
    try:
        archivo = open(nombre, "r")
        contenido = archivo.read()
    except FileNotFoundError:
        print(f"Error: Archivo {nombre} no encontrado")
        return None
    except Exception as e:
        print(f"Error inesperado: {e}")
        return None
    else:
        print("Archivo leído exitosamente")
        return contenido
    finally:
        print("Operación finalizada")

# Nota: En la práctica, usa 'with' para archivos
```

### 8.4 Lanzando excepciones

```python
# Lanzar excepciones
def validar_edad(edad):
    if edad < 0:
        raise ValueError("La edad no puede ser negativa")
    if edad > 150:
        raise ValueError("La edad no puede ser mayor a 150")
    return True

try:
    validar_edad(-5)
except ValueError as e:
    print(f"Error: {e}")
```

```python
# Excepciones personalizadas
class SaldoInsuficienteError(Exception):
    """Excepción para saldo insuficiente."""
    pass

def retirar(saldo, monto):
    if monto > saldo:
        raise SaldoInsuficienteError(f"Saldo insuficiente. Tienes ${saldo}, intentas retirar ${monto}")
    return saldo - monto

try:
    nuevo_saldo = retirar(100, 150)
except SaldoInsuficienteError as e:
    print(f"Error: {e}")
```

## 9. Clases

### 9.3 Un primer vistazo a las clases

```python
# Clase simple
class CuentaBancaria:
    """Clase que representa una cuenta bancaria."""
    
    def __init__(self, titular, saldo_inicial=0):
        """Inicializa la cuenta."""
        self.titular = titular
        self.saldo = saldo_inicial
    
    def depositar(self, monto):
        """Deposita dinero en la cuenta."""
        if monto > 0:
            self.saldo += monto
            return True
        return False
    
    def retirar(self, monto):
        """Retira dinero de la cuenta."""
        if monto > 0 and monto <= self.saldo:
            self.saldo -= monto
            return True
        return False
    
    def consultar_saldo(self):
        """Consulta el saldo actual."""
        return self.saldo
    
    def __str__(self):
        """Representación en string."""
        return f"Cuenta de {self.titular}: ${self.saldo:,.2f}"

# Usar la clase
cuenta = CuentaBancaria("Juan Pérez", 1000)
print(cuenta)
cuenta.depositar(500)
print(f"Después de depositar: ${cuenta.consultar_saldo():,.2f}")
cuenta.retirar(200)
print(f"Después de retirar: ${cuenta.consultar_saldo():,.2f}")
```

### 9.5 Herencia

```python
# Herencia
class CuentaAhorros(CuentaBancaria):
    """Cuenta de ahorros con interés."""
    
    def __init__(self, titular, saldo_inicial=0, tasa_interes=0.02):
        super().__init__(titular, saldo_inicial)
        self.tasa_interes = tasa_interes
    
    def aplicar_interes(self):
        """Aplica interés al saldo."""
        interes = self.saldo * self.tasa_interes
        self.saldo += interes
        return interes

# Usar clase heredada
cuenta_ahorros = CuentaAhorros("María González", 1000, 0.03)
print(cuenta_ahorros)
interes = cuenta_ahorros.aplicar_interes()
print(f"Interés aplicado: ${interes:,.2f}")
print(f"Nuevo saldo: ${cuenta_ahorros.consultar_saldo():,.2f}")
```

### 10.8 Fechas y tiempos

```python
from datetime import datetime, timedelta

hoy = datetime.now()
print(f"Hoy: {hoy}")
print(f"Año: {hoy.year}, Mes: {hoy.month}, Día: {hoy.day}")

mañana = hoy + timedelta(days=1)
print(f"Mañana: {mañana.date()}")

# Formatear fecha
print(f"Formato: {hoy.strftime('%Y-%m-%d %H:%M:%S')}")
```

### 10.10 Medición de rendimiento

```python
import time

# Medir tiempo de ejecución
inicio = time.time()

# Simular operación
suma = sum(range(1000000))

fin = time.time()
tiempo_transcurrido = fin - inicio

print(f"Tiempo transcurrido: {tiempo_transcurrido:.4f} segundos")
print(f"Resultado: {suma}")
```
