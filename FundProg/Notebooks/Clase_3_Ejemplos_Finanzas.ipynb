## Aplicando estructuras de datos a problemas financieros

En este notebook veremos cómo usar listas, diccionarios, tuplas y conjuntos para manejar datos financieros complejos.

## 1. Portafolio de Inversiones con Diccionarios

### 1.1 Gestión de Portafolio

```python
# Portafolio usando diccionarios
portafolio = {
    "AAPL": {"nombre": "Apple Inc.", "cantidad": 100, "precio_compra": 150.25},
    "GOOGL": {"nombre": "Alphabet Inc.", "cantidad": 50, "precio_compra": 2500.00},
    "MSFT": {"nombre": "Microsoft Corp.", "cantidad": 75, "precio_compra": 300.50},
    "TSLA": {"nombre": "Tesla Inc.", "cantidad": 200, "precio_compra": 800.00}
}

# Precios actuales
precios_actuales = {"AAPL": 175.50, "GOOGL": 2800.00, "MSFT": 350.75, "TSLA": 950.00}

# Calcular valor del portafolio
print("=== ANÁLISIS DE PORTAFOLIO ===")
valor_total_compra = 0
valor_total_actual = 0

for simbolo, info in portafolio.items():
    precio_compra = info["precio_compra"]
    cantidad = info["cantidad"]
    precio_actual = precios_actuales[simbolo]
    
    valor_compra = precio_compra * cantidad
    valor_actual = precio_actual * cantidad
    ganancia = valor_actual - valor_compra
    roi = (ganancia / valor_compra) * 100
    
    valor_total_compra += valor_compra
    valor_total_actual += valor_actual
    
    print(f"\n{simbolo} - {info['nombre']}:")
    print(f"  Cantidad: {cantidad}")
    print(f"  Precio compra: ${precio_compra:.2f}")
    print(f"  Precio actual: ${precio_actual:.2f}")
    print(f"  Valor compra: ${valor_compra:,.2f}")
    print(f"  Valor actual: ${valor_actual:,.2f}")
    print(f"  Ganancia/Pérdida: ${ganancia:,.2f} ({roi:+.2f}%)")

ganancia_total = valor_total_actual - valor_total_compra
roi_total = (ganancia_total / valor_total_compra) * 100

print(f"\n{'='*50}")
print(f"Valor total compra: ${valor_total_compra:,.2f}")
print(f"Valor total actual: ${valor_total_actual:,.2f}")
print(f"Ganancia total: ${ganancia_total:,.2f} ({roi_total:+.2f}%)")
```

## 2. Análisis de Transacciones con Listas y Tuplas

### 2.1 Registro de Transacciones

```python
# Transacciones como tuplas: (fecha, tipo, monto, descripcion)
transacciones = [
    ("2024-01-15", "Ingreso", 5000, "Salario"),
    ("2024-01-18", "Egreso", -1200, "Alquiler"),
    ("2024-01-20", "Egreso", -350, "Alimentación"),
    ("2024-01-22", "Ingreso", 500, "Freelance"),
    ("2024-01-25", "Egreso", -200, "Transporte"),
    ("2024-01-28", "Egreso", -150, "Entretenimiento")
]

# Análisis usando comprensión de listas
ingresos = [t[2] for t in transacciones if t[1] == "Ingreso"]
egresos = [abs(t[2]) for t in transacciones if t[1] == "Egreso"]

total_ingresos = sum(ingresos)
total_egresos = sum(egresos)
saldo = total_ingresos - total_egresos

print("=== ANÁLISIS DE TRANSACCIONES ===")
print(f"Total ingresos: ${total_ingresos:,.2f}")
print(f"Total egresos: ${total_egresos:,.2f}")
print(f"Saldo: ${saldo:,.2f}")
print(f"\nNúmero de transacciones: {len(transacciones)}")
print(f"Promedio de ingresos: ${total_ingresos/len(ingresos):,.2f}")
print(f"Promedio de egresos: ${total_egresos/len(egresos):,.2f}")
```

## 3. Análisis de Precios con Comprensión de Listas

### 3.1 Cálculo de Rendimientos

```python
# Precios históricos de una acción
precios = [100, 102, 98, 105, 103, 107, 110, 108, 112, 115]

# Calcular rendimientos diarios usando comprensión de listas
rendimientos = [(precios[i] - precios[i-1]) / precios[i-1] * 100 
                for i in range(1, len(precios))]

print("=== ANÁLISIS DE RENDIMIENTOS ===")
print(f"Precios: {precios}")
print(f"\nRendimientos diarios (%):")
for i, rend in enumerate(rendimientos, 1):
    signo = "+" if rend >= 0 else ""
    print(f"  Día {i}: {signo}{rend:.2f}%")

# Estadísticas
rendimiento_promedio = sum(rendimientos) / len(rendimientos)
rendimiento_max = max(rendimientos)
rendimiento_min = min(rendimientos)

print(f"\nRendimiento promedio: {rendimiento_promedio:.2f}%")
print(f"Rendimiento máximo: {rendimiento_max:.2f}%")
print(f"Rendimiento mínimo: {rendimiento_min:.2f}%")
```

## 4. Categorización de Gastos con Diccionarios

### 4.1 Agrupar Gastos por Categoría

```python
# Gastos con categorías
gastos = [
    ("Alimentación", 350),
    ("Transporte", 120),
    ("Alimentación", 280),
    ("Servicios", 180),
    ("Entretenimiento", 150),
    ("Transporte", 90),
    ("Alimentación", 320),
    ("Servicios", 200),
    ("Entretenimiento", 100)
]

# Agrupar por categoría usando diccionario
gastos_por_categoria = {}
for categoria, monto in gastos:
    if categoria in gastos_por_categoria:
        gastos_por_categoria[categoria].append(monto)
    else:
        gastos_por_categoria[categoria] = [monto]

# Calcular totales por categoría
print("=== GASTOS POR CATEGORÍA ===")
for categoria, montos in gastos_por_categoria.items():
    total = sum(montos)
    promedio = total / len(montos)
    print(f"\n{categoria}:")
    print(f"  Transacciones: {len(montos)}")
    print(f"  Total: ${total:,.2f}")
    print(f"  Promedio: ${promedio:,.2f}")

gasto_total = sum(sum(montos) for montos in gastos_por_categoria.values())
print(f"\n{'='*40}")
print(f"Gasto total: ${gasto_total:,.2f}")
```

## 5. Comparación de Inversiones con Conjuntos

### 5.1 Encontrar Inversiones Comunes

```python
# Portafolios de diferentes inversores
portafolio_maria = {"AAPL", "GOOGL", "MSFT", "AMZN", "TSLA"}
portafolio_juan = {"AAPL", "MSFT", "NVDA", "META", "TSLA"}
portafolio_carlos = {"GOOGL", "AMZN", "NVDA", "META", "NFLX"}

# Análisis usando operaciones de conjuntos
print("=== COMPARACIÓN DE PORTAFOLIOS ===")
print(f"Portafolio María: {portafolio_maria}")
print(f"Portafolio Juan: {portafolio_juan}")
print(f"Portafolio Carlos: {portafolio_carlos}")

# Acciones comunes entre María y Juan
comunes_maria_juan = portafolio_maria & portafolio_juan
print(f"\nAcciones comunes (María y Juan): {comunes_maria_juan}")

# Acciones únicas de María
unicas_maria = portafolio_maria - portafolio_juan - portafolio_carlos
print(f"Acciones únicas de María: {unicas_maria}")

# Todas las acciones únicas
todas_acciones = portafolio_maria | portafolio_juan | portafolio_carlos
print(f"Todas las acciones: {todas_acciones}")
print(f"Total de acciones diferentes: {len(todas_acciones)}")
```

## 6. Análisis de Flujo de Caja con Zip y Enumerate

### 6.1 Proyección Mensual

```python
# Datos mensuales
meses = ["Enero", "Febrero", "Marzo", "Abril", "Mayo", "Junio"]
ingresos = [5000, 5200, 4800, 5500, 5100, 5300]
egresos = [3500, 3800, 4000, 3600, 3700, 3900]

# Calcular flujo neto usando zip
flujos_netos = [ing - eg for ing, eg in zip(ingresos, egresos)]

# Calcular flujo acumulado
flujo_acumulado = []
acumulado = 0
for flujo in flujos_netos:
    acumulado += flujo
    flujo_acumulado.append(acumulado)

print("=== FLUJO DE CAJA MENSUAL ===")
print(f"{'Mes':<12} {'Ingresos':<12} {'Egresos':<12} {'Flujo Neto':<15} {'Acumulado':<15}")
print("="*70)

for mes, ing, eg, neto, acum in zip(meses, ingresos, egresos, flujos_netos, flujo_acumulado):
    print(f"{mes:<12} ${ing:<11,.2f} ${eg:<11,.2f} ${neto:<14,.2f} ${acum:<14,.2f}")

print("="*70)
print(f"Promedio flujo neto: ${sum(flujos_netos)/len(flujos_netos):,.2f}")
print(f"Flujo acumulado final: ${flujo_acumulado[-1]:,.2f}")
```

## 7. Ranking de Inversiones con Sorted

### 7.1 Ordenar por Rendimiento

```python
# Inversiones con rendimientos
inversiones = {
    "Fondo A": 12.5,
    "Fondo B": 8.3,
    "Fondo C": 15.7,
    "Fondo D": 6.2,
    "Fondo E": 18.1
}

# Ordenar por rendimiento (descendente)
ranking = sorted(inversiones.items(), key=lambda x: x[1], reverse=True)

print("=== RANKING DE INVERSIONES ===")
for posicion, (fondo, rendimiento) in enumerate(ranking, 1):
    print(f"{posicion}. {fondo}: {rendimiento:.2f}%")

# Encontrar mejor y peor
mejor = max(inversiones.items(), key=lambda x: x[1])
peor = min(inversiones.items(), key=lambda x: x[1])

print(f"\nMejor rendimiento: {mejor[0]} ({mejor[1]:.2f}%)")
print(f"Peor rendimiento: {peor[0]} ({peor[1]:.2f}%)")
```
