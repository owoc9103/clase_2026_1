## 1. Múltiples bases de datos

En la práctica trabajamos con **varias tablas** (varios DataFrames). Para combinarlas usamos **concat**, **merge** y **join**.

### 1.1 Importar pandas

```
import pandas as pd
import numpy as np
```

### 1.2 pd.concat() — Concatenar DataFrames

**pd.concat()** apila varios DataFrames (filas o columnas). Útil cuando las tablas tienen la **misma estructura**.

```
df1 = pd.DataFrame({"A": [1, 2], "B": [3, 4]})
df2 = pd.DataFrame({"A": [5, 6], "B": [7, 8]})
pd.concat([df1, df2], ignore_index=True)
```

### 1.3 pd.merge() — Unir tablas (JOIN)

**pd.merge()** une dos DataFrames por **columnas clave**. Equivale a JOIN en SQL (inner, left, right, outer).

```
ventas = pd.DataFrame({"id": [1, 2, 3], "producto": ["A", "B", "C"], "monto": [100, 200, 150]})
clientes = pd.DataFrame({"id": [1, 2, 4], "nombre": ["Ana", "Luis", "María"]})
pd.merge(ventas, clientes, on="id", how="left")
```

### 1.4 .join() — Unir por índice

El método **.join()** une por **índice**.

```
izq = pd.DataFrame({"X": [10, 20]}, index=["a", "b"])
der = pd.DataFrame({"Y": [30, 40]}, index=["a", "b"])
izq.join(der)
```
