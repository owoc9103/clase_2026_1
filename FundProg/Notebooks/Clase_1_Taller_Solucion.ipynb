## Ejercicio 1: Calculadora de Interés Simple

Crea un programa que calcule el interés simple y el monto final para una inversión.

**Datos de entrada:**
- Capital inicial: $15,000
- Tasa de interés: 6.5% anual
- Tiempo: 4 años

**Salida esperada:**
- Interés ganado
- Monto final
- Reporte formateado

```python
# Solución Ejercicio 1: Interés simple
capital = 15000
tasa = 0.065
tiempo = 4

interes = capital * tasa * tiempo
monto_final = capital + interes

print("=== REPORTE DE INVERSIÓN ===")
print(f"Capital inicial: ${capital:,.2f}")
print(f"Tasa de interés: {tasa * 100}% anual")
print(f"Tiempo: {tiempo} años")
print(f"Interés ganado: ${interes:,.2f}")
print(f"Monto final: ${monto_final:,.2f}")

```

## Ejercicio 2: Análisis de Portafolio de Inversión

Tienes un portafolio con las siguientes inversiones:
- Acción A: $5,000
- Acción B: $8,000
- Acción C: $3,000
- Acción D: $4,000

**Tareas:**
1. Crea una lista con los valores de las inversiones
2. Calcula el valor total del portafolio
3. Calcula el porcentaje que representa cada inversión
4. Muestra un reporte formateado

```python
# Solución Ejercicio 2: Análisis de portafolio
inversiones = [5000, 8000, 3000, 4000]
nombres = ["Acción A", "Acción B", "Acción C", "Acción D"]

total = sum(inversiones)

print("=== ANÁLISIS DE PORTAFOLIO ===")
for nombre, inversion in zip(nombres, inversiones):
    porcentaje = (inversion / total) * 100
    print(f"{nombre}: ${inversion:,.2f} ({porcentaje:.2f}%)")
print(f"\nTotal del portafolio: ${total:,.2f}")

```

## Ejercicio 3: Registro de Gastos Mensuales

Crea un programa que registre los gastos mensuales y calcule estadísticas.

**Gastos del mes:**
- Alimentación: $350
- Transporte: $120
- Servicios: $180
- Entretenimiento: $150
- Otros: $200

**Tareas:**
1. Almacena los gastos en una lista
2. Calcula el total de gastos
3. Encuentra el gasto máximo y mínimo
4. Calcula el promedio de gastos
5. Crea un reporte con toda la información

```python
# Solución Ejercicio 3: Gastos mensuales
gastos = [350, 120, 180, 150, 200]
categorias = ["Alimentación", "Transporte", "Servicios", "Entretenimiento", "Otros"]

total = sum(gastos)
maximo = max(gastos)
minimo = min(gastos)
promedio = total / len(gastos)

print("=== REGISTRO DE GASTOS MENSUALES ===")
for cat, g in zip(categorias, gastos):
    print(f"  {cat}: ${g:,.2f}")
print(f"\nTotal de gastos: ${total:,.2f}")
print(f"Gasto máximo: ${maximo:,.2f} ({categorias[gastos.index(maximo)]})")
print(f"Gasto mínimo: ${minimo:,.2f} ({categorias[gastos.index(minimo)]})")
print(f"Promedio: ${promedio:,.2f}")

```

## Ejercicio 4: Calculadora de Valor Presente

Eres un analista financiero y necesitas calcular el valor presente de varios flujos de efectivo futuros.

**Datos:**
- Flujo 1: $1,000 en 1 año
- Flujo 2: $2,000 en 2 años
- Flujo 3: $1,500 en 3 años
- Tasa de descuento: 7% anual

**Fórmula:** VP = VF / (1 + r)^t

**Tareas:**
1. Calcula el valor presente de cada flujo
2. Calcula el valor presente total
3. Muestra los resultados en una tabla formateada

```python
# Solución Ejercicio 4: Valor presente
flujos = [1000, 2000, 1500]
anios = [1, 2, 3]
r = 0.07

vps = []
for vf, t in zip(flujos, anios):
    vp = vf / (1 + r) ** t
    vps.append(vp)

vp_total = sum(vps)

print("=== VALOR PRESENTE DE FLUJOS ===")
print(f"{'Flujo':<10} {'VF':>12} {'Años':>6} {'VP':>12}")
print("-" * 42)
for i, (vf, t) in enumerate(zip(flujos, anios), 1):
    print(f"Flujo {i:<4} ${vf:>10,.2f} {t:>6} ${vps[i-1]:>10,.2f}")
print("-" * 42)
print(f"{'TOTAL VP':<10} {'':>12} {'':>6} ${vp_total:>10,.2f}")

```

## Ejercicio 5: Análisis de Precios de Criptomonedas

Tienes los precios de cierre de Bitcoin durante una semana:

**Precios (USD):**
- Lunes: $42,500
- Martes: $43,200
- Miércoles: $41,800
- Jueves: $44,100
- Viernes: $43,900
- Sábado: $45,200
- Domingo: $44,500

**Tareas:**
1. Crea una lista con los precios
2. Calcula la variación diaria (en $ y %)
3. Encuentra el día con mayor ganancia y mayor pérdida
4. Calcula el precio promedio de la semana
5. Crea un reporte completo

```python
# Solución Ejercicio 5: Precios de Bitcoin
dias = ["Lunes", "Martes", "Miércoles", "Jueves", "Viernes", "Sábado", "Domingo"]
precios = [42500, 43200, 41800, 44100, 43900, 45200, 44500]

variacion_dolares = []
variacion_pct = []
for i in range(1, len(precios)):
    var_d = precios[i] - precios[i - 1]
    var_pct = (var_d / precios[i - 1]) * 100
    variacion_dolares.append(var_d)
    variacion_pct.append(var_pct)

max_ganancia_idx = variacion_dolares.index(max(variacion_dolares)) + 1
max_perdida_idx = variacion_dolares.index(min(variacion_dolares)) + 1
promedio_semana = sum(precios) / len(precios)

print("=== ANÁLISIS BITCOIN (USD) ===")
print(f"\nPrecio por día: {precios}")
print(f"\nVariación diaria ($): {[round(v, 2) for v in variacion_dolares]}")
print(f"Variación diaria (%): {[round(v, 2) for v in variacion_pct]} %")
print(f"\nMayor ganancia: {dias[max_ganancia_idx]} (${variacion_dolares[max_ganancia_idx-1]:,.2f})")
print(f"Mayor pérdida: {dias[max_perdida_idx]} (${variacion_dolares[max_perdida_idx-1]:,.2f})")
print(f"Precio promedio de la semana: ${promedio_semana:,.2f}")

```
