## 1. Interés Compuesto con Condicionales

### 1.1 Calculadora de Interés Compuesto

```python
# Función para calcular interés compuesto
def calcular_interes_compuesto(principal, tasa, tiempo, capitalizacion="anual"):
    """
    Calcula el interés compuesto.
    
    capitalizacion: 'anual', 'mensual', 'diaria'
    """
    if capitalizacion == "anual":
        n = 1
    elif capitalizacion == "mensual":
        n = 12
    elif capitalizacion == "diaria":
        n = 365
    else:
        n = 1
    
    monto = principal * (1 + tasa/n) ** (n * tiempo)
    interes = monto - principal
    
    return monto, interes

# Ejemplo
principal = 10000
tasa = 0.06  # 6% anual
tiempo = 5  # años

monto_anual, interes_anual = calcular_interes_compuesto(principal, tasa, tiempo, "anual")
monto_mensual, interes_mensual = calcular_interes_compuesto(principal, tasa, tiempo, "mensual")

print(f"Capital inicial: ${principal:,.2f}")
print(f"\nCapitalización Anual:")
print(f"  Monto final: ${monto_anual:,.2f}")
print(f"  Interés ganado: ${interes_anual:,.2f}")
print(f"\nCapitalización Mensual:")
print(f"  Monto final: ${monto_mensual:,.2f}")
print(f"  Interés ganado: ${interes_mensual:,.2f}")
```

## 2. Tabla de Amortización con Bucles

### 2.1 Generar Tabla de Amortización

```python
def tabla_amortizacion(principal, tasa_anual, num_pagos):
    """
    Genera una tabla de amortización para un préstamo.
    """
    tasa_mensual = tasa_anual / 12
    
    # Calcular pago mensual usando fórmula de anualidad
    if tasa_mensual > 0:
        pago_mensual = principal * (tasa_mensual * (1 + tasa_mensual)**num_pagos) / \
                      ((1 + tasa_mensual)**num_pagos - 1)
    else:
        pago_mensual = principal / num_pagos
    
    saldo = principal
    
    print(f"Préstamo: ${principal:,.2f}")
    print(f"Tasa anual: {tasa_anual*100:.2f}%")
    print(f"Número de pagos: {num_pagos}")
    print(f"Pago mensual: ${pago_mensual:,.2f}")
    print("\n" + "="*70)
    print(f"{'Mes':<6} {'Pago':<12} {'Interés':<12} {'Principal':<12} {'Saldo':<12}")
    print("="*70)
    
    total_interes = 0
    
    for mes in range(1, num_pagos + 1):
        interes_pago = saldo * tasa_mensual
        principal_pago = pago_mensual - interes_pago
        saldo -= principal_pago
        total_interes += interes_pago
        
        if saldo < 0.01:  # Ajuste final
            saldo = 0
        
        print(f"{mes:<6} ${pago_mensual:<11,.2f} ${interes_pago:<11,.2f} "
              f"${principal_pago:<11,.2f} ${saldo:<11,.2f}")
    
    print("="*70)
    print(f"Total pagado: ${pago_mensual * num_pagos:,.2f}")
    print(f"Total interés: ${total_interes:,.2f}")

# Ejemplo
tabla_amortizacion(principal=100000, tasa_anual=0.05, num_pagos=12)
```

## 3. Análisis de Rentabilidad con Condicionales

### 3.1 Clasificar Inversiones por Rentabilidad

```python
def clasificar_inversion(roi):
    """
    Clasifica una inversión según su ROI.
    """
    if roi >= 20:
        categoria = "Excelente"
        recomendacion = "Mantener y considerar aumentar"
    elif roi >= 10:
        categoria = "Buena"
        recomendacion = "Mantener"
    elif roi >= 5:
        categoria = "Aceptable"
        recomendacion = "Revisar estrategia"
    elif roi >= 0:
        categoria = "Baja"
        recomendacion = "Considerar alternativas"
    else:
        categoria = "Pérdida"
        recomendacion = "Revisar urgentemente o salir"
    
    return categoria, recomendacion

# Analizar múltiples inversiones
inversiones = [
    ("Acción A", 25.5),
    ("Acción B", 12.3),
    ("Acción C", 6.8),
    ("Acción D", 2.1),
    ("Acción E", -5.2)
]

print("=== ANÁLISIS DE RENTABILIDAD ===")
for nombre, roi in inversiones:
    categoria, recomendacion = clasificar_inversion(roi)
    signo = "+" if roi >= 0 else ""
    print(f"\n{nombre}:")
    print(f"  ROI: {signo}{roi:.2f}%")
    print(f"  Categoría: {categoria}")
    print(f"  Recomendación: {recomendacion}")
```

## 4. Proyección de Crecimiento con Bucles

### 4.1 Proyectar Valor Futuro de Inversión

```python
def proyectar_inversion(principal, tasa_anual, años, aporte_mensual=0):
    """
    Proyecta el crecimiento de una inversión con aportes mensuales opcionales.
    """
    tasa_mensual = tasa_anual / 12
    saldo = principal
    
    print(f"Proyección de inversión:")
    print(f"Capital inicial: ${principal:,.2f}")
    print(f"Tasa anual: {tasa_anual*100:.2f}%")
    print(f"Aporte mensual: ${aporte_mensual:,.2f}")
    print(f"Periodo: {años} años\n")
    print(f"{'Año':<6} {'Saldo Inicial':<15} {'Aportes':<12} {'Interés':<12} {'Saldo Final':<15}")
    print("="*70)
    
    for año in range(1, años + 1):
        saldo_inicial_año = saldo
        total_aportes = 0
        total_interes = 0
        
        # Calcular mes a mes
        for mes in range(12):
            saldo += aporte_mensual
            total_aportes += aporte_mensual
            interes_mes = saldo * tasa_mensual
            saldo += interes_mes
            total_interes += interes_mes
        
        print(f"{año:<6} ${saldo_inicial_año:<14,.2f} ${total_aportes:<11,.2f} "
              f"${total_interes:<11,.2f} ${saldo:<14,.2f}")
    
    print("="*70)
    print(f"Saldo final: ${saldo:,.2f}")
    print(f"Total aportado: ${principal + aporte_mensual * 12 * años:,.2f}")
    print(f"Ganancia total: ${saldo - principal - aporte_mensual * 12 * años:,.2f}")
    
    return saldo

# Ejemplo
proyectar_inversion(principal=5000, tasa_anual=0.07, años=5, aporte_mensual=200)
```

## 5. Análisis de Riesgo con Funciones

### 5.1 Calcular Métricas de Riesgo

```python
def calcular_metricas_riesgo(rendimientos):
    """
    Calcula métricas de riesgo de una serie de rendimientos.
    """
    if len(rendimientos) == 0:
        return None
    
    # Rendimiento promedio
    rendimiento_promedio = sum(rendimientos) / len(rendimientos)
    
    # Varianza
    varianza = sum((r - rendimiento_promedio)**2 for r in rendimientos) / len(rendimientos)
    
    # Desviación estándar (volatilidad)
    volatilidad = varianza ** 0.5
    
    # Rendimiento mínimo y máximo
    rend_min = min(rendimientos)
    rend_max = max(rendimientos)
    
    # Clasificar riesgo
    if volatilidad < 0.05:
        nivel_riesgo = "Bajo"
    elif volatilidad < 0.15:
        nivel_riesgo = "Moderado"
    elif volatilidad < 0.30:
        nivel_riesgo = "Alto"
    else:
        nivel_riesgo = "Muy Alto"
    
    return {
        "rendimiento_promedio": rendimiento_promedio,
        "volatilidad": volatilidad,
        "rendimiento_min": rend_min,
        "rendimiento_max": rend_max,
        "nivel_riesgo": nivel_riesgo
    }

# Ejemplo: Rendimientos mensuales de una acción (%)
rendimientos_mensuales = [2.5, -1.2, 3.8, 0.5, -2.1, 4.2, 1.8, -0.9, 2.3, 1.5, -1.8, 3.1]

metricas = calcular_metricas_riesgo(rendimientos_mensuales)

if metricas:
    print("=== MÉTRICAS DE RIESGO ===")
    print(f"Rendimiento promedio: {metricas['rendimiento_promedio']:.2f}%")
    print(f"Volatilidad: {metricas['volatilidad']:.2f}%")
    print(f"Rendimiento mínimo: {metricas['rendimiento_min']:.2f}%")
    print(f"Rendimiento máximo: {metricas['rendimiento_max']:.2f}%")
    print(f"Nivel de riesgo: {metricas['nivel_riesgo']}")
```

## 6. Simulador de Ahorro con Funciones Avanzadas

### 6.1 Plan de Ahorro con Objetivo

```python
def simular_ahorro(objetivo, tasa_anual, aporte_mensual, ahorro_inicial=0):
    """
    Simula un plan de ahorro hasta alcanzar un objetivo.
    """
    tasa_mensual = tasa_anual / 12
    saldo = ahorro_inicial
    meses = 0
    
    print(f"Objetivo: ${objetivo:,.2f}")
    print(f"Ahorro inicial: ${ahorro_inicial:,.2f}")
    print(f"Aporte mensual: ${aporte_mensual:,.2f}")
    print(f"Tasa anual: {tasa_anual*100:.2f}%\n")
    
    while saldo < objetivo and meses < 600:  # Límite de 50 años
        meses += 1
        saldo += aporte_mensual
        interes = saldo * tasa_mensual
        saldo += interes
        
        if meses % 12 == 0:  # Mostrar cada año
            años = meses // 12
            print(f"Año {años}: ${saldo:,.2f}")
    
    años = meses / 12
    
    print(f"\n{'='*50}")
    print(f"Tiempo necesario: {años:.1f} años ({meses} meses)")
    print(f"Saldo final: ${saldo:,.2f}")
    print(f"Total aportado: ${ahorro_inicial + aporte_mensual * meses:,.2f}")
    print(f"Intereses ganados: ${saldo - ahorro_inicial - aporte_mensual * meses:,.2f}")
    
    return meses, saldo

# Ejemplo: Ahorrar para compra de casa
simular_ahorro(objetivo=50000, tasa_anual=0.05, aporte_mensual=800, ahorro_inicial=5000)
```
