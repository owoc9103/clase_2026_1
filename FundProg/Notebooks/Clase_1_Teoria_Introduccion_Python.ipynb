## 1. ¿Qué es Python?

Python es un lenguaje de programación potente y fácil de aprender. Tiene:

- **Estructuras de datos de alto nivel** eficientes
- **Sistema de programación orientado a objetos** simple pero efectivo
- **Sintaxis elegante** y tipado dinámico
- **Naturaleza interpretada** - ideal para scripting y desarrollo rápido

Python es apropiado para muchas áreas y plataformas.

## 3. Una Introducción Informal a Python

### 3.1 Usando Python como una calculadora

#### 3.1.1 Números

```python
# Operaciones aritméticas básicas
print("Suma:", 2 + 3)
print("Resta:", 5 - 2)
print("Multiplicación:", 4 * 7)
print("División:", 15 / 3)
print("División entera:", 15 // 3)
print("Módulo (resto):", 17 % 5)
print("Potencia:", 2 ** 8)
```

```python
# Variables y asignación
ancho = 20
alto = 5 * 9
area = ancho * alto
print(f"Área: {area}")
```

```python
# Tipos de números: int y float
entero = 5
decimal = 3.14
print(f"Tipo de {entero}: {type(entero)}")
print(f"Tipo de {decimal}: {type(decimal)}")
print(f"Suma: {entero + decimal}")
```

#### 3.1.2 Texto (Strings)

```python
# Crear strings
texto1 = 'Hola'
texto2 = "Mundo"
texto3 = '''Este es un
texto de múltiples
líneas'''

print(texto1)
print(texto2)
print(texto3)
```

```python
# Operaciones con strings
nombre = "Python"
print("Concatenación:", nombre + " es genial")
print("Repetición:", nombre * 3)
print("Longitud:", len(nombre))
print("Mayúsculas:", nombre.upper())
print("Minúsculas:", nombre.lower())
```

```python
# Indexación y slicing
texto = "Python"
print(f"Primer carácter: {texto[0]}")
print(f"Último carácter: {texto[-1]}")
print(f"Primeros 3: {texto[0:3]}")
print(f"Desde el índice 2: {texto[2:]}")
```

#### 3.1.3 Listas

```python
# Crear listas
cuadrados = [1, 4, 9, 16, 25]
print("Lista de cuadrados:", cuadrados)
print("Primer elemento:", cuadrados[0])
print("Último elemento:", cuadrados[-1])
print("Slicing:", cuadrados[1:3])
```

```python
# Operaciones con listas
lista1 = [1, 2, 3]
lista2 = [4, 5, 6]

print("Concatenación:", lista1 + lista2)
print("Repetición:", lista1 * 2)
print("Longitud:", len(lista1))

# Agregar elementos
lista1.append(4)
print("Después de append:", lista1)

# Insertar elemento
lista1.insert(0, 0)
print("Después de insert:", lista1)
```

```python
# Listas anidadas
matriz = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
print("Matriz:", matriz)
print("Primera fila:", matriz[0])
print("Elemento [1][2]:", matriz[1][2])
```

## 3.2 Primeros pasos hacia la programación

### Secuencia de Fibonacci

```python
# Ejemplo: Secuencia de Fibonacci
a, b = 0, 1
while a < 10:
    print(a, end=' ')
    a, b = b, a + b
print()  # Nueva línea
```
