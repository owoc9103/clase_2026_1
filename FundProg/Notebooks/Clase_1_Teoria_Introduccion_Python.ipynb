## 1. ¿Qué es Python?

### 1.1 Definición y Origen

**Python** es un lenguaje de programación de alto nivel, interpretado y de propósito general creado por Guido van Rossum y lanzado por primera vez en 1991. El nombre proviene del programa de comedia británico "Monty Python's Flying Circus", reflejando el sentido del humor de su creador.

### 1.2 Características Fundamentales

Python se distingue por varias características clave:

#### ✨ **Sintaxis Simple y Legible**
- **Indentación obligatoria**: Python usa espacios o tabs para definir bloques de código, lo que fuerza un estilo consistente y legible
- **Menos símbolos**: No requiere llaves `{}` o punto y coma `;` como otros lenguajes
- **Palabras clave en inglés**: `if`, `for`, `while`, `def` son intuitivas

#### 📊 **Tipado Dinámico**
- No necesitas declarar el tipo de variable explícitamente
- Python infiere el tipo automáticamente
- Las variables pueden cambiar de tipo durante la ejecución

#### 🎯 **Interpretado vs Compilado**
- **Interpretado**: El código se ejecuta línea por línea por un intérprete
- **Ventajas**: Desarrollo rápido, fácil depuración, multiplataforma
- **Desventajas**: Generalmente más lento que lenguajes compilados (aunque esto se mitiga con optimizaciones)

#### ⚡ **Multiplataforma**
- Funciona en Windows, macOS, Linux
- El mismo código funciona en diferentes sistemas operativos
- Portabilidad garantizada por la máquina virtual de Python (PVM)

#### 🔧 **Extensible**
- Puede integrarse con código C/C++ para mejor rendimiento
- Interfaz con otros lenguajes y sistemas

#### 📚 **Biblioteca Estándar Rica**
- Incluye módulos para casi todo: archivos, redes, matemáticas, fechas, etc.
- "Batteries included" - viene con todo lo necesario

## 3. Números en Python 🔢

### 3.1 Tipos de Números

Python soporta varios tipos numéricos:

#### **int (Enteros)**
- Números enteros positivos, negativos o cero
- En Python 3, no hay límite práctico de tamaño (solo limitado por memoria)
- Ejemplos: `42`, `-17`, `0`, `1000000`

#### **float (Punto Flotante)**
- Números decimales
- Precisión limitada (aproximadamente 15-17 dígitos decimales)
- Ejemplos: `3.14`, `-0.5`, `2.0`, `1e10` (notación científica)

#### **complex (Complejos)**
- Números complejos con parte real e imaginaria
- Formato: `a + bj` donde `j` es la unidad imaginaria
- Ejemplos: `3 + 4j`, `-2j`, `1.5 + 0.5j`

```python
# Enteros (int)
entero_positivo = 42
entero_negativo = -17
cero = 0
entero_grande = 10**100  # 10 elevado a 100

print(f"Entero positivo: {entero_positivo}, tipo: {type(entero_positivo)}")
print(f"Entero negativo: {entero_negativo}, tipo: {type(entero_negativo)}")
print(f"Entero grande: {entero_grande}")
```

### 3.2 Operaciones Aritméticas

Python soporta todas las operaciones aritméticas estándar:

| Operador | Operación | Ejemplo | Resultado |
|----------|-----------|---------|-----------|
| `+` | Suma | `5 + 3` | `8` |
| `-` | Resta | `5 - 3` | `2` |
| `*` | Multiplicación | `5 * 3` | `15` |
| `/` | División (siempre float) | `5 / 2` | `2.5` |
| `//` | División entera | `5 // 2` | `2` |
| `%` | Módulo (resto) | `5 % 2` | `1` |
| `**` | Potencia | `5 ** 2` | `25` |

#### Precedencia de Operadores
1. Paréntesis `()`
2. Potencia `**`
3. Multiplicación, División, Módulo `*`, `/`, `//`, `%`
4. Suma y Resta `+`, `-`

```python
# Números complejos
complejo1 = 3 + 4j
complejo2 = -2j

print(f"Complejo: {complejo1}, tipo: {type(complejo1)}")
print(f"Parte real: {complejo1.real}, Parte imaginaria: {complejo1.imag}")
```

```python
# Operaciones aritméticas básicas
a, b = 10, 3

print(f"a = {a}, b = {b}")
print(f"Suma: {a} + {b} = {a + b}")
print(f"Resta: {a} - {b} = {a - b}")
print(f"Multiplicación: {a} * {b} = {a * b}")
print(f"División: {a} / {b} = {a / b}")
print(f"División entera: {a} // {b} = {a // b}")
print(f"Módulo (resto): {a} % {b} = {a % b}")
print(f"Potencia: {a} ** {b} = {a ** b}")

# Precedencia de operadores
print(f"\nPrecedencia: 2 + 3 * 4 = {2 + 3 * 4}")  # 14, no 20
print(f"Con paréntesis: (2 + 3) * 4 = {(2 + 3) * 4}")  # 20
```

### 3.4 Variables y Asignación

#### Conceptos Clave:

- **Variable**: Un nombre que referencia a un valor almacenado en memoria
- **Asignación**: El proceso de asociar un valor a una variable usando `=`
- **Identificador**: El nombre de la variable (debe seguir reglas específicas)

#### Reglas para Nombres de Variables:
1. Pueden contener letras, números y guiones bajos `_`
2. Deben empezar con letra o guion bajo (no con número)
3. Son case-sensitive (mayúsculas y minúsculas importan)
4. No pueden ser palabras reservadas (`if`, `for`, `def`, etc.)
5. Convención: usar `snake_case` (minúsculas con guiones bajos)

```python
# Asignación simple
x = 10
y = 20

# Asignación múltiple
a, b, c = 1, 2, 3
print(f"a={a}, b={b}, c={c}")

# Asignación del mismo valor a múltiples variables
x = y = z = 0
print(f"x={x}, y={y}, z={z}")

# Intercambio de valores (Pythonic way)
a, b = 5, 10
print(f"Antes: a={a}, b={b}")
a, b = b, a  # Intercambio
print(f"Después: a={a}, b={b}")

# Asignación con operaciones
contador = 5
contador += 1  # Equivale a: contador = contador + 1
print(f"Contador: {contador}")

contador *= 2  # Equivale a: contador = contador * 2
print(f"Contador después de *= 2: {contador}")
```

### 4.4 Indexación y Slicing Avanzado

#### Indexación:
- **Positiva**: Empieza en 0 desde la izquierda
- **Negativa**: Empieza en -1 desde la derecha

#### Slicing:
- Formato: `[inicio:fin:paso]`
- `inicio`: Índice donde empieza (incluido)
- `fin`: Índice donde termina (excluido)
- `paso`: Saltos entre elementos (default: 1)

```python
texto = "Python Programming"

print(f"Texto completo: {texto}")
print(f"Longitud: {len(texto)}")

# Indexación positiva
print(f"\nPrimer carácter [0]: {texto[0]}")
print(f"Segundo carácter [1]: {texto[1]}")
print(f"Último carácter [len-1]: {texto[len(texto)-1]}")

# Indexación negativa
print(f"\nÚltimo carácter [-1]: {texto[-1]}")
print(f"Penúltimo carácter [-2]: {texto[-2]}")
print(f"Primer carácter [-len]: {texto[-len(texto)]}")

# Slicing básico
print(f"\nPrimeros 6 caracteres [0:6]: {texto[0:6]}")
print(f"Desde índice 7 hasta el final [7:]: {texto[7:]}")
print(f"Hasta índice 6 [:6]: {texto[:6]}")
print(f"Todo el string [:] : {texto[:]}")

# Slicing con pasos
print(f"\nCada 2 caracteres [::2]: {texto[::2]}")
print(f"Invertido [::-1]: {texto[::-1]}")
print(f"Cada 3 caracteres [::3]: {texto[::3]}")
```

### 4.7 Métodos de Strings (Parte 3: División y Unión)

```python
texto = "Python,Java,JavaScript,C++"

# Dividir strings
lenguajes = texto.split(",")
print(f"Texto original: {texto}")
print(f"split(','): {lenguajes}")

# Dividir con límite
print(f"split(',', 2): {texto.split(',', 2)}")  # Máximo 2 divisiones

# Dividir por líneas
multilinea = "Línea 1\nLínea 2\nLínea 3"
lineas = multilinea.splitlines()
print(f"\nDividir por líneas: {lineas}")

# Unir strings
lista = ["Python", "Java", "JavaScript"]
unido = ", ".join(lista)
print(f"\njoin(', '): {unido}")

# Particionar
texto2 = "Python es genial"
print(f"\npartition('es'): {texto2.partition('es')}")  # (antes, separador, después)
print(f"rpartition('es'): {texto2.rpartition('es')}")  # Desde la derecha
```

### 4.8 Formateo de Strings

Python ofrece varias formas de formatear strings:

1. **f-strings** (recomendado, Python 3.6+)
2. **format()** method
3. **% formatting** (estilo antiguo)

```python
# Formateo avanzado - Números
edad = 30
saldo = 5000.50
print(f"Entero: {edad:05d}")  # 5 dígitos con ceros a la izquierda
print(f"Decimal: {saldo:,.2f}")  # Con comas y 2 decimales
print(f"Porcentaje: {0.25:.2%}")  # Porcentaje
```

```python
nombre = "María"
edad = 30
saldo = 5000.50

# f-strings (más moderno y legible)
mensaje1 = f"{nombre} tiene {edad} años y un saldo de ${saldo:,.2f}"
print(f"f-string: {mensaje1}")

# format() method
mensaje2 = "{} tiene {} años".format(nombre, edad)
mensaje3 = "{0} tiene {1} años".format(nombre, edad)  # Con índices
mensaje4 = "{nombre} tiene {edad} años".format(nombre=nombre, edad=edad)  # Con nombres
print(f"\nformat(): {mensaje2}")

# Formateo avanzado
print(f"\nAlineación:")
print(f"'{nombre:>10}'" )  # Derecha
print(f"'{nombre:<10}'" )  # Izquierda
print(f"'{nombre:^10}'" )  # Centro

print(f"\nFormato de números:")
print(f"Entero: {edad:05d}")  # 5 dígitos con ceros a la izquierda
print(f"Decimal: {saldo:,.2f}")  # Con comas y 2 decimales
print(f"Porcentaje: {0.25:.2%}")  # Porcentaje
```

## 5. Listas 📋

### 5.1 ¿Qué es una Lista?

Una **lista** es una colección ordenada y mutable de elementos. Es una de las estructuras de datos más importantes en Python.

#### Características:
- **Ordenada**: Los elementos mantienen el orden de inserción
- **Mutable**: Puedes modificar, agregar o eliminar elementos
- **Heterogénea**: Puede contener diferentes tipos de datos
- **Indexable**: Acceso por índice (empezando en 0)
- **Permite duplicados**: Puede tener el mismo elemento varias veces

```python
# Crear listas de diferentes formas
lista1 = [1, 2, 3, 4, 5]  # Lista de enteros
lista2 = ["a", "b", "c"]  # Lista de strings
lista3 = [1, "texto", 3.14, True]  # Lista heterogénea
lista4 = []  # Lista vacía
lista5 = list(range(5))  # Usando constructor list()
lista6 = [0] * 5  # Repetición: [0, 0, 0, 0, 0]

print(f"Lista de enteros: {lista1}")
print(f"Lista de strings: {lista2}")
print(f"Lista heterogénea: {lista3}")
print(f"Lista vacía: {lista4}")
print(f"Con range(): {lista5}")
print(f"Con repetición: {lista6}")
```

### 5.2 Acceso a Elementos

Las listas usan la misma indexación que los strings:

```python
numeros = [10, 20, 30, 40, 50]

print(f"Lista: {numeros}")
print(f"Primer elemento [0]: {numeros[0]}")
print(f"Segundo elemento [1]: {numeros[1]}")
print(f"Último elemento [-1]: {numeros[-1]}")
print(f"Penúltimo elemento [-2]: {numeros[-2]}")

# Slicing
print(f"\nPrimeros 3 [0:3]: {numeros[0:3]}")
print(f"Desde índice 2 [2:]: {numeros[2:]}")
print(f"Últimos 2 [:-2]: {numeros[:-2]}")
print(f"Todos [:] : {numeros[:]}")
print(f"Invertida [::-1]: {numeros[::-1]}")
```

### 5.3 Métodos de Listas (Parte 1: Agregar Elementos)

```python
lista = [1, 2, 3]
print(f"Lista inicial: {lista}")

# append() - Agregar al final
lista.append(4)
print(f"Después de append(4): {lista}")

# insert() - Insertar en posición específica
lista.insert(0, 0)  # Insertar 0 en posición 0
print(f"Después de insert(0, 0): {lista}")

# extend() - Agregar múltiples elementos
lista.extend([5, 6, 7])
print(f"Después de extend([5,6,7]): {lista}")

# Concatenación con +
lista = lista + [8, 9]
print(f"Después de + [8,9]: {lista}")

# += (equivalente a extend)
lista += [10]
print(f"Después de += [10]: {lista}")
```

```python
# sort() - Ordenar la lista (modifica la lista original)
lista = [3, 1, 4, 1, 5, 9, 2, 6, 5]
lista_copia = lista.copy()
lista_copia.sort()
print(f"Después de sort(): {lista_copia}")

# sort() descendente
lista_copia2 = lista.copy()
lista_copia2.sort(reverse=True)
print(f"sort(reverse=True): {lista_copia2}")
```

### 5.4 Métodos de Listas (Parte 2: Eliminar Elementos)

```python
# reverse() - Invertir orden
lista = [3, 1, 4, 1, 5, 9, 2, 6, 5]
lista.reverse()
print(f"Después de reverse(): {lista}")
```

```python
lista = [1, 2, 3, 4, 5, 3, 6]
print(f"Lista inicial: {lista}")

# remove() - Eliminar primera ocurrencia de valor
lista.remove(3)
print(f"Después de remove(3): {lista}")

# pop() - Eliminar y retornar por índice
elemento = lista.pop()  # Sin índice, elimina el último
print(f"Elemento eliminado con pop(): {elemento}")
print(f"Lista después de pop(): {lista}")

elemento2 = lista.pop(0)  # Eliminar primer elemento
print(f"Elemento eliminado con pop(0): {elemento2}")
print(f"Lista después de pop(0): {lista}")

# del - Eliminar por índice o slice
del lista[0]  # Eliminar primer elemento
print(f"Después de del lista[0]: {lista}")

# clear() - Vaciar toda la lista
lista.clear()
print(f"Después de clear(): {lista}")
```

### 5.5 Métodos de Listas (Parte 3: Búsqueda y Ordenamiento)

```python
lista = [3, 1, 4, 1, 5, 9, 2, 6, 5]
print(f"Lista inicial: {lista}")

# index() - Encontrar índice de elemento
indice = lista.index(4)
print(f"\nindex(4): {indice}")

# count() - Contar ocurrencias
cantidad = lista.count(1)
print(f"count(1): {cantidad}")

# Verificar pertenencia
print(f"\n5 in lista: {5 in lista}")
print(f"10 in lista: {10 in lista}")

# sort() - Ordenar la lista (modifica la lista original)
lista_copia = lista.copy()
lista_copia.sort()
print(f"\nDespués de sort(): {lista_copia}")

# sort() descendente
lista_copia2 = lista.copy()
lista_copia2.sort(reverse=True)
print(f"sort(reverse=True): {lista_copia2}")

# sorted() - Ordenar sin modificar (retorna nueva lista)
lista_ordenada = sorted(lista)
print(f"\nsorted() (nueva lista): {lista_ordenada}")
print(f"Lista original sin cambios: {lista}")

# reverse() - Invertir orden
lista.reverse()
print(f"\nDespués de reverse(): {lista}")
```

### 5.6 Listas Anidadas (Matrices)

Las listas pueden contener otras listas, creando estructuras multidimensionales:

```python
# Matriz 3x3
matriz = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

print("Matriz 3x3:")
for fila in matriz:
    print(fila)

print(f"\nPrimera fila: {matriz[0]}")
print(f"Elemento [1][2] (fila 1, columna 2): {matriz[1][2]}")

# Acceder a una columna completa
columna_0 = [fila[0] for fila in matriz]
print(f"\nPrimera columna: {columna_0}")
```
