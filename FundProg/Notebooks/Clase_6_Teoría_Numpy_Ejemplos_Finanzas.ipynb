## 2. Ejemplos aplicados (finanzas)

### Ejemplo 1: indexacion y filtrado sobre retornos


```python
precios = np.array([100, 103, 101, 105, 110, 108, 115], dtype=float)
retornos = precios[1:] / precios[:-1] - 1

# Indexacion booleana y fancy indexing
positivos = retornos[retornos > 0]
idx_mejores = np.argsort(retornos)[-2:]

print("precios:", precios)
print("retornos:", np.round(retornos, 4))
print("retornos positivos:", np.round(positivos, 4))
print("indices 2 mejores retornos:", idx_mejores)
print("valores 2 mejores retornos:", np.round(retornos[idx_mejores], 4))
```

### Ejemplo 2: apilado, particion y transposicion de series

En analitica financiera es comun juntar varias series en una sola matriz para modelar escenarios.


```python
serie_a = np.array([100, 102, 101, 103], dtype=float)
serie_b = np.array([50, 49, 51, 52], dtype=float)
serie_c = np.array([80, 81, 79, 82], dtype=float)

M = np.vstack([serie_a, serie_b, serie_c])
print("Matriz por activo (filas=activos):\n", M)

print("\nTranspuesta (columnas=activos):\n", M.T)
print("\nConcatenacion horizontal A y B:", np.hstack([serie_a, serie_b]))

parte_1, parte_2 = np.hsplit(M, 2)
print("\nPrimera mitad de columnas:\n", parte_1)
print("\nSegunda mitad de columnas:\n", parte_2)
```

### Ejemplo 3: algebra lineal y vectorizacion en portafolios


```python
# Rendimientos esperados y matriz de covarianza simplificada
mu = np.array([0.012, 0.008, 0.010])
Sigma = np.array([
    [0.040, 0.006, 0.004],
    [0.006, 0.030, 0.005],
    [0.004, 0.005, 0.035],
])
w = np.array([0.4, 0.3, 0.3])

retorno_esp = w @ mu
riesgo = np.sqrt(w @ Sigma @ w)

print("retorno esperado portafolio:", round(retorno_esp, 5))
print("volatilidad portafolio:", round(riesgo, 5))
print("det(Sigma):", round(np.linalg.det(Sigma), 8))
print("traza(Sigma):", round(np.trace(Sigma), 6))

# Comparacion loop vs vectorizacion para calcular PnL
rng = np.random.default_rng(123)
ret = rng.normal(0.001, 0.02, size=20000)
capital = 10000.0

pnl_loop = np.empty_like(ret)
for i, r in enumerate(ret):
    pnl_loop[i] = capital * r

pnl_vec = capital * ret
print("\nCoinciden loop y vectorizado:", np.allclose(pnl_loop, pnl_vec))
print("PnL promedio:", round(pnl_vec.mean(), 4))
```
