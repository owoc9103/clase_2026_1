## 1. Teoria: fundamentos de NumPy

### 1.1 Que es NumPy

NumPy (Numerical Python) es una libreria para computo numerico eficiente en Python. Su estructura principal es `ndarray`, un arreglo multidimensional homogeneo.

Ventajas clave:
- Operaciones rapidas implementadas en C.
- Menor consumo de memoria que listas anidadas.
- Algebra lineal, estadistica y funciones matematicas con sintaxis clara.
- Trabajo natural con vectores y matrices para analisis numerico.

### 1.2 Propiedades importantes de un arreglo

- `ndim`: numero de ejes.
- `shape`: tamano por eje.
- `size`: cantidad total de elementos.
- `dtype`: tipo de dato almacenado.
- `itemsize`: bytes por elemento.



```python
import numpy as np

x = np.array([10, 20, 30, 40], dtype=np.int32)
A = np.array([[1, 2, 3], [4, 5, 6]], dtype=np.float64)

print("x:", x)
print("x -> ndim:", x.ndim, "shape:", x.shape, "size:", x.size, "dtype:", x.dtype, "itemsize:", x.itemsize)

print("\nA:\n", A)
print("A -> ndim:", A.ndim, "shape:", A.shape, "size:", A.size, "dtype:", A.dtype, "itemsize:", A.itemsize)
```

### 1.3 Creacion de arreglos (adaptado de `tools_numpy.ipynb`)

Funciones mas usadas:

- `np.zeros`, `np.ones`, `np.full`, `np.empty`
- `np.array`
- `np.arange`, `np.linspace`
- `np.random.rand`, `np.random.randn`
- `np.fromfunction`

Nota: `np.empty` reserva memoria sin inicializar valores.


```python
z = np.zeros((2, 3))
o = np.ones((2, 3))
f = np.full((2, 3), np.pi)
e = np.empty((2, 3))
a = np.array([[1, 2], [3, 4]])
r = np.arange(1, 5, 0.5)
l = np.linspace(0, 5/3, 6)
rand_u = np.random.rand(2, 3)
rand_n = np.random.randn(2, 3)
ff = np.fromfunction(lambda i, j: 10 * i + j, (3, 4), dtype=int)

print("zeros:\n", z)
print("\nones:\n", o)
print("\nfull(pi):\n", np.round(f, 3))
print("\nempty (valores no inicializados):\n", e)
print("\narray:\n", a)
print("\narange:", r)
print("\nlinspace:", np.round(l, 4))
print("\nrandom uniforme:\n", np.round(rand_u, 4))
print("\nrandom normal:\n", np.round(rand_n, 4))
print("\nfromfunction:\n", ff)
```

### 1.4 Reorganizacion, operaciones y condicionales

- `reshape` cambia la forma sin cambiar datos.
- `ravel` aplana el arreglo en 1D.
- Operaciones aritmeticas son elemento a elemento.
- NumPy aplica *broadcasting* cuando las formas son compatibles.
- Operadores condicionales (`>`, `<`, `==`) devuelven mascaras booleanas.


```python
g = np.arange(24)
G = g.reshape(4, 6)
print("reshape 4x6:\n", G)
print("\nravel:", G.ravel())

A = np.array([[1.0, 2.0], [3.0, 4.0]])
B = np.array([[10.0, 20.0], [30.0, 40.0]])

print("\nA + B:\n", A + B)
print("\nA * B (elemento a elemento):\n", A * B)
print("\nA @ B (producto matricial):\n", A @ B)

v = np.array([100.0, 200.0])
print("\nBroadcasting A + v:\n", A + v)

mascara = (A > 2.0)
print("\nMascara A > 2:\n", mascara)
print("Valores filtrados:", A[mascara])
```

### 1.5 Funciones matematicas y estadisticas

Funciones frecuentes de agregacion:
- `sum`, `mean`, `min`, `max`, `std`, `var`, `prod`

Funciones universales (ufunc):
- `np.abs`, `np.sqrt`, `np.exp`, `np.log`, `np.sign`
- `np.ceil`, `np.modf`, `np.isnan`
- `np.maximum`, `np.add`, `np.greater`


```python
x = np.array([-2.5, -1.0, 0.0, 1.0, 4.0])
y = np.array([1.0, 3.0, 2.0, 5.0, 4.0])

print("x:", x)
print("abs(x):", np.abs(x))
print("sqrt(|x|):", np.sqrt(np.abs(x)))
print("exp(y):", np.round(np.exp(y), 3))
print("log(y):", np.round(np.log(y), 3))
print("sign(x):", np.sign(x))
print("ceil(x):", np.ceil(x))

frac, ent = np.modf(np.array([2.75, -1.20, 3.0]))
print("\nmodf fracc:", frac, "ent:", ent)
print("isnan:", np.isnan(np.array([1.0, np.nan, 2.0])))
print("maximum(x, y):", np.maximum(x, y))
print("greater(y, 3):", np.greater(y, 3))

print("\n--- Estadisticos sobre y ---")
print("sum:", y.sum(), "mean:", y.mean(), "min:", y.min(), "max:", y.max())
print("std:", round(y.std(), 4), "var:", round(y.var(), 4), "prod:", y.prod())
```

## 2. Ejemplos aplicados (finanzas)

### Ejemplo 1: indexacion y filtrado sobre retornos


```python
precios = np.array([100, 103, 101, 105, 110, 108, 115], dtype=float)
retornos = precios[1:] / precios[:-1] - 1

# Indexacion booleana y fancy indexing
positivos = retornos[retornos > 0]
idx_mejores = np.argsort(retornos)[-2:]

print("precios:", precios)
print("retornos:", np.round(retornos, 4))
print("retornos positivos:", np.round(positivos, 4))
print("indices 2 mejores retornos:", idx_mejores)
print("valores 2 mejores retornos:", np.round(retornos[idx_mejores], 4))
```

### Ejemplo 2: apilado, particion y transposicion de series

En analitica financiera es comun juntar varias series en una sola matriz para modelar escenarios.


```python
serie_a = np.array([100, 102, 101, 103], dtype=float)
serie_b = np.array([50, 49, 51, 52], dtype=float)
serie_c = np.array([80, 81, 79, 82], dtype=float)

M = np.vstack([serie_a, serie_b, serie_c])
print("Matriz por activo (filas=activos):\n", M)

print("\nTranspuesta (columnas=activos):\n", M.T)
print("\nConcatenacion horizontal A y B:", np.hstack([serie_a, serie_b]))

parte_1, parte_2 = np.hsplit(M, 2)
print("\nPrimera mitad de columnas:\n", parte_1)
print("\nSegunda mitad de columnas:\n", parte_2)
```

### Ejemplo 3: algebra lineal y vectorizacion en portafolios


```python
# Rendimientos esperados y matriz de covarianza simplificada
mu = np.array([0.012, 0.008, 0.010])
Sigma = np.array([
    [0.040, 0.006, 0.004],
    [0.006, 0.030, 0.005],
    [0.004, 0.005, 0.035],
])
w = np.array([0.4, 0.3, 0.3])

retorno_esp = w @ mu
riesgo = np.sqrt(w @ Sigma @ w)

print("retorno esperado portafolio:", round(retorno_esp, 5))
print("volatilidad portafolio:", round(riesgo, 5))
print("det(Sigma):", round(np.linalg.det(Sigma), 8))
print("traza(Sigma):", round(np.trace(Sigma), 6))

# Comparacion loop vs vectorizacion para calcular PnL
rng = np.random.default_rng(123)
ret = rng.normal(0.001, 0.02, size=20000)
capital = 10000.0

pnl_loop = np.empty_like(ret)
for i, r in enumerate(ret):
    pnl_loop[i] = capital * r

pnl_vec = capital * ret
print("\nCoinciden loop y vectorizado:", np.allclose(pnl_loop, pnl_vec))
print("PnL promedio:", round(pnl_vec.mean(), 4))
```

## 3. Taller (practica guiada)

### Ejercicio 1 - creacion, reshape e indexacion

1. Crea `a = np.arange(1, 25)`.
2. Conviertelo a matriz `4 x 6`.
3. Extrae la tercera fila y las columnas 2 a 4.
4. Filtra los elementos mayores al promedio de la matriz.


```python
# Solucion sugerida

a = np.arange(1, 25)
M = a.reshape(4, 6)
sub = M[2, 1:4]
filtrados = M[M > M.mean()]

print("Matriz:\n", M)
print("\nTercera fila, col 2..4:", sub)
print("\nPromedio matriz:", M.mean())
print("Elementos > promedio:", filtrados)
```

### Ejercicio 2 - estadistica y funciones universales

Con `x = np.array([-3.5, -1.2, 0.0, 2.8, 4.1])`:

1. Calcula `abs`, `sign`, `ceil` y `sqrt(abs(x))`.
2. Reporta media, desviacion estandar y varianza.
3. Crea mascara de valores positivos y extrae dichos valores.


```python
# Solucion sugerida

x = np.array([-3.5, -1.2, 0.0, 2.8, 4.1])

print("abs:", np.abs(x))
print("sign:", np.sign(x))
print("ceil:", np.ceil(x))
print("sqrt(abs(x)):", np.round(np.sqrt(np.abs(x)), 4))
print("media:", round(x.mean(), 4))
print("std:", round(x.std(), 4), "var:", round(x.var(), 4))

mask = x > 0
print("mascara positivos:", mask)
print("valores positivos:", x[mask])
```

### Ejercicio 3 - algebra lineal, vectorizacion y guardado

1. Resuelve el sistema lineal:
   \[
   \begin{bmatrix}3 & 1 \\ 1 & 2\end{bmatrix}
   x =
   \begin{bmatrix}9 \\ 8\end{bmatrix}
   \]
2. Simula 5,000 retornos normales y calcula `pnl = capital * retorno` con vectorizacion.
3. Guarda el arreglo `pnl` en formato `npy` y `csv`.


```python
# Solucion sugerida

A = np.array([[3.0, 1.0], [1.0, 2.0]])
b = np.array([9.0, 8.0])
x = np.linalg.solve(A, b)

print("Solucion del sistema:", np.round(x, 4))
print("Verificacion A @ x:", np.round(A @ x, 4))

rng = np.random.default_rng(2026)
ret = rng.normal(loc=0.001, scale=0.02, size=5000)
capital = 15000.0
pnl = capital * ret

print("PnL promedio:", round(pnl.mean(), 4))
print("PnL std:", round(pnl.std(), 4))

np.save("pnl_clase6.npy", pnl)
np.savetxt("pnl_clase6.csv", pnl, delimiter=",", fmt="%.6f")
print("Archivos guardados: pnl_clase6.npy y pnl_clase6.csv")
```
