## Ejercicio 1: Clase Cuenta de Ahorros

Crea una clase `CuentaAhorros` con las siguientes características:

**Atributos:**
- titular (string)
- saldo (float)
- tasa_interes (float)

**Métodos:**
- `depositar(monto)` - Deposita dinero
- `retirar(monto)` - Retira dinero (con validación de saldo)
- `aplicar_interes()` - Aplica interés al saldo
- `consultar_saldo()` - Retorna el saldo actual

**Tareas:**
1. Crea la clase con manejo de errores
2. Valida que los montos sean positivos
3. Valida que no se retire más del saldo disponible
4. Prueba la clase con diferentes operaciones

````python
# Solución Ejercicio 1: Clase Cuenta de Ahorros
class SaldoInsuficienteError(Exception):
    pass

class CuentaAhorros:
    def __init__(self, titular, saldo=0, tasa_interes=0.02):
        if saldo < 0:
            raise ValueError("El saldo no puede ser negativo")
        self.titular = titular
        self.saldo = saldo
        self.tasa_interes = tasa_interes

    def depositar(self, monto):
        if monto <= 0:
            raise ValueError("El monto debe ser positivo")
        self.saldo += monto
        return True

    def retirar(self, monto):
        if monto <= 0:
            raise ValueError("El monto debe ser positivo")
        if monto > self.saldo:
            raise SaldoInsuficienteError(f"Saldo insuficiente: ${self.saldo:,.2f}")
        self.saldo -= monto
        return True

    def aplicar_interes(self):
        interes = self.saldo * self.tasa_interes
        self.saldo += interes
        return interes

    def consultar_saldo(self):
        return self.saldo

cuenta = CuentaAhorros("Juan", 1000, 0.03)
cuenta.depositar(500)
print(f"Saldo después de depósito: ${cuenta.consultar_saldo():,.2f}")
cuenta.aplicar_interes()
print(f"Saldo después de interés: ${cuenta.consultar_saldo():,.2f}")
cuenta.retirar(200)
print(f"Saldo final: ${cuenta.consultar_saldo():,.2f}")

````

## Ejercicio 3: Calculadora Financiera con Manejo de Errores

Crea funciones financieras con manejo robusto de errores:

**Funciones:**
1. `calcular_interes_compuesto(principal, tasa, tiempo)`
2. `calcular_pago_prestamo(principal, tasa_anual, años)`
3. `calcular_valor_presente(valor_futuro, tasa, periodos)`

**Tareas:**
1. Valida todos los parámetros
2. Maneja divisiones por cero
3. Maneja valores negativos o inválidos
4. Retorna None en caso de error
5. Prueba con casos válidos e inválidos

````python
# Solución Ejercicio 3: Calculadora financiera con manejo de errores
def calcular_interes_compuesto(principal, tasa, tiempo):
    if principal <= 0 or tiempo < 0:
        return None
    try:
        return principal * (1 + tasa) ** tiempo
    except (TypeError, ZeroDivisionError):
        return None

def calcular_pago_prestamo(principal, tasa_anual, años):
    if principal <= 0 or años <= 0:
        return None
    try:
        r = tasa_anual / 12
        n = años * 12
        if (1 + r)**n - 1 == 0:
            return None
        return principal * (r * (1 + r)**n) / ((1 + r)**n - 1)
    except (TypeError, ZeroDivisionError):
        return None

def calcular_valor_presente(valor_futuro, tasa, periodos):
    if valor_futuro <= 0 or periodos < 0:
        return None
    try:
        if (1 + tasa)**periodos == 0:
            return None
        return valor_futuro / (1 + tasa)**periodos
    except (TypeError, ZeroDivisionError):
        return None

print("Válidos:", calcular_interes_compuesto(10000, 0.05, 5), calcular_pago_prestamo(100000, 0.06, 5))
print("Inválidos:", calcular_interes_compuesto(-1000, 0.05, 5), calcular_pago_prestamo(100000, 0.06, 0))

````

## Ejercicio 4: Clase Portafolio con Herencia

Crea una clase base `Inversion` y una clase `Portafolio` que herede de ella.

**Clase Inversion (base):**
- nombre, monto_inicial, tasa
- `calcular_rendimiento()`

**Clase Portafolio (hereda de Inversion):**
- Agrega lista de inversiones
- `agregar_inversion()`
- `calcular_valor_total()`

**Tareas:**
1. Crea la clase base
2. Crea la clase derivada con herencia
3. Usa super() para llamar métodos de la clase base
4. Prueba ambas clases

````python
# Solución Ejercicio 4: Clase Portafolio con herencia
class Inversion:
    def __init__(self, nombre, monto_inicial, tasa):
        self.nombre = nombre
        self.monto_inicial = monto_inicial
        self.tasa = tasa

    def calcular_rendimiento(self, años=1):
        return self.monto_inicial * (1 + self.tasa) ** años

class Portafolio(Inversion):
    def __init__(self, nombre, monto_inicial=0, tasa=0):
        super().__init__(nombre, monto_inicial, tasa)
        self.inversiones = []

    def agregar_inversion(self, inv):
        self.inversiones.append(inv)

    def calcular_valor_total(self, años=1):
        return sum(inv.calcular_rendimiento(años) for inv in self.inversiones)

p = Portafolio("Mi Portafolio")
p.agregar_inversion(Inversion("AAPL", 5000, 0.08))
p.agregar_inversion(Inversion("GOOGL", 3000, 0.06))
print(f"Valor total portafolio (1 año): ${p.calcular_valor_total(1):,.2f}")

````

## Ejercicio 6: Sistema de Préstamos con Clases

Crea un sistema completo de préstamos usando clases.

**Clase Prestamo:**
- principal, tasa_anual, años, prestatario
- `calcular_pago_mensual()`
- `calcular_total_pagar()`
- `generar_cronograma()` - Lista de fechas de pago

**Tareas:**
1. Crea la clase con validaciones
2. Usa datetime para generar fechas
3. Maneja errores apropiadamente
4. Genera un cronograma de pagos
5. Muestra un resumen del préstamo

````python
# Solución Ejercicio 6: Sistema de préstamos con clases
from datetime import datetime, timedelta

class Prestamo:
    def __init__(self, principal, tasa_anual, años, prestatario):
        if principal <= 0 or años <= 0:
            raise ValueError("Principal y años deben ser positivos")
        self.principal = principal
        self.tasa_anual = tasa_anual
        self.años = años
        self.prestatario = prestatario

    def calcular_pago_mensual(self):
        r = self.tasa_anual / 12
        n = self.años * 12
        return self.principal * (r * (1 + r)**n) / ((1 + r)**n - 1)

    def calcular_total_pagar(self):
        return self.calcular_pago_mensual() * self.años * 12

    def generar_cronograma(self, fecha_inicio=None):
        inicio = fecha_inicio or datetime.now()
        return [inicio + timedelta(days=30*i) for i in range(1, self.años * 12 + 1)]

prestamo = Prestamo(100000, 0.06, 3, "María López")
pago = prestamo.calcular_pago_mensual()
total = prestamo.calcular_total_pagar()
cronograma = prestamo.generar_cronograma()
print("=== RESUMEN PRÉSTAMO ===")
print(f"Prestatario: {prestamo.prestatario} | Principal: ${prestamo.principal:,.2f}")
print(f"Pago mensual: ${pago:,.2f} | Total a pagar: ${total:,.2f}")
print(f"Primeras 3 fechas de pago: {[d.strftime('%d/%m/%Y') for d in cronograma[:3]]}")

````
