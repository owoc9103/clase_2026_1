## Ejercicio 1: Calculadora de Descuentos por Volumen

Crea una función que calcule el precio final de una inversión considerando descuentos por volumen:

**Reglas de descuento:**
- Si la inversión es >= $100,000: 5% de descuento
- Si la inversión es >= $50,000: 3% de descuento
- Si la inversión es >= $25,000: 1% de descuento
- Si la inversión es < $25,000: Sin descuento

**Tareas:**
1. Crea la función `calcular_precio_final(monto)`
2. Prueba con los siguientes montos: $15,000, $30,000, $60,000, $150,000
3. Muestra el monto original, descuento y monto final

````python
# Solución Ejercicio 1: Descuentos por volumen
def calcular_precio_final(monto):
    if monto >= 100000:
        descuento = 0.05
    elif monto >= 50000:
        descuento = 0.03
    elif monto >= 25000:
        descuento = 0.01
    else:
        descuento = 0
    monto_descuento = monto * descuento
    precio_final = monto - monto_descuento
    return precio_final, monto_descuento, descuento * 100

montos = [15000, 30000, 60000, 150000]
print("=== CALCULADORA DE DESCUENTOS ===")
for monto in montos:
    final, desc, porc = calcular_precio_final(monto)
    print(f"Monto: ${monto:,.2f} | Descuento: {porc:.0f}% | Final: ${final:,.2f}")

````

## Ejercicio 2: Generador de Tabla de Interés Compuesto

Crea una función que genere una tabla mostrando el crecimiento de una inversión año por año.

**Parámetros:**
- Capital inicial: $10,000
- Tasa de interés: 8% anual
- Periodo: 10 años

**Fórmula:** Monto = Principal × (1 + tasa)^tiempo

**Tareas:**
1. Crea la función `tabla_interes_compuesto(principal, tasa, años)`
2. Muestra una tabla con: Año, Capital Inicial, Interés Ganado, Capital Final
3. Calcula el total de intereses ganados

````python
# Solución Ejercicio 2: Tabla de interés compuesto
def tabla_interes_compuesto(principal, tasa, años):
    capital = principal
    total_interes = 0
    print(f"{'Año':<6} {'Capital Inicial':<18} {'Interés':<12} {'Capital Final':<15}")
    print("=" * 55)
    for año in range(1, años + 1):
        capital_inicial = capital
        interes = capital * tasa
        capital = capital * (1 + tasa)
        total_interes += interes
        print(f"{año:<6} ${capital_inicial:<17,.2f} ${interes:<11,.2f} ${capital:<14,.2f}")
    print("=" * 55)
    print(f"Total intereses: ${total_interes:,.2f}")
    print(f"Capital final: ${capital:,.2f}")

tabla_interes_compuesto(10000, 0.08, 10)

````

## Ejercicio 3: Analizador de Flujo de Caja

Crea un programa que analice un flujo de caja mensual y determine:

**Datos:**
- Ingresos: [5000, 5200, 4800, 5500, 5100, 5300]
- Egresos: [3500, 3800, 4000, 3600, 3700, 3900]

**Tareas:**
1. Calcula el flujo neto de cada mes (ingresos - egresos)
2. Identifica los meses con flujo positivo y negativo
3. Calcula el flujo acumulado mes a mes
4. Determina el mes con mayor flujo positivo y el mes con mayor flujo negativo
5. Calcula el promedio de ingresos y egresos

````python
# Solución Ejercicio 3: Analizador de flujo de caja
ingresos = [5000, 5200, 4800, 5500, 5100, 5300]
egresos = [3500, 3800, 4000, 3600, 3700, 3900]
meses = ["Ene", "Feb", "Mar", "Abr", "May", "Jun"]

flujo_neto = [i - e for i, e in zip(ingresos, egresos)]
flujo_acum = []
acum = 0
for fn in flujo_neto:
    acum += fn
    flujo_acum.append(acum)

positivos = [meses[i] for i in range(len(flujo_neto)) if flujo_neto[i] > 0]
negativos = [meses[i] for i in range(len(flujo_neto)) if flujo_neto[i] < 0]
max_pos_idx = flujo_neto.index(max(flujo_neto))
max_neg_idx = flujo_neto.index(min(flujo_neto))
prom_ing = sum(ingresos) / len(ingresos)
prom_egr = sum(egresos) / len(egresos)

print("=== FLUJO DE CAJA MENSUAL ===")
for m, i, e, fn, fa in zip(meses, ingresos, egresos, flujo_neto, flujo_acum):
    print(f"{m}: Ingreso ${i:,} | Egreso ${e:,} | Neto ${fn:,} | Acum ${fa:,}")
print(f"\nMeses con flujo positivo: {positivos}")
print(f"Meses con flujo negativo: {negativos}")
print(f"Mayor flujo positivo: {meses[max_pos_idx]} (${flujo_neto[max_pos_idx]:,})")
print(f"Mayor flujo negativo: {meses[max_neg_idx]} (${flujo_neto[max_neg_idx]:,})")
print(f"Promedio ingresos: ${prom_ing:,.2f} | Promedio egresos: ${prom_egr:,.2f}")

````

## Ejercicio 4: Calculadora de Pago de Préstamo

Crea una función que calcule el pago mensual de un préstamo usando la fórmula de anualidad.

**Fórmula:** P = (P0 × r × (1+r)^n) / ((1+r)^n - 1)

Donde:
- P = Pago mensual
- P0 = Monto del préstamo
- r = Tasa de interés mensual
- n = Número de pagos

**Tareas:**
1. Crea la función `calcular_pago_mensual(monto_prestamo, tasa_anual, años)`
2. Prueba con: $200,000 a 5 años con tasa del 6% anual
3. Muestra el pago mensual y el total a pagar

````python
# Solución Ejercicio 4: Pago mensual de préstamo
def calcular_pago_mensual(monto_prestamo, tasa_anual, años):
    r = tasa_anual / 12
    n = años * 12
    pago = (monto_prestamo * r * (1 + r)**n) / ((1 + r)**n - 1)
    return pago

monto = 200000
tasa = 0.06
años = 5
pago = calcular_pago_mensual(monto, tasa, años)
total = pago * años * 12
print("=== PAGO DE PRÉSTAMO ===")
print(f"Monto: ${monto:,.2f} | Tasa: {tasa*100}% anual | Plazo: {años} años")
print(f"Pago mensual: ${pago:,.2f}")
print(f"Total a pagar: ${total:,.2f}")

````

## Ejercicio 5: Simulador de Inversión con Aportes

Crea una función que simule una inversión con aportes mensuales regulares.

**Parámetros:**
- Capital inicial: $5,000
- Aporte mensual: $300
- Tasa de interés: 7% anual
- Duración: 3 años

**Tareas:**
1. Crea la función `simular_inversion(capital_inicial, aporte_mensual, tasa_anual, años)`
2. Calcula mes a mes: aporte, interés ganado, saldo acumulado
3. Muestra un resumen cada 6 meses
4. Al final muestra: total aportado, intereses ganados, saldo final

````python
# Solución Ejercicio 5: Simulador de inversión con aportes
def simular_inversion(capital_inicial, aporte_mensual, tasa_anual, años):
    tasa_mensual = tasa_anual / 12
    meses_tot = años * 12
    saldo = capital_inicial
    total_aportado = capital_inicial
    total_interes = 0
    for mes in range(1, meses_tot + 1):
        interes = saldo * tasa_mensual
        saldo += interes + aporte_mensual
        total_aportado += aporte_mensual
        total_interes += interes
        if mes % 6 == 0:
            print(f"Mes {mes:3}: Saldo ${saldo:,.2f} | Aportado ${total_aportado:,.2f} | Intereses ${total_interes:,.2f}")
    return total_aportado, total_interes, saldo

print("=== SIMULACIÓN INVERSIÓN CON APORTES ===")
ta, ti, sf = simular_inversion(5000, 300, 0.07, 3)
print(f"\nTotal aportado: ${ta:,.2f} | Intereses ganados: ${ti:,.2f} | Saldo final: ${sf:,.2f}")

````

## Ejercicio 6: Clasificador de Perfiles de Inversor

Crea una función que clasifique el perfil de un inversor según su edad, ingresos y tolerancia al riesgo.

**Criterios:**
- **Conservador**: Edad >= 50 O ingresos < $50,000 O tolerancia = "baja"
- **Moderado**: (Edad 30-49 Y ingresos $50,000-$100,000) O tolerancia = "media"
- **Agresivo**: Edad < 30 Y ingresos > $100,000 Y tolerancia = "alta"
- **Equilibrado**: Cualquier otro caso

**Tareas:**
1. Crea la función `clasificar_perfil(edad, ingresos, tolerancia_riesgo)`
2. Prueba con diferentes combinaciones
3. Muestra el perfil y una recomendación de estrategia

````python
# Solución Ejercicio 6: Clasificador de perfil de inversor
def clasificar_perfil(edad, ingresos, tolerancia_riesgo):
    if edad >= 50 or ingresos < 50000 or tolerancia_riesgo == "baja":
        return "Conservador"
    if (30 <= edad <= 49 and 50000 <= ingresos <= 100000) or tolerancia_riesgo == "media":
        return "Moderado"
    if edad < 30 and ingresos > 100000 and tolerancia_riesgo == "alta":
        return "Agresivo"
    return "Equilibrado"

pruebas = [(55, 60000, "media"), (35, 80000, "media"), (28, 120000, "alta"), (40, 40000, "baja")]
print("=== CLASIFICACIÓN DE PERFIL ===")
for edad, ing, tol in pruebas:
    perfil = clasificar_perfil(edad, ing, tol)
    print(f"Edad {edad}, Ingresos ${ing:,}, Tolerancia {tol} → Perfil: {perfil}")

````

## Ejercicio 7: Calculadora de ROI Múltiple

Crea una función que calcule el ROI de múltiples inversiones y determine cuál es la mejor.

**Datos:**
```python
inversiones = [
    ("Proyecto A", 10000, 12500),
    ("Proyecto B", 5000, 6200),
    ("Proyecto C", 15000, 16000),
    ("Proyecto D", 8000, 9500)
]
```

Cada tupla es: (nombre, inversión_inicial, valor_actual)

**Tareas:**
1. Crea la función `analizar_inversiones(lista_inversiones)`
2. Calcula el ROI de cada inversión
3. Identifica la inversión con mayor ROI
4. Ordena las inversiones de mayor a menor ROI
5. Muestra un reporte completo

````python
# Solución Ejercicio 7: ROI múltiple
def analizar_inversiones(lista_inversiones):
    resultados = []
    for nombre, inv_ini, valor_act in lista_inversiones:
        roi = ((valor_act - inv_ini) / inv_ini) * 100
        resultados.append((nombre, inv_ini, valor_act, roi))
    resultados.sort(key=lambda x: x[3], reverse=True)
    return resultados

inversiones = [
    ("Proyecto A", 10000, 12500),
    ("Proyecto B", 5000, 6200),
    ("Proyecto C", 15000, 16000),
    ("Proyecto D", 8000, 9500)
]
res = analizar_inversiones(inversiones)
print("=== ANÁLISIS ROI (ordenado mayor a menor) ===")
for nombre, inv, val, roi in res:
    print(f"{nombre}: Inversión ${inv:,} → Valor ${val:,} | ROI {roi:.2f}%")
print(f"\nMejor inversión: {res[0][0]} (ROI {res[0][3]:.2f}%)")

````
