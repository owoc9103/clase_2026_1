## 4. Más herramientas para control de flujo

### 4.1 La sentencia if

```python
# Ejemplo básico de if
x = 10
if x > 5:
    print(f"{x} es mayor que 5")
```

```python
# if-else
edad = 18
if edad >= 18:
    print("Es mayor de edad")
else:
    print("Es menor de edad")
```

```python
# if-elif-else
nota = 85

if nota >= 90:
    calificacion = "A"
elif nota >= 80:
    calificacion = "B"
elif nota >= 70:
    calificacion = "C"
elif nota >= 60:
    calificacion = "D"
else:
    calificacion = "F"

print(f"Nota: {nota}, Calificación: {calificacion}")
```

```python
# Operadores de comparación
a, b = 5, 10

print(f"{a} == {b}: {a == b}")  # Igual
print(f"{a} != {b}: {a != b}")  # Diferente
print(f"{a} < {b}: {a < b}")     # Menor que
print(f"{a} > {b}: {a > b}")     # Mayor que
print(f"{a} <= {b}: {a <= b}")   # Menor o igual
print(f"{a} >= {b}: {a >= b}")   # Mayor o igual
```

```python
# Operadores lógicos: and, or, not
edad = 25
tiene_licencia = True

if edad >= 18 and tiene_licencia:
    print("Puede conducir")
else:
    print("No puede conducir")

# Ejemplo con or
es_estudiante = True
es_profesor = False

if es_estudiante or es_profesor:
    print("Tiene acceso al campus")
```

### 4.2 La sentencia for

```python
# Iterar sobre una lista
frutas = ["manzana", "banana", "naranja"]
for fruta in frutas:
    print(fruta)
```

```python
# Iterar sobre una cadena
palabra = "Python"
for letra in palabra:
    print(letra, end=" ")
print()
```

```python
# Usar enumerate para obtener índice y valor
meses = ["Enero", "Febrero", "Marzo", "Abril"]
for indice, mes in enumerate(meses, start=1):
    print(f"Mes {indice}: {mes}")
```

### 4.3 La función range()

```python
# range(stop)
for i in range(5):
    print(i, end=" ")
print()
```

```python
# range(start, stop)
for i in range(2, 8):
    print(i, end=" ")
print()
```

```python
# range(start, stop, step)
for i in range(0, 10, 2):
    print(i, end=" ")
print()

# Números negativos
for i in range(10, 0, -1):
    print(i, end=" ")
print()
```

```python
# Crear lista con range
numeros = list(range(1, 11))
print("Lista:", numeros)

# Sumar números del 1 al 10
suma = sum(range(1, 11))
print(f"Suma del 1 al 10: {suma}")
```

### 4.4 break and continue Statements

```python
# break: salir del bucle
for numero in range(1, 11):
    if numero == 5:
        break
    print(numero, end=" ")
print("\nBucle terminado")
```

```python
# continue: saltar a la siguiente iteración
for numero in range(1, 11):
    if numero % 2 == 0:  # Si es par
        continue  # Saltar números pares
    print(numero, end=" ")
print()
```

### 4.5 else Clauses on Loops

```python
# else en bucles for
for i in range(5):
    print(i, end=" ")
else:
    print("\nBucle completado normalmente")
```

```python
# else no se ejecuta si hay break
for i in range(5):
    if i == 3:
        break
    print(i, end=" ")
else:
    print("\nEsto no se imprimirá")
```

### 4.6 La sentencia pass

```python
# pass: placeholder que no hace nada
x = 10
if x > 5:
    pass  # Implementar más tarde
else:
    print("x es menor o igual a 5")
```

### 4.7 La sentencia match (Python 3.10+)

```python
# match-case (similar a switch en otros lenguajes)
dia = "lunes"

match dia:
    case "lunes" | "martes" | "miércoles" | "jueves" | "viernes":
        print("Día laboral")
    case "sábado" | "domingo":
        print("Fin de semana")
    case _:
        print("Día no válido")
```

### 4.8 Definir funciones

```python
# Función simple
def saludar():
    print("¡Hola, bienvenido a Python!")

saludar()
```

```python
# Función con parámetros
def saludar_persona(nombre):
    print(f"¡Hola, {nombre}!")

saludar_persona("María")
saludar_persona("Juan")
```

```python
# Función con múltiples parámetros
def calcular_area(ancho, alto):
    area = ancho * alto
    return area

resultado = calcular_area(5, 10)
print(f"Área: {resultado}")
```

```python
# Función con valor de retorno
def es_par(numero):
    return numero % 2 == 0

print(f"¿4 es par? {es_par(4)}")
print(f"¿5 es par? {es_par(5)}")
```

### 4.9 Más sobre definición de funciones

#### 4.9.1 Argumentos con valores por omisión

```python
# Valores por defecto
def potencia(base, exponente=2):
    return base ** exponente

print(f"2^3 = {potencia(2, 3)}")
print(f"2^2 = {potencia(2)}")  # Usa valor por defecto
```

#### 4.9.2 Palabras claves como argumentos

```python
# Argumentos con palabras clave
def presentar(nombre, edad, ciudad):
    print(f"{nombre}, {edad} años, de {ciudad}")

# Llamada con argumentos posicionales
presentar("Ana", 25, "Bogotá")

# Llamada con argumentos por nombre
presentar(ciudad="Medellín", nombre="Carlos", edad=30)
```

#### 4.9.4 Listas de argumentos arbitrarios

```python
# *args: argumentos posicionales arbitrarios
def sumar(*numeros):
    total = 0
    for num in numeros:
        total += num
    return total

print(f"Suma: {sumar(1, 2, 3)}")
print(f"Suma: {sumar(1, 2, 3, 4, 5)}")
```

```python
# **kwargs: argumentos de palabra clave arbitrarios
def mostrar_info(**datos):
    for clave, valor in datos.items():
        print(f"{clave}: {valor}")

mostrar_info(nombre="Pedro", edad=28, ciudad="Cali")
```

#### 4.9.6 Expresiones lambda

```python
# Función lambda (anónima)
cuadrado = lambda x: x ** 2
print(f"Cuadrado de 5: {cuadrado(5)}")

# Lambda con múltiples parámetros
sumar = lambda a, b: a + b
print(f"Suma: {sumar(3, 4)}")
```

```python
# Usar lambda con map, filter, etc.
numeros = [1, 2, 3, 4, 5]
cuadrados = list(map(lambda x: x ** 2, numeros))
print(f"Cuadrados: {cuadrados}")

# Filtrar números pares
pares = list(filter(lambda x: x % 2 == 0, numeros))
print(f"Pares: {pares}")
```

#### 4.9.7 Cadenas de texto de documentación

```python
def calcular_interes(principal, tasa, tiempo):
    """
    Calcula el interés simple.
    
    Parámetros:
    -----------
    principal : float
        Capital inicial
    tasa : float
        Tasa de interés (decimal, ej: 0.05 para 5%)
    tiempo : float
        Tiempo en años
    
    Retorna:
    --------
    float
        Interés calculado
    """
    return principal * tasa * tiempo

# Ver documentación
help(calcular_interes)
```
