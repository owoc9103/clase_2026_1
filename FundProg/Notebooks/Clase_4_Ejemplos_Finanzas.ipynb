## 1. Módulo de Funciones Financieras

### 1.1 Crear y usar módulo financiero

```python
# Simular un módulo de funciones financieras
# En la práctica, esto estaría en un archivo separado: finanzas.py

def calcular_interes_simple(principal, tasa, tiempo):
    """Calcula interés simple."""
    return principal * tasa * tiempo

def calcular_interes_compuesto(principal, tasa, tiempo):
    """Calcula interés compuesto."""
    return principal * ((1 + tasa) ** tiempo - 1)

def calcular_valor_presente(valor_futuro, tasa, periodos):
    """Calcula valor presente."""
    return valor_futuro / ((1 + tasa) ** periodos)

def calcular_pago_prestamo(principal, tasa_anual, años):
    """Calcula pago mensual de préstamo."""
    tasa_mensual = tasa_anual / 12
    num_pagos = años * 12
    if tasa_mensual > 0:
        pago = principal * (tasa_mensual * (1 + tasa_mensual)**num_pagos) / \
               ((1 + tasa_mensual)**num_pagos - 1)
    else:
        pago = principal / num_pagos
    return pago

# Usar las funciones
print("=== FUNCIONES FINANCIERAS ===")
print(f"Interés simple: ${calcular_interes_simple(10000, 0.05, 2):,.2f}")
print(f"Interés compuesto: ${calcular_interes_compuesto(10000, 0.05, 2):,.2f}")
print(f"Valor presente: ${calcular_valor_presente(5000, 0.08, 3):,.2f}")
print(f"Pago mensual préstamo: ${calcular_pago_prestamo(100000, 0.06, 5):,.2f}")
```

## 2. Guardar y Leer Datos Financieros en JSON

### 2.1 Guardar Portafolio en JSON

```python
import json
from datetime import datetime

# Datos del portafolio
portafolio = {
    "fecha_actualizacion": datetime.now().strftime("%Y-%m-%d"),
    "inversor": "Juan Pérez",
    "inversiones": [
        {
            "simbolo": "AAPL",
            "nombre": "Apple Inc.",
            "cantidad": 50,
            "precio_compra": 150.25,
            "fecha_compra": "2024-01-15"
        },
        {
            "simbolo": "GOOGL",
            "nombre": "Alphabet Inc.",
            "cantidad": 20,
            "precio_compra": 2500.00,
            "fecha_compra": "2024-02-01"
        }
    ]
}

# Guardar en JSON
with open("portafolio.json", "w", encoding="utf-8") as archivo:
    json.dump(portafolio, archivo, indent=2, ensure_ascii=False)

print("Portafolio guardado exitosamente")
print(json.dumps(portafolio, indent=2, ensure_ascii=False))
```

```python
# Leer portafolio desde JSON
with open("portafolio.json", "r", encoding="utf-8") as archivo:
    portafolio_cargado = json.load(archivo)

print("=== PORTAFOLIO CARGADO ===")
print(f"Inversor: {portafolio_cargado['inversor']}")
print(f"Fecha actualización: {portafolio_cargado['fecha_actualizacion']}")
print(f"\nInversiones:")
for inv in portafolio_cargado['inversiones']:
    valor = inv['cantidad'] * inv['precio_compra']
    print(f"  {inv['simbolo']} - {inv['nombre']}: {inv['cantidad']} acciones, "
          f"Valor: ${valor:,.2f}")
```

## 3. Procesar Archivos CSV de Transacciones

### 3.1 Leer y Analizar Transacciones

```python
# Crear archivo CSV de ejemplo
csv_contenido = """Fecha,Tipo,Monto,Descripcion
2024-01-15,Ingreso,5000,Salario
2024-01-18,Egreso,-1200,Alquiler
2024-01-20,Egreso,-350,Alimentación
2024-01-22,Ingreso,500,Freelance
2024-01-25,Egreso,-200,Transporte
"""

with open("transacciones.csv", "w", encoding="utf-8") as archivo:
    archivo.write(csv_contenido)

print("Archivo CSV creado")
```

```python
# Leer y procesar CSV
transacciones = []

with open("transacciones.csv", "r", encoding="utf-8") as archivo:
    lineas = archivo.readlines()
    # Saltar encabezado
    for linea in lineas[1:]:
        linea = linea.strip()
        if linea:
            partes = linea.split(",")
            transacciones.append({
                "fecha": partes[0],
                "tipo": partes[1],
                "monto": float(partes[2]),
                "descripcion": partes[3]
            })

print("=== TRANSACCIONES ===")
for trans in transacciones:
    print(f"{trans['fecha']} | {trans['tipo']:8} | ${trans['monto']:8,.2f} | {trans['descripcion']}")

# Análisis
ingresos = sum(t['monto'] for t in transacciones if t['tipo'] == 'Ingreso')
egresos = sum(abs(t['monto']) for t in transacciones if t['tipo'] == 'Egreso')

print(f"\nTotal ingresos: ${ingresos:,.2f}")
print(f"Total egresos: ${egresos:,.2f}")
print(f"Saldo: ${ingresos - egresos:,.2f}")
```

## 4. Generar Reportes Financieros

### 4.1 Crear Reporte en Archivo de Texto

```python
from datetime import datetime

def generar_reporte_financiero(cliente, saldo, transacciones, nombre_archivo):
    """Genera un reporte financiero en archivo de texto."""
    
    with open(nombre_archivo, "w", encoding="utf-8") as archivo:
        archivo.write("=" * 60 + "\n")
        archivo.write("REPORTE FINANCIERO\n")
        archivo.write("=" * 60 + "\n\n")
        archivo.write(f"Cliente: {cliente}\n")
        archivo.write(f"Fecha: {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}\n")
        archivo.write(f"Saldo actual: ${saldo:,.2f}\n\n")
        
        archivo.write("TRANSACCIONES:\n")
        archivo.write("-" * 60 + "\n")
        archivo.write(f"{'Fecha':<12} {'Tipo':<10} {'Monto':<15} {'Descripción'}\n")
        archivo.write("-" * 60 + "\n")
        
        for trans in transacciones:
            archivo.write(f"{trans['fecha']:<12} {trans['tipo']:<10} "
                        f"${trans['monto']:>13,.2f} {trans['descripcion']}\n")
        
        archivo.write("\n" + "=" * 60 + "\n")
    
    print(f"Reporte guardado en {nombre_archivo}")

# Datos de ejemplo
cliente = "María González"
saldo = 5000.50
transacciones = [
    {"fecha": "2024-01-15", "tipo": "Ingreso", "monto": 5000, "descripcion": "Salario"},
    {"fecha": "2024-01-18", "tipo": "Egreso", "monto": -1200, "descripcion": "Alquiler"},
    {"fecha": "2024-01-20", "tipo": "Egreso", "monto": -350, "descripcion": "Alimentación"}
]

generar_reporte_financiero(cliente, saldo, transacciones, "reporte_financiero.txt")
```

```python
# Leer el reporte generado
with open("reporte_financiero.txt", "r", encoding="utf-8") as archivo:
    print(archivo.read())
```

## 5. Usar Módulos Estándar para Análisis Financiero

### 5.1 Análisis con Collections

```python
from collections import defaultdict, Counter

# Agrupar gastos por categoría
gastos = [
    ("Alimentación", 350),
    ("Transporte", 120),
    ("Alimentación", 280),
    ("Servicios", 180),
    ("Transporte", 90)
]

# Usar defaultdict
gastos_por_categoria = defaultdict(list)
for categoria, monto in gastos:
    gastos_por_categoria[categoria].append(monto)

print("=== GASTOS POR CATEGORÍA ===")
for categoria, montos in gastos_por_categoria.items():
    total = sum(montos)
    print(f"{categoria}: ${total:,.2f} ({len(montos)} transacciones)")
```

### 5.2 Trabajar con Fechas

```python
from datetime import datetime, timedelta

# Calcular fechas de vencimiento
fecha_inicio = datetime(2024, 1, 1)
plazo_dias = 90
fecha_vencimiento = fecha_inicio + timedelta(days=plazo_dias)

print(f"Fecha inicio: {fecha_inicio.strftime('%Y-%m-%d')}")
print(f"Plazo: {plazo_dias} días")
print(f"Fecha vencimiento: {fecha_vencimiento.strftime('%Y-%m-%d')}")

# Días restantes
hoy = datetime.now()
if hoy < fecha_vencimiento:
    dias_restantes = (fecha_vencimiento - hoy).days
    print(f"Días restantes: {dias_restantes}")
```
