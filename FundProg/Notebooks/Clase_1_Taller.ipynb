### Solución Ejercicio 1

```python
# Solución Ejercicio 1
capital = 15000
tasa = 0.065
tiempo = 4

interes = capital * tasa * tiempo
monto_final = capital + interes

print(f"Capital inicial: ${capital:,.2f}")
print(f"Tasa de interés: {tasa * 100}% anual")
print(f"Tiempo: {tiempo} años")
print(f"Interés ganado: ${interes:,.2f}")
print(f"Monto final: ${monto_final:,.2f}")
```

### Solución Ejercicio 2

```python
# Solución Ejercicio 2
inversiones = [5000, 8000, 3000, 4000]
nombres = ["Acción A", "Acción B", "Acción C", "Acción D"]

total = sum(inversiones)

print("=== ANÁLISIS DE PORTAFOLIO ===")
for nombre, inversion in zip(nombres, inversiones):
    porcentaje = (inversion / total) * 100
    print(f"{nombre}: ${inversion:,.2f} ({porcentaje:.2f}%)")

print(f"\nTotal del portafolio: ${total:,.2f}")
```
