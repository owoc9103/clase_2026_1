## 1. ¿Qué es Python? 🐍

### 1.1 Definición Simple

**Python** es un lenguaje de programación de alto nivel, interpretado y de propósito general. En términos simples, es una forma de dar instrucciones a una computadora usando un lenguaje que se parece mucho al lenguaje humano.

### ¿Por qué se llama Python?

Python fue creado por Guido van Rossum a finales de los años 80. El nombre viene del programa de comedia británico "Monty Python's Flying Circus", del cual Guido era fan.

### 1.2 Características Principales de Python

Python tiene varias características que lo hacen especial:

#### ✨ **Sintaxis Simple y Legible**
- El código Python se lee casi como inglés
- No requiere muchos símbolos especiales
- Facilita el aprendizaje y la lectura del código

#### 📊 **Versátil y Poderoso**
- Se usa en muchas áreas: web, ciencia de datos, inteligencia artificial, automatización, etc.
- Tiene una gran biblioteca estándar
- Miles de librerías disponibles (pandas, numpy, matplotlib, etc.)

#### 🎯 **Multiplataforma**
- Funciona en Windows, Mac, Linux
- El mismo código funciona en diferentes sistemas operativos

#### ⚡ **Interpretado**
- No necesita compilación previa
- Ejecuta el código línea por línea
- Ideal para desarrollo rápido y experimentación

`````python
# Ejemplo de código Python - Simple y legible
nombre = "María"
edad = 25

if edad >= 18:
    print(f"{nombre} es mayor de edad")
else:
    print(f"{nombre} es menor de edad")
`````

## 2. ¿Qué es Google Colab? 📓

### 2.1 Definición

### 3.6 Magic Commands (Comandos Mágicos)

Los **magic commands** son comandos especiales que empiezan con `%` o `%%`. Son atajos para tareas comunes.

#### Magic Commands Útiles:

- `%time` - Mide el tiempo de ejecución de una línea
- `%%time` - Mide el tiempo de ejecución de toda la celda
- `%run` - Ejecuta un script de Python
- `%who` - Lista todas las variables
- `%whos` - Lista variables con detalles
- `%reset` - Borra todas las variables
- `%pwd` - Muestra el directorio actual
- `%cd` - Cambia de directorio
- `%ls` - Lista archivos
- `%%writefile` - Escribe contenido a un archivo
- `%load` - Carga código desde un archivo

`````python
# Magic command: Ver variables actuales
%whos
`````

`````python
# Magic command: Medir tiempo de ejecución
%%time

suma = sum(range(1000000))
print(f"Suma: {suma}")
`````

`````python
# Magic command: Escribir a un archivo
%%writefile ejemplo.txt
Este es un archivo de ejemplo
Creado desde Google Colab
Línea 3
`````

`````python
# Leer el archivo creado
with open('ejemplo.txt', 'r') as f:
    contenido = f.read()
    print(contenido)
`````

### 3.7 Visualización de Datos 📊

Google Colab soporta muchas librerías de visualización. Las más comunes son:

- **matplotlib** - Gráficos básicos y avanzados
- **seaborn** - Gráficos estadísticos
- **plotly** - Gráficos interactivos
- **pandas** - Visualización integrada

`````python
# Ejemplo de visualización con matplotlib
import matplotlib.pyplot as plt
import numpy as np

# Crear datos
x = np.linspace(0, 10, 100)
y = np.sin(x)

# Crear gráfico
plt.figure(figsize=(10, 6))
plt.plot(x, y, label='sin(x)', linewidth=2)
plt.xlabel('X')
plt.ylabel('Y')
plt.title('Gráfico de sin(x)')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
`````

### 3.10 Tipos de Runtime 🖥️

Google Colab ofrece diferentes tipos de runtime (entornos de ejecución):

#### Runtime Estándar
- CPU estándar
- Memoria RAM limitada
- Gratis

#### GPU (Graphics Processing Unit)
- Acelera cálculos paralelos
- Ideal para Machine Learning y Deep Learning
- Gratis con límites de uso

#### TPU (Tensor Processing Unit)
- Unidad de procesamiento especializada de Google
- Muy rápida para TensorFlow
- Disponible bajo ciertas condiciones

#### Cómo Cambiar el Runtime
1. Ve a **Runtime** → **Change runtime type**
2. Selecciona el tipo de hardware
3. Guarda los cambios

**Nota**: Cambiar el runtime reinicia el entorno y se pierden las variables en memoria.

`````python
# Verificar el tipo de runtime actual
import tensorflow as tf

print(f"Versión de TensorFlow: {tf.__version__}")
print(f"GPU disponible: {tf.config.list_physical_devices('GPU')}")
print(f"TPU disponible: {tf.config.list_physical_devices('TPU')}")
`````

### 3.12 Widgets Interactivos 🎛️

Puedes crear interfaces interactivas usando widgets:

`````python
# Ejemplo de widgets interactivos
from ipywidgets import interact, interactive, fixed
import ipywidgets as widgets

def graficar_funcion(a=1, b=0):
    import matplotlib.pyplot as plt
    import numpy as np
    
    x = np.linspace(-10, 10, 100)
    y = a * x + b
    
    plt.figure(figsize=(8, 6))
    plt.plot(x, y, linewidth=2)
    plt.grid(True, alpha=0.3)
    plt.title(f'y = {a}x + {b}')
    plt.show()

# Crear widget interactivo
interact(graficar_funcion, a=(-5, 5, 0.1), b=(-10, 10, 1))
`````

## 4. Mejores Prácticas para Notebooks 📚

### 4.1 Organización del Notebook

1. **Estructura clara**
   - Usa títulos y subtítulos
   - Divide en secciones lógicas
   - Mantén un flujo coherente

2. **Documentación**
   - Explica qué hace cada sección
   - Comenta código complejo
   - Incluye ejemplos

3. **Orden de ejecución**
   - Ejecuta celdas en orden
   - No saltes celdas
   - Reinicia y ejecuta todo antes de compartir

### 4.2 Gestión de Celdas

#### Buenas Prácticas:
- **Celdas pequeñas**: Una idea por celda
- **Nombres descriptivos**: Usa markdown para explicar
- **Evita celdas muy largas**: Divide código complejo
- **Limpia salidas**: Elimina salidas innecesarias

#### Organizar Celdas:
- Usa el menú contextual (clic derecho)
- Mover celdas arriba/abajo
- Duplicar celdas para experimentar
- Eliminar celdas no usadas

### 4.3 Manejo de Errores

Cuando encuentres errores:

1. **Lee el mensaje de error** - Python da mensajes claros
2. **Revisa la línea indicada** - El error muestra dónde está
3. **Verifica variables** - Usa `%whos` para ver el estado
4. **Reinicia y ejecuta de nuevo** - A veces ayuda limpiar el estado

`````python
# Ejemplo de manejo de errores
try:
    resultado = 10 / 0
except ZeroDivisionError as e:
    print(f"Error capturado: {e}")
    print("No se puede dividir por cero")
else:
    print("Operación exitosa")
finally:
    print("Finalizando...")
`````

### 4.4 Optimización de Rendimiento

Para notebooks que procesan muchos datos:

1. **Usa GPU cuando sea necesario**
2. **Evita recalcular**: Guarda resultados intermedios
3. **Limpia memoria**: Elimina variables grandes no usadas
4. **Mide tiempo**: Usa `%%time` para identificar cuellos de botella

`````python
# Ejemplo: Medir tiempo de ejecución
%%time

import numpy as np

# Operación que puede ser lenta
datos = np.random.rand(1000000)
resultado = np.sum(datos ** 2)

print(f"Resultado: {resultado:.2f}")
`````
