## Soluciones de Referencia

> **Nota:** Intenta resolver los ejercicios antes de ver las soluciones.

### Solución Ejercicio 1

````python
# Solución Ejercicio 1
portafolio = {
    "AAPL": {"cantidad": 50, "precio_compra": 150},
    "GOOGL": {"cantidad": 20, "precio_compra": 2500},
    "MSFT": {"cantidad": 30, "precio_compra": 300}
}

precios_actuales = {"AAPL": 175, "GOOGL": 2800, "MSFT": 350}

resultados = []
for simbolo, info in portafolio.items():
    valor_compra = info["cantidad"] * info["precio_compra"]
    valor_actual = info["cantidad"] * precios_actuales[simbolo]
    ganancia = valor_actual - valor_compra
    roi = (ganancia / valor_compra) * 100
    resultados.append((simbolo, ganancia, roi))

# Ordenar por ROI
resultados.sort(key=lambda x: x[2], reverse=True)
print("Mejor:", resultados[0])
print("Peor:", resultados[-1])
````
