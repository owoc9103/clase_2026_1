## 1. Clase para Cuenta Bancaria con Manejo de Errores

### 1.1 Cuenta Bancaria Robusta

```python
# Excepciones personalizadas
class SaldoInsuficienteError(Exception):
    """Excepción para saldo insuficiente."""
    pass

class MontoInvalidoError(Exception):
    """Excepción para monto inválido."""
    pass

# Clase Cuenta Bancaria
class CuentaBancaria:
    """Clase que representa una cuenta bancaria."""
    
    def __init__(self, titular, saldo_inicial=0, numero_cuenta=None):
        if saldo_inicial < 0:
            raise ValueError("El saldo inicial no puede ser negativo")
        self.titular = titular
        self.saldo = saldo_inicial
        self.numero_cuenta = numero_cuenta or f"ACC-{id(self)}"
        self.transacciones = []
    
    def depositar(self, monto):
        """Deposita dinero en la cuenta."""
        try:
            if monto <= 0:
                raise MontoInvalidoError(f"El monto debe ser positivo: ${monto}")
            self.saldo += monto
            self.transacciones.append(("Depósito", monto))
            return True
        except MontoInvalidoError as e:
            print(f"Error: {e}")
            return False
    
    def retirar(self, monto):
        """Retira dinero de la cuenta."""
        try:
            if monto <= 0:
                raise MontoInvalidoError(f"El monto debe ser positivo: ${monto}")
            if monto > self.saldo:
                raise SaldoInsuficienteError(
                    f"Saldo insuficiente. Tienes ${self.saldo:,.2f}, intentas retirar ${monto:,.2f}"
                )
            self.saldo -= monto
            self.transacciones.append(("Retiro", -monto))
            return True
        except (MontoInvalidoError, SaldoInsuficienteError) as e:
            print(f"Error: {e}")
            return False
    
    def consultar_saldo(self):
        """Consulta el saldo actual."""
        return self.saldo
    
    def obtener_historial(self):
        """Obtiene el historial de transacciones."""
        return self.transacciones
    
    def __str__(self):
        return f"Cuenta {self.numero_cuenta} - {self.titular}: ${self.saldo:,.2f}"

# Usar la clase
cuenta = CuentaBancaria("Juan Pérez", 1000)
print(cuenta)
cuenta.depositar(500)
print(f"Saldo: ${cuenta.consultar_saldo():,.2f}")
cuenta.retirar(200)
print(f"Saldo: ${cuenta.consultar_saldo():,.2f}")
cuenta.retirar(2000)  # Intentar retirar más de lo disponible
print(f"Historial: {cuenta.obtener_historial()}")
```

## 2. Clase para Inversión con Cálculos Financieros

### 2.1 Clase Inversion

```python
from datetime import datetime

class Inversion:
    """Clase que representa una inversión."""
    
    def __init__(self, nombre, monto_inicial, tasa_anual, fecha_inicio=None):
        if monto_inicial <= 0:
            raise ValueError("El monto inicial debe ser positivo")
        if tasa_anual < 0:
            raise ValueError("La tasa no puede ser negativa")
        
        self.nombre = nombre
        self.monto_inicial = monto_inicial
        self.tasa_anual = tasa_anual
        self.fecha_inicio = fecha_inicio or datetime.now()
    
    def calcular_valor_futuro(self, años):
        """Calcula el valor futuro de la inversión."""
        if años < 0:
            raise ValueError("Los años no pueden ser negativos")
        return self.monto_inicial * (1 + self.tasa_anual) ** años
    
    def calcular_interes_ganado(self, años):
        """Calcula el interés ganado."""
        valor_futuro = self.calcular_valor_futuro(años)
        return valor_futuro - self.monto_inicial
    
    def calcular_roi(self, valor_actual):
        """Calcula el ROI basado en el valor actual."""
        if valor_actual < 0:
            raise ValueError("El valor actual no puede ser negativo")
        return ((valor_actual - self.monto_inicial) / self.monto_inicial) * 100
    
    def __str__(self):
        return f"Inversión: {self.nombre}, Monto: ${self.monto_inicial:,.2f}, "
               f"Tasa: {self.tasa_anual*100:.2f}%"

# Usar la clase
inversion = Inversion("Fondo de Ahorro", 10000, 0.07)
print(inversion)
valor_5_años = inversion.calcular_valor_futuro(5)
print(f"Valor en 5 años: ${valor_5_años:,.2f}")
interes = inversion.calcular_interes_ganado(5)
print(f"Interés ganado: ${interes:,.2f}")
```

## 3. Clase Portafolio con Herencia

### 3.1 Portafolio de Inversiones

```python
class Portafolio:
    """Clase que representa un portafolio de inversiones."""
    
    def __init__(self, nombre, propietario):
        self.nombre = nombre
        self.propietario = propietario
        self.inversiones = {}
    
    def agregar_inversion(self, simbolo, inversion):
        """Agrega una inversión al portafolio."""
        if simbolo in self.inversiones:
            raise ValueError(f"La inversión {simbolo} ya existe")
        self.inversiones[simbolo] = inversion
    
    def eliminar_inversion(self, simbolo):
        """Elimina una inversión del portafolio."""
        if simbolo not in self.inversiones:
            raise KeyError(f"La inversión {simbolo} no existe")
        del self.inversiones[simbolo]
    
    def calcular_valor_total(self, precios_actuales):
        """Calcula el valor total del portafolio."""
        valor_total = 0
        for simbolo, inversion in self.inversiones.items():
            if simbolo not in precios_actuales:
                raise KeyError(f"Precio no disponible para {simbolo}")
            precio_actual = precios_actuales[simbolo]
            valor_total += precio_actual * inversion.get("cantidad", 0)
        return valor_total
    
    def __str__(self):
        return f"Portafolio {self.nombre} - Propietario: {self.propietario}, "
               f"Inversiones: {len(self.inversiones)}"

# Usar la clase
portafolio = Portafolio("Mi Portafolio", "Juan Pérez")

portafolio.agregar_inversion("AAPL", {"cantidad": 50, "precio_compra": 150})
portafolio.agregar_inversion("GOOGL", {"cantidad": 20, "precio_compra": 2500})

print(portafolio)

precios = {"AAPL": 175, "GOOGL": 2800}
valor_total = portafolio.calcular_valor_total(precios)
print(f"Valor total: ${valor_total:,.2f}")
```

## 4. Función con Manejo Robusto de Errores

### 4.1 Calculadora Financiera Segura

```python
def calcular_pago_prestamo_seguro(principal, tasa_anual, años):
    """Calcula el pago mensual de un préstamo con manejo de errores."""
    try:
        # Validar entradas
        if principal <= 0:
            raise ValueError("El principal debe ser positivo")
        if tasa_anual < 0 or tasa_anual > 1:
            raise ValueError("La tasa debe estar entre 0 y 1")
        if años <= 0:
            raise ValueError("Los años deben ser positivos")
        
        # Calcular
        tasa_mensual = tasa_anual / 12
        num_pagos = años * 12
        
        if tasa_mensual == 0:
            pago = principal / num_pagos
        else:
            pago = principal * (tasa_mensual * (1 + tasa_mensual)**num_pagos) / \
                   ((1 + tasa_mensual)**num_pagos - 1)
        
        return pago
    
    except ValueError as e:
        print(f"Error de validación: {e}")
        return None
    except Exception as e:
        print(f"Error inesperado: {e}")
        return None

# Pruebas
print(f"Pago válido: ${calcular_pago_prestamo_seguro(100000, 0.06, 5):,.2f}")
calcular_pago_prestamo_seguro(-100000, 0.06, 5)  # Error
calcular_pago_prestamo_seguro(100000, 1.5, 5)  # Error
```

## 5. Usar Biblioteca Estándar para Análisis

### 5.1 Análisis con Statistics

```python
import statistics
from datetime import datetime, timedelta

# Rendimientos mensuales
rendimientos = [2.5, -1.2, 3.8, 0.5, -2.1, 4.2, 1.8, -0.9, 2.3, 1.5, -1.8, 3.1]

print("=== ANÁLISIS ESTADÍSTICO ===")
print(f"Promedio: {statistics.mean(rendimientos):.2f}%")
print(f"Mediana: {statistics.median(rendimientos):.2f}%")
print(f"Desviación estándar: {statistics.stdev(rendimientos):.2f}%")
print(f"Mínimo: {min(rendimientos):.2f}%")
print(f"Máximo: {max(rendimientos):.2f}%")
```

### 5.2 Trabajar con Fechas Financieras

```python
from datetime import datetime, timedelta

def generar_fechas_pago(fecha_inicio, num_pagos, frecuencia="mensual"):
    """Genera fechas de pago para un préstamo."""
    fechas = []
    fecha_actual = fecha_inicio
    
    if frecuencia == "mensual":
        delta = timedelta(days=30)
    elif frecuencia == "trimestral":
        delta = timedelta(days=90)
    else:
        delta = timedelta(days=365)
    
    for i in range(num_pagos):
        fechas.append(fecha_actual)
        fecha_actual += delta
    
    return fechas

# Generar fechas de pago
fecha_inicio = datetime(2024, 1, 1)
fechas = generar_fechas_pago(fecha_inicio, 12, "mensual")

print("Fechas de pago:")
for i, fecha in enumerate(fechas, 1):
    print(f"Pago {i}: {fecha.strftime('%Y-%m-%d')}")
```
