### 6.2 Crear módulos propios

```python
# Crear un módulo de funciones financieras
# Guardar esto en un archivo: finanzas.py

# Ejemplo de contenido del módulo:
def calcular_interes_simple(principal, tasa, tiempo):
    """Calcula interés simple."""
    return principal * tasa * tiempo

def calcular_interes_compuesto(principal, tasa, tiempo):
    """Calcula interés compuesto."""
    return principal * (1 + tasa) ** tiempo - principal

# Para usar: import finanzas
# Luego: finanzas.calcular_interes_simple(1000, 0.05, 2)
```

### 6.3 Módulos estándar útiles

```python
# datetime - Trabajar con fechas
from datetime import datetime, timedelta

hoy = datetime.now()
print(f"Hoy: {hoy}")
print(f"Fecha formateada: {hoy.strftime('%Y-%m-%d %H:%M:%S')}")

mañana = hoy + timedelta(days=1)
print(f"Mañana: {mañana.date()}")
```

```python
# random - Números aleatorios
import random

print(f"Número aleatorio: {random.randint(1, 100)}")
print(f"Número decimal: {random.uniform(0, 1):.2f}")

lista = [1, 2, 3, 4, 5]
print(f"Elemento aleatorio: {random.choice(lista)}")
print(f"Muestra: {random.sample(lista, 3)}")
```

```python
# collections - Estructuras de datos avanzadas
from collections import defaultdict, Counter

# defaultdict - diccionario con valor por defecto
gastos = defaultdict(list)
gastos["Alimentación"].append(350)
gastos["Transporte"].append(120)
print("Gastos:", dict(gastos))

# Counter - contar elementos
transacciones = ["Ingreso", "Egreso", "Ingreso", "Ingreso", "Egreso"]
contador = Counter(transacciones)
print(f"Conteo: {contador}")
```

## 7. Entrada y Salida

### 7.1 Formateo elegante de la salida

```python
# f-strings (Python 3.6+)
nombre = "María"
edad = 30
saldo = 5000.50

mensaje = f"{nombre} tiene {edad} años y un saldo de ${saldo:,.2f}"
print(mensaje)
```

```python
# format() method
texto = "El precio es {precio:.2f} y la cantidad es {cantidad}"
print(texto.format(precio=99.99, cantidad=5))

# Con índices
print("Coordenadas: ({0}, {1})".format(10, 20))
```

### 7.2 Leyendo y escribiendo archivos

```python
# Escribir a un archivo
contenido = """Línea 1
Línea 2
Línea 3"""

with open("ejemplo.txt", "w", encoding="utf-8") as archivo:
    archivo.write(contenido)

print("Archivo escrito exitosamente")
```

```python
# Leer de un archivo
with open("ejemplo.txt", "r", encoding="utf-8") as archivo:
    contenido = archivo.read()
    print("Contenido completo:")
    print(contenido)
```

```python
# Leer línea por línea
with open("ejemplo.txt", "r", encoding="utf-8") as archivo:
    print("Línea por línea:")
    for i, linea in enumerate(archivo, 1):
        print(f"{i}: {linea.strip()}")
```

```python
# Leer todas las líneas en una lista
with open("ejemplo.txt", "r", encoding="utf-8") as archivo:
    lineas = archivo.readlines()
    print(f"Total de líneas: {len(lineas)}")
    print(f"Líneas: {lineas}")
```

### 7.3 Guardar datos estructurados con JSON

```python
import json

# Datos a guardar
datos_financieros = {
    "cliente": "Juan Pérez",
    "saldo": 5000.50,
    "transacciones": [
        {"fecha": "2024-01-15", "tipo": "Ingreso", "monto": 1000},
        {"fecha": "2024-01-20", "tipo": "Egreso", "monto": -500}
    ]
}

# Escribir JSON
with open("datos.json", "w", encoding="utf-8") as archivo:
    json.dump(datos_financieros, archivo, indent=2, ensure_ascii=False)

print("JSON guardado exitosamente")
```

```python
# Leer JSON
with open("datos.json", "r", encoding="utf-8") as archivo:
    datos = json.load(archivo)
    print("Datos cargados:")
    print(json.dumps(datos, indent=2, ensure_ascii=False))
```

## Funciones Avanzadas

### Decoradores básicos

```python
# Decorador simple
def medir_tiempo(func):
    def wrapper(*args, **kwargs):
        import time
        inicio = time.time()
        resultado = func(*args, **kwargs)
        fin = time.time()
        print(f"{func.__name__} tardó {fin - inicio:.4f} segundos")
        return resultado
    return wrapper

@medir_tiempo
def calcular_cuadrados(n):
    return [x**2 for x in range(n)]

resultado = calcular_cuadrados(1000000)
```

### Generadores

```python
# Generador simple
def numeros_pares(n):
    for i in range(n):
        if i % 2 == 0:
            yield i

# Usar generador
for num in numeros_pares(10):
    print(num, end=" ")
print()
```

```python
# Expresión generadora
cuadrados = (x**2 for x in range(10))
print("Generador:", cuadrados)
print("Valores:", list(cuadrados))
```
