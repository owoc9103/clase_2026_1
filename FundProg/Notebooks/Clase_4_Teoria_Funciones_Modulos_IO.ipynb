## 6. Módulos

### 6.1 ¿Qué es un Módulo?

Un **módulo** es un archivo que contiene definiciones y declaraciones de Python. Es como una caja de herramientas: guardas funciones, clases y variables relacionadas en un archivo y las reutilizas cuando las necesitas.

#### Ventajas de usar módulos:
- **Organización**: Código organizado por funcionalidad
- **Reutilización**: Escribe una vez, usa muchas veces
- **Mantenibilidad**: Fácil de mantener y actualizar
- **Namespace**: Evita conflictos de nombres
- **Colaboración**: Múltiples personas pueden trabajar en diferentes módulos

### 6.2 Formas de Importar Módulos

Python ofrece varias formas de importar módulos, cada una con sus ventajas:

### 6.3 Crear Módulos Propios

Para crear un módulo, simplemente guarda código Python en un archivo `.py`. Luego puedes importarlo.

```python
# Ejemplo: Crear un módulo de funciones financieras
# Guarda esto en un archivo llamado: finanzas.py

# Contenido del archivo finanzas.py:
codigo_modulo = '''
def calcular_interes_simple(principal, tasa, tiempo):
    """Calcula interés simple."""
    return principal * tasa * tiempo

def calcular_interes_compuesto(principal, tasa, tiempo):
    """Calcula interés compuesto."""
    return principal * (1 + tasa) ** tiempo - principal

def calcular_valor_presente(valor_futuro, tasa, periodos):
    """Calcula valor presente."""
    return valor_futuro / ((1 + tasa) ** periodos)

# Constantes del módulo
TAXA_REFERENCIA = 0.05  # 5%
'''

# Crear el archivo (solo para demostración)
with open("finanzas.py", "w", encoding="utf-8") as f:
    f.write(codigo_modulo)

print("Módulo finanzas.py creado")
print("\nPara usarlo en otro archivo:")
print("  import finanzas")
print("  resultado = finanzas.calcular_interes_simple(1000, 0.05, 2)")
print("\nO:")
print("  from finanzas import calcular_interes_simple")
print("  resultado = calcular_interes_simple(1000, 0.05, 2)")
```

```python
# Importar el módulo que acabamos de crear
import finanzas

# Usar las funciones del módulo
interes = finanzas.calcular_interes_simple(1000, 0.05, 2)
print(f"Interés simple: ${interes:,.2f}")

interes_comp = finanzas.calcular_interes_compuesto(1000, 0.05, 2)
print(f"Interés compuesto: ${interes_comp:,.2f}")

vp = finanzas.calcular_valor_presente(5000, 0.08, 3)
print(f"Valor presente: ${vp:,.2f}")

# Acceder a constantes del módulo
print(f"\nTasa de referencia del módulo: {finanzas.TAXA_REFERENCIA*100}%")
```

### 6.5 El Atributo __name__

El atributo `__name__` es especial en Python. Te permite hacer que un módulo se comporte diferente cuando se ejecuta directamente vs cuando se importa.

```python
# __name__ en el módulo actual
print(f"__name__ en este notebook: {__name__}")

# Cuando un archivo se ejecuta directamente, __name__ == '__main__'
# Cuando se importa, __name__ == nombre_del_modulo

# Ejemplo de uso en un módulo:
codigo_ejemplo = '''
def funcion_util():
    return "Función útil"

if __name__ == '__main__':
    # Este código solo se ejecuta si el archivo se ejecuta directamente
    print("Ejecutando como script principal")
    print(funcion_util())
else:
    # Este código se ejecuta cuando se importa
    print("Módulo importado")
'''

print("\nEsto permite que un módulo sea tanto importable como ejecutable")
```

## 7. Entrada y Salida

### 7.1 Formateo de Salida

Python ofrece varias formas de formatear la salida. Las más importantes son f-strings (recomendado) y el método format().

```python
# f-strings (Python 3.6+) - RECOMENDADO
nombre = "María"
edad = 30
saldo = 5000.50

# Básico
mensaje = f"{nombre} tiene {edad} años"
print(mensaje)

# Con formato de números
mensaje2 = f"Saldo: ${saldo:,.2f}"
print(mensaje2)

# Con expresiones
mensaje3 = f"En 10 años tendrá {edad + 10} años"
print(mensaje3)

# Con alineación
print(f"\nAlineación:")
print(f"'{nombre:>10}'" )  # Derecha (10 caracteres)
print(f"'{nombre:<10}'" )  # Izquierda
print(f"'{nombre:^10}'" )  # Centro

# Con formato de números
print(f"\nFormato de números:")
print(f"Entero con ceros: {edad:05d}")  # 5 dígitos con ceros
print(f"Decimal: {saldo:,.2f}")  # Con comas y 2 decimales
print(f"Porcentaje: {0.25:.2%}")  # Porcentaje
print(f"Notación científica: {1234567:.2e}")  # Científica
```

```python
# format() method - Alternativa a f-strings
nombre = "María"
edad = 30
saldo = 5000.50

# Con nombres
mensaje = "{} tiene {} años y un saldo de ${:,.2f}".format(nombre, edad, saldo)
print(mensaje)

# Con índices
mensaje2 = "{0} tiene {1} años. {0} es mayor de edad.".format(nombre, edad)
print(mensaje2)

# Con nombres de parámetros
mensaje3 = "{nombre} tiene {edad} años".format(nombre=nombre, edad=edad)
print(mensaje3)

# Con formato avanzado
print(f"\nCoordenadas: ({10:>5}, {20:>5})")  # Alineación
```

### 7.2 Trabajar con Archivos

#### ¿Qué es un Archivo?

Un **archivo** es una secuencia de bytes almacenada en el disco. Python permite leer y escribir archivos de forma sencilla.

#### Modos de Apertura:

| Modo | Descripción |
|------|-------------|
| `'r'` | Lectura (read) - por defecto |
| `'w'` | Escritura (write) - sobrescribe si existe |
| `'a'` | Anexar (append) - agrega al final |
| `'x'` | Crear - falla si el archivo existe |
| `'b'` | Modo binario (ej: 'rb', 'wb') |
| `'t'` | Modo texto (por defecto) |
| `'+'` | Lectura y escritura (ej: 'r+', 'w+') |

#### Context Manager (with):

Usar `with` es la forma recomendada porque cierra el archivo automáticamente, incluso si hay errores.

```python
# Escribir a un archivo
contenido = """Línea 1: Introducción
Línea 2: Desarrollo
Línea 3: Conclusión"""

with open("ejemplo.txt", "w", encoding="utf-8") as archivo:
    archivo.write(contenido)

print("Archivo escrito exitosamente")
print("\n" + "="*50)

# Leer archivo completo
with open("ejemplo.txt", "r", encoding="utf-8") as archivo:
    contenido_leido = archivo.read()
    print("Contenido completo:")
    print(contenido_leido)

print("\n" + "="*50)

# Leer línea por línea (eficiente para archivos grandes)
with open("ejemplo.txt", "r", encoding="utf-8") as archivo:
    print("Línea por línea:")
    for i, linea in enumerate(archivo, 1):
        print(f"{i}: {linea.strip()}")

print("\n" + "="*50)

# Leer todas las líneas en una lista
with open("ejemplo.txt", "r", encoding="utf-8") as archivo:
    lineas = archivo.readlines()
    print(f"Total de líneas: {len(lineas)}")
    print(f"Líneas: {lineas}")
```

```python
# Modo append (agregar al final)
with open("ejemplo.txt", "a", encoding="utf-8") as archivo:
    archivo.write("\nLínea 4: Agregada después")

print("Línea agregada")
print("\n" + "="*50)

# Leer el archivo actualizado
with open("ejemplo.txt", "r", encoding="utf-8") as archivo:
    print("Archivo actualizado:")
    print(archivo.read())

print("\n" + "="*50)

# Modo lectura y escritura (r+)
with open("ejemplo.txt", "r+", encoding="utf-8") as archivo:
    contenido = archivo.read()
    archivo.seek(0)  # Volver al inicio
    archivo.write("NUEVA PRIMERA LÍNEA\n" + contenido)

print("Archivo modificado con r+")
```

### 7.3 JSON (JavaScript Object Notation)

#### ¿Qué es JSON?

**JSON** es un formato de texto para intercambiar datos. Es muy popular porque es legible por humanos y fácil de procesar por máquinas. Python tiene soporte nativo para JSON.

#### Características:
- Formato de texto plano
- Fácil de leer y escribir
- Compatible con muchos lenguajes
- Ideal para APIs y configuración
- Soporta: objetos (diccionarios), arrays (listas), strings, números, booleanos, null

```python
import json

# Datos a guardar en JSON
datos_financieros = {
    "cliente": "Juan Pérez",
    "saldo": 5000.50,
    "activo": True,
    "transacciones": [
        {"fecha": "2024-01-15", "tipo": "Ingreso", "monto": 1000},
        {"fecha": "2024-01-20", "tipo": "Egreso", "monto": -500},
        {"fecha": "2024-01-25", "tipo": "Ingreso", "monto": 200}
    ],
    "configuracion": {
        "moneda": "USD",
        "formato_fecha": "YYYY-MM-DD"
    }
}

# Escribir JSON a archivo
with open("datos.json", "w", encoding="utf-8") as archivo:
    json.dump(datos_financieros, archivo, indent=2, ensure_ascii=False)

print("JSON guardado exitosamente")
print("\n" + "="*50)

# Leer JSON desde archivo
with open("datos.json", "r", encoding="utf-8") as archivo:
    datos_cargados = json.load(archivo)
    print("Datos cargados:")
    print(json.dumps(datos_cargados, indent=2, ensure_ascii=False))

print("\n" + "="*50)

# Convertir objeto Python a string JSON
json_string = json.dumps(datos_financieros, indent=2)
print("JSON como string:")
print(json_string)

print("\n" + "="*50)

# Convertir string JSON a objeto Python
datos_desde_string = json.loads(json_string)
print(f"Tipo: {type(datos_desde_string)}")
print(f"Cliente: {datos_desde_string['cliente']}")
```

### 7.4 Manejo de Errores con Archivos

Es importante manejar errores al trabajar con archivos:

```python
# Manejo seguro de archivos
def leer_archivo_seguro(nombre_archivo):
    """Lee un archivo de forma segura con manejo de errores."""
    try:
        with open(nombre_archivo, "r", encoding="utf-8") as archivo:
            return archivo.read()
    except FileNotFoundError:
        print(f"Error: El archivo '{nombre_archivo}' no existe")
        return None
    except PermissionError:
        print(f"Error: No tienes permiso para leer '{nombre_archivo}'")
        return None
    except Exception as e:
        print(f"Error inesperado: {e}")
        return None

# Probar con archivo que existe
contenido = leer_archivo_seguro("ejemplo.txt")
if contenido:
    print(f"Archivo leído: {len(contenido)} caracteres")

# Probar con archivo que no existe
contenido2 = leer_archivo_seguro("archivo_inexistente.txt")
print(f"Resultado: {contenido2}")
```

## 8. Funciones Avanzadas

### 8.1 Decoradores

#### ¿Qué es un Decorador?

Un **decorador** es una función que toma otra función y extiende su comportamiento sin modificar su código. Es como un "envoltorio" que agrega funcionalidad.

#### Conceptos Clave:
- **Funciones como objetos**: En Python, las funciones son objetos de primera clase
- **Closures**: Funciones que recuerdan variables del ámbito exterior
- **Sintaxis @**: Azúcar sintáctico para aplicar decoradores

```python
# Decorador simple para medir tiempo
def medir_tiempo(func):
    """Decorador que mide el tiempo de ejecución de una función."""
    def wrapper(*args, **kwargs):
        import time
        inicio = time.time()
        resultado = func(*args, **kwargs)
        fin = time.time()
        print(f"{func.__name__} tardó {fin - inicio:.4f} segundos")
        return resultado
    return wrapper

# Usar decorador con @
@medir_tiempo
def calcular_cuadrados(n):
    """Calcula cuadrados del 0 al n-1."""
    return [x**2 for x in range(n)]

# Al llamar la función, automáticamente mide el tiempo
resultado = calcular_cuadrados(1000000)
print(f"Primeros 10 cuadrados: {resultado[:10]}")

print("\n" + "="*50)

# Equivalente sin @ (forma explícita)
def otra_funcion(n):
    return sum(range(n))

otra_funcion = medir_tiempo(otra_funcion)  # Aplicar decorador manualmente
resultado2 = otra_funcion(1000000)
```

```python
# Decorador con parámetros
def repetir(veces):
    """Decorador que repite la ejecución de una función."""
    def decorador(func):
        def wrapper(*args, **kwargs):
            resultados = []
            for i in range(veces):
                resultado = func(*args, **kwargs)
                resultados.append(resultado)
            return resultados
        return wrapper
    return decorador

@repetir(3)
def saludar(nombre):
    return f"Hola, {nombre}!"

resultados = saludar("Python")
print(f"Resultados: {resultados}")
```

### 8.2 Generadores

#### ¿Qué es un Generador?

Un **generador** es una función que produce valores uno por uno usando `yield` en lugar de `return`. Es más eficiente en memoria porque no crea toda la lista de una vez.

#### Ventajas:
- **Eficiencia en memoria**: Solo genera valores cuando se necesitan
- **Lazy evaluation**: Evalúa bajo demanda
- **Útil para secuencias grandes**: No carga todo en memoria
- **Puede ser infinito**: Puede generar valores indefinidamente

```python
# Generador simple
def numeros_pares(n):
    """Genera números pares hasta n."""
    for i in range(n):
        if i % 2 == 0:
            yield i  # Produce valor y pausa

# Usar generador
print("Números pares del 0 al 9:")
for num in numeros_pares(10):
    print(num, end=" ")
print()

print("\n" + "="*50)

# Convertir a lista
lista_pares = list(numeros_pares(10))
print(f"Como lista: {lista_pares}")

print("\n" + "="*50)

# Generador infinito
def numeros_fibonacci():
    """Genera la secuencia de Fibonacci infinitamente."""
    a, b = 0, 1
    while True:
        yield a
        a, b = b, a + b

# Usar generador infinito (con límite)
fib = numeros_fibonacci()
print("Primeros 10 números de Fibonacci:")
for i in range(10):
    print(next(fib), end=" ")
print()
```

```python
# Expresión generadora (generator expression)
# Similar a comprensión de listas, pero con paréntesis
cuadrados_gen = (x**2 for x in range(10))
print(f"Tipo: {type(cuadrados_gen)}")
print(f"Valores: {list(cuadrados_gen)}")

print("\n" + "="*50)

# Ventaja: usa menos memoria
import sys

# Lista (ocupa más memoria)
lista_cuadrados = [x**2 for x in range(1000000)]
print(f"Tamaño de lista: {sys.getsizeof(lista_cuadrados)} bytes")

# Generador (ocupa menos memoria)
gen_cuadrados = (x**2 for x in range(1000000))
print(f"Tamaño de generador: {sys.getsizeof(gen_cuadrados)} bytes")

print("\n" + "="*50)

# Usar generador en funciones que aceptan iterables
suma_cuadrados = sum(x**2 for x in range(10))
print(f"Suma de cuadrados: {suma_cuadrados}")
```
