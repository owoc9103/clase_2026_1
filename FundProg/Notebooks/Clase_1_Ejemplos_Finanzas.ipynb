## Aplicando conceptos básicos de Python a problemas financieros

En este notebook veremos cómo aplicar los conceptos aprendidos (números, strings, listas) a situaciones financieras reales.

## 1. Cálculos Financieros Básicos

### 1.1 Interés Simple

**Fórmula:** I = P × r × t

Donde:
- P = Principal (capital inicial)
- r = Tasa de interés (decimal)
- t = Tiempo (años)

```python
# Ejemplo 1: Calcular interés simple
principal = 10000  # $10,000
tasa_interes = 0.05  # 5% anual
tiempo = 2  # 2 años

interes = principal * tasa_interes * tiempo
monto_final = principal + interes

print(f"Capital inicial: ${principal:,.2f}")
print(f"Tasa de interés: {tasa_interes * 100}% anual")
print(f"Tiempo: {tiempo} años")
print(f"Interés ganado: ${interes:,.2f}")
print(f"Monto final: ${monto_final:,.2f}")
```

### 1.2 Valor Presente (VP)

**Fórmula:** VP = VF / (1 + r)^t

```python
# Ejemplo 2: Calcular valor presente
valor_futuro = 5000  # $5,000 en el futuro
tasa_descuento = 0.08  # 8% anual
periodos = 3  # 3 años

valor_presente = valor_futuro / ((1 + tasa_descuento) ** periodos)

print(f"Valor futuro: ${valor_futuro:,.2f}")
print(f"Tasa de descuento: {tasa_descuento * 100}% anual")
print(f"Periodos: {periodos} años")
print(f"Valor presente: ${valor_presente:,.2f}")
```

## 2. Trabajando con Strings en Contexto Financiero

### 2.1 Formateo de Reportes Financieros

```python
# Ejemplo 3: Crear un reporte financiero básico
nombre_cliente = "Juan Pérez"
numero_cuenta = "ACC-12345"
saldo_inicial = 5000.50
deposito = 1500.25
saldo_final = saldo_inicial + deposito

# Crear reporte usando strings
reporte = f"""
=== EXTRACTO BANCARIO ===
Cliente: {nombre_cliente}
Cuenta: {numero_cuenta}
Saldo inicial: ${saldo_inicial:,.2f}
Depósito: ${deposito:,.2f}
Saldo final: ${saldo_final:,.2f}
=======================
"""

print(reporte)
```

```python
# Ejemplo 4: Formatear números como moneda
cantidad = 1234567.89

# Diferentes formatos
print(f"Sin formato: {cantidad}")
print(f"Con comas: ${cantidad:,.2f}")
print(f"Sin decimales: ${cantidad:,.0f}")
print(f"Con separador de miles: {cantidad:,.2f}")
```

## 3. Usando Listas para Datos Financieros

### 3.1 Registro de Transacciones

```python
# Ejemplo 5: Registrar transacciones en una lista
transacciones = []

# Agregar transacciones
transacciones.append(1000.00)  # Depósito
transacciones.append(-250.50)  # Retiro
transacciones.append(500.00)   # Depósito
transacciones.append(-100.00)  # Retiro

print("Transacciones:", transacciones)
print(f"Total de transacciones: {len(transacciones)}")
print(f"Suma total: ${sum(transacciones):,.2f}")
```

```python
# Ejemplo 6: Calcular saldo acumulado
saldo_inicial = 5000
transacciones = [1000, -250, 500, -100, 2000, -50]

saldo_actual = saldo_inicial
saldos = [saldo_inicial]  # Lista para guardar saldos históricos

for transaccion in transacciones:
    saldo_actual += transaccion
    saldos.append(saldo_actual)

print("Historial de saldos:")
for i, saldo in enumerate(saldos):
    if i == 0:
        print(f"Saldo inicial: ${saldo:,.2f}")
    else:
        print(f"Después de transacción {i}: ${saldo:,.2f}")

print(f"\nSaldo final: ${saldo_actual:,.2f}")
```

### 3.2 Análisis de Precios de Acciones

```python
# Ejemplo 7: Precios de cierre de una acción
precios_cierre = [100.50, 102.30, 98.75, 105.20, 103.10, 107.50]

print("Precios de cierre:")
for i, precio in enumerate(precios_cierre, 1):
    print(f"Día {i}: ${precio:.2f}")

# Calcular estadísticas básicas
precio_min = min(precios_cierre)
precio_max = max(precios_cierre)
precio_promedio = sum(precios_cierre) / len(precios_cierre)

print(f"\nPrecio mínimo: ${precio_min:.2f}")
print(f"Precio máximo: ${precio_max:.2f}")
print(f"Precio promedio: ${precio_promedio:.2f}")
```

```python
# Ejemplo 8: Calcular variaciones diarias
precios = [100.50, 102.30, 98.75, 105.20, 103.10, 107.50]
variaciones = []

for i in range(1, len(precios)):
    cambio = precios[i] - precios[i-1]
    variacion_porcentual = (cambio / precios[i-1]) * 100
    variaciones.append(variacion_porcentual)

print("Variaciones porcentuales diarias:")
for i, var in enumerate(variaciones, 1):
    signo = "+" if var >= 0 else ""
    print(f"Día {i} a {i+1}: {signo}{var:.2f}%")
```

## 4. Ejercicios Integrados

### 4.1 Calculadora de Retorno de Inversión

```python
# Ejemplo 9: Calcular retorno de inversión (ROI)
# ROI = ((Valor Final - Valor Inicial) / Valor Inicial) × 100

inversion_inicial = 10000
valor_actual = 12500

roi = ((valor_actual - inversion_inicial) / inversion_inicial) * 100
ganancia = valor_actual - inversion_inicial

reporte_roi = f"""
=== ANÁLISIS DE INVERSIÓN ===
Inversión inicial: ${inversion_inicial:,.2f}
Valor actual: ${valor_actual:,.2f}
Ganancia: ${ganancia:,.2f}
ROI: {roi:.2f}%
===========================
"""

print(reporte_roi)
```
