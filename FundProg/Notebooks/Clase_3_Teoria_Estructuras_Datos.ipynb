## 5. Estructuras de datos

### 5.1 Más sobre listas

#### Usar listas como pilas

```python
# LIFO: Last In, First Out
pila = [3, 4, 5]
pila.append(6)  # Agregar al final
pila.append(7)
print("Pila:", pila)

# Sacar del final
ultimo = pila.pop()
print(f"Elemento sacado: {ultimo}")
print("Pila después de pop:", pila)
```

#### Usar listas como colas

```python
# FIFO: First In, First Out
from collections import deque

cola = deque(["a", "b", "c"])
cola.append("d")  # Agregar al final
print("Cola:", cola)

# Sacar del principio
primero = cola.popleft()
print(f"Elemento sacado: {primero}")
print("Cola después de popleft:", cola)
```

#### Comprensión de listas

```python
# Forma tradicional
cuadrados = []
for x in range(10):
    cuadrados.append(x**2)
print("Tradicional:", cuadrados)

# Con comprensión de listas
cuadrados = [x**2 for x in range(10)]
print("Comprensión:", cuadrados)
```

```python
# Comprensión con condición
pares = [x for x in range(20) if x % 2 == 0]
print("Números pares:", pares)

# Comprensión con transformación y condición
cuadrados_pares = [x**2 for x in range(10) if x % 2 == 0]
print("Cuadrados de pares:", cuadrados_pares)
```

```python
# Comprensión anidada
matriz = [[i*j for j in range(1, 4)] for i in range(1, 4)]
print("Matriz:")
for fila in matriz:
    print(fila)
```

### 5.3 Tuplas y secuencias

```python
# Tuplas: inmutables
tupla = (1, 2, 3)
print("Tupla:", tupla)
print("Tipo:", type(tupla))

# Tupla con un elemento (necesita coma)
tupla_simple = (42,)
print("Tupla simple:", tupla_simple)

# Desempaquetado
x, y, z = tupla
print(f"x={x}, y={y}, z={z}")
```

```python
# Tuplas como claves en diccionarios
coordenadas = {(0, 0): "Origen", (1, 2): "Punto A", (3, 4): "Punto B"}
print("Diccionario con tuplas:", coordenadas)
print("Valor en (1,2):", coordenadas[(1, 2)])
```

### 5.4 Conjuntos

```python
# Crear conjuntos
frutas = {"manzana", "banana", "naranja"}
print("Conjunto:", frutas)

# Agregar elementos
frutas.add("uva")
print("Después de add:", frutas)

# Eliminar elementos
frutas.remove("banana")
print("Después de remove:", frutas)
```

```python
# Operaciones con conjuntos
a = {1, 2, 3, 4, 5}
b = {4, 5, 6, 7, 8}

print(f"A: {a}")
print(f"B: {b}")
print(f"Unión: {a | b}")
print(f"Intersección: {a & b}")
print(f"Diferencia: {a - b}")
print(f"Diferencia simétrica: {a ^ b}")
```

### 5.5 Diccionarios

```python
# Crear diccionarios
telefonos = {"Juan": "123-4567", "María": "234-5678", "Pedro": "345-6789"}
print("Diccionario:", telefonos)

# Acceder a valores
print("Teléfono de Juan:", telefonos["Juan"])

# Agregar/modificar
telefonos["Ana"] = "456-7890"
print("Después de agregar:", telefonos)

# Eliminar
del telefonos["Pedro"]
print("Después de eliminar:", telefonos)
```

```python
# Métodos de diccionarios
persona = {"nombre": "Carlos", "edad": 30, "ciudad": "Bogotá"}

print("Claves:", list(persona.keys()))
print("Valores:", list(persona.values()))
print("Items:", list(persona.items()))

# get() con valor por defecto
print("Edad:", persona.get("edad", "No especificada"))
print("Email:", persona.get("email", "No especificado"))
```

```python
# Comprensión de diccionarios
cuadrados_dict = {x: x**2 for x in range(1, 6)}
print("Diccionario de cuadrados:", cuadrados_dict)

# Con condición
pares_dict = {x: x*2 for x in range(1, 11) if x % 2 == 0}
print("Diccionario de pares:", pares_dict)
```

### 5.6 Técnicas de iteración

```python
# Iterar sobre diccionarios
inventario = {"manzanas": 50, "bananas": 30, "naranjas": 40}

print("Iterar sobre claves:")
for fruta in inventario:
    print(fruta)

print("\nIterar sobre items:")
for fruta, cantidad in inventario.items():
    print(f"{fruta}: {cantidad}")
```

```python
# zip() para iterar sobre múltiples secuencias
nombres = ["Ana", "Luis", "Carmen"]
edades = [25, 30, 28]
ciudades = ["Bogotá", "Medellín", "Cali"]

for nombre, edad, ciudad in zip(nombres, edades, ciudades):
    print(f"{nombre}, {edad} años, de {ciudad}")
```

```python
# enumerate() para obtener índice y valor
meses = ["Enero", "Febrero", "Marzo"]
for indice, mes in enumerate(meses, start=1):
    print(f"Mes {indice}: {mes}")
```

```python
# sorted() para ordenar
numeros = [3, 1, 4, 1, 5, 9, 2, 6]
print("Original:", numeros)
print("Ordenado:", sorted(numeros))
print("Ordenado descendente:", sorted(numeros, reverse=True))

# Ordenar diccionario por valor
ventas = {"Producto A": 150, "Producto B": 200, "Producto C": 120}
ventas_ordenadas = sorted(ventas.items(), key=lambda x: x[1], reverse=True)
print("Ventas ordenadas:", ventas_ordenadas)
```
