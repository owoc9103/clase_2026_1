## 5. Estructuras de Datos

### 5.1 ¿Qué son las Estructuras de Datos?

Las **estructuras de datos** son formas de organizar y almacenar información en la memoria de la computadora. Python ofrece varias estructuras de datos integradas, cada una optimizada para diferentes tipos de operaciones.

### 5.2 Listas Avanzadas

#### Conceptos Fundamentales:

- **Mutabilidad**: Las listas pueden modificarse después de crearse
- **Orden**: Los elementos mantienen el orden de inserción
- **Heterogeneidad**: Pueden contener diferentes tipos de datos
- **Indexación**: Acceso rápido por posición (O(1) en promedio)
- **Dinámicas**: Crecen y se reducen automáticamente

#### Usar Listas como Pilas (LIFO - Last In, First Out)

Una **pila** es una estructura donde el último elemento agregado es el primero en salir. Es como una pila de platos: pones uno encima y sacas el de arriba.

**Operaciones principales:**
- `append()` - Agregar al final (push)
- `pop()` - Sacar del final (pop)

````python
# Implementar una pila
pila = []

# Agregar elementos (push)
pila.append(3)
pila.append(4)
pila.append(5)
print(f"Pila después de agregar: {pila}")

# Sacar elementos (pop) - siempre el último
ultimo = pila.pop()
print(f"Elemento sacado: {ultimo}")
print(f"Pila después de pop: {pila}")

# Ver el último sin sacarlo
if pila:
    print(f"Último elemento (sin sacar): {pila[-1]}")
````

#### Usar Listas como Colas (FIFO - First In, First Out)

Una **cola** es una estructura donde el primer elemento agregado es el primero en salir. Es como una fila: el primero en llegar es el primero en ser atendido.

**Nota**: Usar `collections.deque` es más eficiente para colas que usar listas.

````python
# Implementar cola con deque (eficiente)
from collections import deque

cola = deque(["a", "b", "c"])
print(f"Cola inicial: {cola}")

# Agregar al final (enqueue)
cola.append("d")
cola.append("e")
print(f"Cola después de agregar: {cola}")

# Sacar del principio (dequeue)
primero = cola.popleft()
print(f"Elemento sacado: {primero}")
print(f"Cola después de popleft: {cola}")

# Agregar al principio
cola.appendleft("z")
print(f"Cola después de appendleft: {cola}")

print("\n" + "="*50)

# Cola con lista (menos eficiente, pero funciona)
cola_lista = ["a", "b", "c"]
cola_lista.append("d")  # Agregar al final
primero_lista = cola_lista.pop(0)  # Sacar del principio (lento para listas grandes)
print(f"Cola con lista - Elemento sacado: {primero_lista}")
print(f"Cola restante: {cola_lista}")
````

### 5.3 Comprensión de Listas (List Comprehensions)

#### ¿Qué es la Comprensión de Listas?

La **comprensión de listas** es una forma elegante y concisa de crear listas. Es más rápida y legible que usar bucles tradicionales.

#### Sintaxis:
```python
[expresión for elemento in iterable]
[expresión for elemento in iterable if condición]
[expresión for elemento1 in iterable1 for elemento2 in iterable2]  # Anidada
```

````python
# Forma tradicional vs comprensión
import time

# Forma tradicional (3 líneas)
cuadrados_tradicional = []
for x in range(10):
    cuadrados_tradicional.append(x**2)

# Con comprensión de listas (1 línea)
cuadrados_comprension = [x**2 for x in range(10)]

print(f"Tradicional: {cuadrados_tradicional}")
print(f"Comprensión: {cuadrados_comprension}")
print(f"¿Son iguales? {cuadrados_tradicional == cuadrados_comprension}")
````

````python
# Comprensión con condición
# Solo números pares
pares = [x for x in range(20) if x % 2 == 0]
print(f"Números pares del 0 al 19: {pares}")

# Cuadrados de números pares
cuadrados_pares = [x**2 for x in range(10) if x % 2 == 0]
print(f"Cuadrados de pares: {cuadrados_pares}")

# Números mayores que 5
mayores_5 = [x for x in range(10) if x > 5]
print(f"Números mayores que 5: {mayores_5}")
````

````python
# Comprensión anidada
# Matriz 3x3
matriz = [[i*j for j in range(1, 4)] for i in range(1, 4)]
print("Matriz usando comprensión:")
for fila in matriz:
    print(fila)

print("\n" + "="*40)

# Tabla de multiplicar
tabla_multiplicar = [[f"{i}x{j}={i*j}" for j in range(1, 6)] for i in range(1, 6)]
print("Tabla de multiplicar:")
for fila in tabla_multiplicar:
    print(fila)

print("\n" + "="*40)

# Aplanar lista de listas
lista_anidada = [[1, 2, 3], [4, 5], [6, 7, 8, 9]]
lista_plana = [elemento for sublista in lista_anidada for elemento in sublista]
print(f"Lista anidada: {lista_anidada}")
print(f"Lista plana: {lista_plana}")
````

### 5.4 Tuplas y Secuencias

#### ¿Qué es una Tupla?

Una **tupla** es una secuencia **inmutable** de elementos. Una vez creada, no se puede modificar. Es como una lista congelada.

#### Características:
- **Inmutable**: No se puede modificar después de crearse
- **Ordenada**: Mantiene el orden de los elementos
- **Indexable**: Acceso por índice
- **Puede contener duplicados**
- **Heterogénea**: Puede contener diferentes tipos

#### ¿Cuándo usar tuplas?
- Cuando los datos no deben cambiar
- Como claves en diccionarios (porque son inmutables)
- Para retornar múltiples valores de una función
- Para desempaquetar valores

````python
# Crear tuplas
tupla1 = (1, 2, 3)
tupla2 = 1, 2, 3  # Los paréntesis son opcionales
tupla3 = tuple([1, 2, 3])  # Desde una lista

print(f"tupla1: {tupla1}, tipo: {type(tupla1)}")
print(f"tupla2: {tupla2}")
print(f"tupla3: {tupla3}")

# Tupla con un elemento (necesita coma)
tupla_simple = (42,)  # Con coma
no_tupla = (42)  # Sin coma, es solo un número
print(f"\nTupla simple: {tupla_simple}, tipo: {type(tupla_simple)}")
print(f"No tupla: {no_tupla}, tipo: {type(no_tupla)}")

# Tupla vacía
tupla_vacia = ()
print(f"\nTupla vacía: {tupla_vacia}")

# Acceso a elementos (igual que listas)
print(f"\nPrimer elemento: {tupla1[0]}")
print(f"Último elemento: {tupla1[-1]}")
print(f"Slicing: {tupla1[1:3]}")
````

````python
# Desempaquetado de tuplas
coordenadas = (10, 20)
x, y = coordenadas  # Desempaquetado
print(f"Coordenadas: {coordenadas}")
print(f"x = {x}, y = {y}")

# Intercambio de variables (muy Pythonic)
a, b = 5, 10
print(f"\nAntes: a={a}, b={b}")
a, b = b, a  # Intercambio usando tuplas
print(f"Después: a={a}, b={b}")

# Desempaquetado con * (resto de elementos)
numeros = (1, 2, 3, 4, 5)
primero, *medio, ultimo = numeros
print(f"\nPrimero: {primero}, Medio: {medio}, Último: {ultimo}")

# Retornar múltiples valores (en realidad retorna una tupla)
def dividir_con_resto(dividendo, divisor):
    cociente = dividendo // divisor
    resto = dividendo % divisor
    return cociente, resto  # Retorna tupla

coc, res = dividir_con_resto(17, 5)
print(f"\n17 ÷ 5 = {coc} con resto {res}")
````

````python
# Tuplas como claves en diccionarios
# Las tuplas son inmutables, por lo que pueden ser claves
coordenadas_puntos = {
    (0, 0): "Origen",
    (1, 2): "Punto A",
    (3, 4): "Punto B",
    (-1, -1): "Punto C"
}

print("Diccionario con tuplas como claves:")
for coord, nombre in coordenadas_puntos.items():
    print(f"  {coord}: {nombre}")

print(f"\nValor en (1,2): {coordenadas_puntos[(1, 2)]}")

# Las listas NO pueden ser claves (son mutables)
# coordenadas_puntos[[1, 2]] = "Error"  # Esto daría error
````

### 5.5 Conjuntos (Sets)

#### ¿Qué es un Conjunto?

Un **conjunto** es una colección **desordenada** de elementos **únicos** (sin duplicados). Está basado en la teoría matemática de conjuntos.

#### Características:
- **Sin duplicados**: Cada elemento aparece solo una vez
- **Desordenado**: No mantiene orden de inserción (aunque en Python 3.7+ mantiene orden)
- **Mutable**: Puedes agregar y eliminar elementos
- **No indexable**: No puedes acceder por índice
- **Muy rápido** para verificar pertenencia (O(1) en promedio)

````python
# Crear conjuntos
conjunto1 = {1, 2, 3, 4, 5}
conjunto2 = set([1, 2, 3, 4, 5])  # Desde lista
conjunto3 = set("Python")  # Desde string (cada carácter único)

print(f"Conjunto 1: {conjunto1}")
print(f"Conjunto 2: {conjunto2}")
print(f"Conjunto 3 (de string): {conjunto3}")

# Los duplicados se eliminan automáticamente
conjunto_duplicados = {1, 2, 2, 3, 3, 3, 4}
print(f"\nConjunto con duplicados: {conjunto_duplicados}")

# Conjunto vacío (usar set(), no {})
conjunto_vacio = set()
print(f"Conjunto vacío: {conjunto_vacio}, tipo: {type(conjunto_vacio)}")
# {} es un diccionario vacío, no un conjunto vacío
````

#### Métodos de Conjuntos (Completos)

````python
a = {1, 2, 3, 4, 5}
b = {4, 5, 6, 7, 8}

print(f"Conjunto A: {a}")
print(f"Conjunto B: {b}")

# Agregar elementos
a.add(6)
print(f"\nA después de add(6): {a}")

# Agregar múltiples elementos
a.update([7, 8, 9])
print(f"A después de update([7,8,9]): {a}")

# Eliminar elementos
a.remove(9)  # Error si no existe
print(f"A después de remove(9): {a}")

a.discard(10)  # No da error si no existe
print(f"A después de discard(10): {a}")

elemento = a.pop()  # Elimina y retorna elemento aleatorio
print(f"\nElemento eliminado con pop(): {elemento}")
print(f"A después de pop(): {a}")

# Operaciones de conjunto
print(f"\n{'='*50}")
print(f"A: {a}")
print(f"B: {b}")
print(f"Unión (A | B): {a | b}")
print(f"Intersección (A & B): {a & b}")
print(f"Diferencia (A - B): {a - b}")
print(f"Diferencia simétrica (A ^ B): {a ^ b}")

# Métodos equivalentes
print(f"\nUnión con método: {a.union(b)}")
print(f"Intersección con método: {a.intersection(b)}")
print(f"Diferencia con método: {a.difference(b)}")
print(f"Diferencia simétrica con método: {a.symmetric_difference(b)}")

# Verificar relaciones
c = {1, 2, 3}
print(f"\n{'='*50}")
print(f"C: {c}")
print(f"C es subconjunto de A: {c.issubset(a)}")
print(f"A es superconjunto de C: {a.issuperset(c)}")
print(f"A y B son disjuntos: {a.isdisjoint(b)}")

# Verificar pertenencia
print(f"\n3 in A: {3 in a}")
print(f"10 in A: {10 in a}")

# Limpiar conjunto
a.clear()
print(f"\nA después de clear(): {a}")
````

### 5.6 Diccionarios

#### ¿Qué es un Diccionario?

Un **diccionario** es una colección de pares **clave-valor**. Es como un diccionario real: buscas una palabra (clave) y encuentras su significado (valor).

#### Características:
- **Clave-valor**: Cada elemento tiene una clave única y un valor asociado
- **Mutable**: Puedes modificar, agregar y eliminar elementos
- **Desordenado** (antes de Python 3.7), **ordenado** (desde Python 3.7+)
- **Claves inmutables**: Las claves deben ser tipos inmutables (strings, números, tuplas)
- **Acceso rápido**: O(1) en promedio para buscar, insertar y eliminar
- **Muy eficiente** para mapear relaciones

````python
# Crear diccionarios de diferentes formas
dic1 = {"nombre": "Juan", "edad": 30}
dic2 = dict(nombre="María", edad=25)  # Con constructor
dic3 = dict([("nombre", "Pedro"), ("edad", 28)])  # Desde lista de tuplas
dic4 = {x: x**2 for x in range(5)}  # Comprensión de diccionario

print(f"Diccionario 1: {dic1}")
print(f"Diccionario 2: {dic2}")
print(f"Diccionario 3: {dic3}")
print(f"Diccionario 4 (comprensión): {dic4}")

# Diccionario vacío
dic_vacio = {}
print(f"\nDiccionario vacío: {dic_vacio}")
````

#### Métodos de Diccionarios (Completos - Parte 1: Acceso)

````python
persona = {"nombre": "Carlos", "edad": 30, "ciudad": "Bogotá"}

# Acceso básico
print(f"Nombre: {persona['nombre']}")  # Con []
print(f"Edad: {persona.get('edad')}")  # Con get()

# get() con valor por defecto (no lanza error si no existe)
email = persona.get("email", "No especificado")
print(f"Email: {email}")

# Obtener todas las claves
claves = list(persona.keys())
print(f"\nClaves: {claves}")

# Obtener todos los valores
valores = list(persona.values())
print(f"Valores: {valores}")

# Obtener pares clave-valor
items = list(persona.items())
print(f"Items (pares): {items}")

# Iterar sobre diccionario
print(f"\nIteración:")
for clave in persona:  # Itera sobre claves
    print(f"  {clave}: {persona[clave]}")

print()
for clave, valor in persona.items():  # Itera sobre items
    print(f"  {clave}: {valor}")
````

#### Métodos de Diccionarios (Completos - Parte 2: Modificación)

````python
persona = {"nombre": "Carlos", "edad": 30}

# Agregar/modificar elementos
persona["ciudad"] = "Bogotá"  # Agregar
persona["edad"] = 31  # Modificar
print(f"Después de agregar/modificar: {persona}")

# update() - Actualizar con otro diccionario
persona.update({"profesion": "Ingeniero", "edad": 32})
print(f"Después de update(): {persona}")

# setdefault() - Obtener valor o establecer si no existe
ciudad = persona.setdefault("ciudad", "Desconocida")
telefono = persona.setdefault("telefono", "000-0000")
print(f"\nCiudad (ya existía): {ciudad}")
print(f"Teléfono (nuevo): {telefono}")
print(f"Diccionario: {persona}")

# pop() - Eliminar y retornar valor
edad_eliminada = persona.pop("edad")
print(f"\nEdad eliminada: {edad_eliminada}")
print(f"Diccionario después de pop('edad'): {persona}")

# popitem() - Eliminar y retornar último par (Python 3.7+)
ultimo_par = persona.popitem()
print(f"\nÚltimo par eliminado: {ultimo_par}")
print(f"Diccionario después de popitem(): {persona}")

# del - Eliminar elemento
del persona["ciudad"]
print(f"\nDespués de del persona['ciudad']: {persona}")

# clear() - Vaciar diccionario
persona.clear()
print(f"Después de clear(): {persona}")
````

#### Comprensión de Diccionarios

````python
# Comprensión básica
cuadrados_dict = {x: x**2 for x in range(1, 6)}
print(f"Diccionario de cuadrados: {cuadrados_dict}")

# Con condición
pares_dict = {x: x*2 for x in range(1, 11) if x % 2 == 0}
print(f"Diccionario de pares: {pares_dict}")

# Invertir clave-valor
original = {"a": 1, "b": 2, "c": 3}
invertido = {valor: clave for clave, valor in original.items()}
print(f"\nOriginal: {original}")
print(f"Invertido: {invertido}")

# Filtrar diccionario
estudiantes = {"Juan": 85, "María": 92, "Pedro": 78, "Ana": 95}
aprobados = {nombre: nota for nombre, nota in estudiantes.items() if nota >= 80}
print(f"\nEstudiantes: {estudiantes}")
print(f"Aprobados (>=80): {aprobados}")
````

### 5.7 Técnicas de Iteración Avanzadas

#### zip() - Combinar Múltiples Secuencias

`zip()` combina elementos de múltiples iterables en pares (o tuplas). Es como abrochar una cremallera.

````python
# zip() básico
nombres = ["Ana", "Luis", "Carmen"]
edades = [25, 30, 28]
ciudades = ["Bogotá", "Medellín", "Cali"]

# Combinar en pares
for nombre, edad in zip(nombres, edades):
    print(f"{nombre}: {edad} años")

print("\n" + "="*40)

# Combinar múltiples secuencias
for nombre, edad, ciudad in zip(nombres, edades, ciudades):
    print(f"{nombre}, {edad} años, de {ciudad}")

print("\n" + "="*40)

# Convertir a lista de tuplas
combinado = list(zip(nombres, edades, ciudades))
print(f"Lista de tuplas: {combinado}")

# Desempaquetar zip
nombres2, edades2, ciudades2 = zip(*combinado)
print(f"\nDesempaquetado:")
print(f"Nombres: {nombres2}")
print(f"Edades: {edades2}")
print(f"Ciudades: {ciudades2}")

# zip() con secuencias de diferente longitud (se detiene en la más corta)
print("\n" + "="*40)
cortos = [1, 2, 3]
largos = [10, 20, 30, 40, 50]
for a, b in zip(cortos, largos):
    print(f"{a}, {b}")
````

````python
# enumerate() - Índice y valor
meses = ["Enero", "Febrero", "Marzo", "Abril"]

# enumerate básico (índice empieza en 0)
for indice, mes in enumerate(meses):
    print(f"Índice {indice}: {mes}")

print("\n" + "="*40)

# enumerate con start personalizado
for indice, mes in enumerate(meses, start=1):
    print(f"Mes {indice}: {mes}")

print("\n" + "="*40)

# Convertir a diccionario
meses_dict = {indice: mes for indice, mes in enumerate(meses, start=1)}
print(f"Diccionario de meses: {meses_dict}")
````

````python
# sorted() - Ordenar
numeros = [3, 1, 4, 1, 5, 9, 2, 6]

print(f"Original: {numeros}")
print(f"Ordenado: {sorted(numeros)}")
print(f"Ordenado descendente: {sorted(numeros, reverse=True)}")
print(f"Original sin cambios: {numeros}")  # sorted() no modifica la lista original

print("\n" + "="*40)

# Ordenar strings
palabras = ["python", "java", "c++", "javascript"]
print(f"Palabras ordenadas alfabéticamente: {sorted(palabras)}")
print(f"Palabras ordenadas por longitud: {sorted(palabras, key=len)}")

print("\n" + "="*40)

# Ordenar diccionario por valor
ventas = {"Producto A": 150, "Producto B": 200, "Producto C": 120, "Producto D": 180}
ventas_ordenadas = sorted(ventas.items(), key=lambda x: x[1], reverse=True)
print("Ventas ordenadas por valor (descendente):")
for producto, cantidad in ventas_ordenadas:
    print(f"  {producto}: {cantidad}")

print("\n" + "="*40)

# Ordenar lista de diccionarios
estudiantes = [
    {"nombre": "Juan", "nota": 85},
    {"nombre": "María", "nota": 92},
    {"nombre": "Pedro", "nota": 78}
]
estudiantes_ordenados = sorted(estudiantes, key=lambda x: x["nota"], reverse=True)
print("Estudiantes ordenados por nota:")
for est in estudiantes_ordenados:
    print(f"  {est['nombre']}: {est['nota']}")
````
