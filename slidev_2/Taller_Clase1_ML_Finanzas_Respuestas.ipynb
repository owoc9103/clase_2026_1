## Bloque 1: Historia del Machine Learning (timeline de la clase)

### 1.1 Orden de los hitos

En la presentación se recorre esta secuencia: **1950** Test de Turing → **1956** Conferencia de Dartmouth → **1986** Backpropagation → **1997** Deep Blue vs Kasparov → **2006** Renacimiento del Deep Learning → **2012** AlexNet → **2016** AlphaGo.

**Pregunta:** ¿Por qué tiene sentido este orden histórico? Elige **dos** hitos consecutivos y explica en una frase cómo el primero permitió o motivó el segundo.

**Respuesta modelo:**

- **Backpropagation (1986) → Deep Blue (1997):** El backpropagation permitió entrenar redes neuronales multicapa; aunque Deep Blue no se basaba principalmente en redes neuronales sino en búsqueda y evaluación, el avance en algoritmos de aprendizaje sentó las bases para que más adelante se pudieran combinar búsqueda y aprendizaje (como en AlphaGo).

- **Deep Learning (2006) → AlexNet (2012):** Sin el renacimiento del Deep Learning y las técnicas para entrenar redes profundas (y el uso de GPUs), no habría sido posible entrenar una red tan grande como AlexNet sobre millones de imágenes; el avance de 2006 en representaciones jerárquicas y entrenamiento profundo fue condición necesaria para el éxito de AlexNet en ImageNet.

### 1.2 Test de Turing (1950)

En clase se citó a Turing y el **concepto clave**: si una máquina puede engañar a un humano haciéndole creer que es otro humano, entonces puede considerarse "inteligente".

**Pregunta:** ¿Crees que "engañar" es un criterio adecuado para definir inteligencia? ¿Qué otros criterios podrías proponer? Si piensas en un *chatbot* que atiende clientes en un banco, ¿pasar el Test de Turing sería suficiente para que el banco lo considere "inteligente" o útil?

**Respuesta modelo:**

«Engañar» es un criterio **operativo** (fácil de evaluar en un juego de imitación) pero **limitado**: no distingue entre «parecer humano» y «entender, razonar o explicar». Otros criterios podrían ser: capacidad de resolver problemas nuevos, de generalizar, de explicar sus decisiones o de adaptarse a reglas cambiantes; cada uno tiene limitaciones (por ejemplo, «explicar» puede ser difícil de medir de forma objetiva).

Para un **chatbot** en un banco, **pasar el Test de Turing no es suficiente**: el banco suele necesitar que el sistema sea fiable, que no dé información incorrecta sobre productos o regulación, que pueda escalar y que, en muchos casos, sea explicable o auditable. Un bot que «engañe» en una conversación corta puede fallar en casos edge o en cumplimiento normativo; por tanto, la utilidad depende de criterios de negocio y regulación, no solo de parecer humano.

### 1.3 Conferencia de Dartmouth (1956)

La **cita relevante** dice: *"Cada aspecto del aprendizaje o cualquier otra característica de la inteligencia puede en principio describirse con tal precisión que una máquina pueda simularla."*

**Pregunta:** ¿Estás de acuerdo? ¿Qué aspectos del aprendizaje humano o de la toma de decisiones en finanzas son más difíciles de describir con precisión para que una máquina los simule? Da al menos un ejemplo en el ámbito financiero.

**Respuesta modelo:**

Parcialmente de acuerdo: muchos aspectos **sí** se pueden describir y simular (clasificación, regresión, patrones en datos), pero otros son **más difíciles o discutibles**: la intuición bajo incertidumbre extrema, el uso de contexto cultural o ético, la creatividad en el diseño de productos, o la confianza y relación con el cliente.

**Ejemplo en finanzas:** La decisión de un oficial de crédito que «hace caso omiso» del score cuando conoce al cliente y su contexto (empresa familiar, crisis puntual) es difícil de formalizar por completo: implica criterio humano, experiencia y hasta valores. Un modelo que solo use variables estructuradas no captura ese «juicio excepcional»; por eso muchas instituciones mantienen un humano en el loop para casos límite.

### 1.4 Backpropagation (1986)

El **concepto clave** es que el backpropagation permite entrenar redes neuronales profundas ajustando los pesos de manera sistemática.

**Pregunta:** ¿Por qué es importante poder "ajustar los pesos de manera sistemática" para aplicaciones en finanzas? Piensa en un modelo de riesgo crediticio o de clasificación de fraude.

**Respuesta modelo:**

Ajustar los pesos de manera sistemática permite que el modelo **aprenda desde datos** la relación entre variables (historial crediticio, ingresos, transacciones, etc.) y la variable objetivo (impago, fraude), sin tener que especificar a mano cada regla. En riesgo crediticio o detección de fraude, hay muchas señales y interacciones; el backpropagation (y en general el entrenamiento por gradiente) permite encontrar una combinación óptima de pesos que minimice el error. Sin muchos datos y sin un algoritmo que ajuste parámetros de forma sistemática, los modelos serían muy limitados o totalmente diseñados por expertos, con poca capacidad de capturar patrones complejos o cambiantes.

### 1.5 Deep Blue (1997) y AlphaGo (2016)

**Concepto clave** Deep Blue: la combinación de poder computacional masivo y algoritmos sofisticados puede superar la intuición humana en dominios específicos. AlphaGo: aprendizaje profundo + búsqueda para problemas intratables.

**Pregunta:** ¿En qué sentido son "inteligentes" y en qué sentido no? ¿Qué puede hacer un jugador humano que estas máquinas no? Relaciona con el Test de Turing y con "superar la intuición humana en dominios específicos".

**Respuesta modelo:**

**En qué sentido son inteligentes:** En el dominio concreto (ajedrez, Go) superan a los mejores humanos: evalúan posiciones, buscan jugadas y optimizan el resultado. Eso es una forma de «inteligencia» operativa en un espacio de reglas bien definidas.

**En qué sentido no:** No explican su jugada en lenguaje natural, no generalizan a otros juegos o dominios sin reentrenar, no «sienten» presión ni adaptan su estilo a la personalidad del rival. Un humano puede dar una clase de ajedrez o de Go; la máquina solo juega.

**Relación con el Test de Turing:** Pasar el Test exigiría engañar en una conversación; Deep Blue y AlphaGo no están diseñados para eso. Ilustran más bien la idea de la clase: **en dominios específicos**, poder computacional y algoritmos pueden superar la intuición humana, pero eso no implica inteligencia general ni capacidad de explicación o generalización.

### 1.6 Deep Learning (2006) y AlexNet (2012) – Representaciones jerárquicas

**Pregunta:** ¿Qué es una "representación jerárquica" de los datos? Da un ejemplo en finanzas con niveles de abstracción (más bajo → más alto).

**Respuesta modelo:**

Una **representación jerárquica** es una organización de la información en **niveles**: desde características simples y locales (bajo nivel) hasta patrones más abstractos y globales (alto nivel). Las capas profundas de una red aprenden progresivamente representaciones más complejas.

**Ejemplo en finanzas:**
- **Nivel bajo:** Precios de cierre diarios, volumen, tick-by-tick.
- **Nivel medio:** Indicadores técnicos (medias móviles, RSI, bandas), volatilidad realizada, correlaciones entre activos.
- **Nivel alto:** Señales de trading (tendencia, reversión, régimen de mercado), riesgo de portafolio, o decisión comprar/vender/mantener.

Otro ejemplo: **transacciones** (bajo) → **patrones de comportamiento** (medio) → **riesgo de fraude o segmento de cliente** (alto).

## Bloque 3: Los Pilares del Éxito

### 3.1 Los tres pilares

**Pregunta:** ¿Qué puede salir mal si solo hay infraestructura potente pero herramientas/algoritmos mal elegidos? ¿Y si hay buenos algoritmos y herramientas pero infraestructura débil?

**Respuesta modelo:**

**Solo infraestructura (datos y cómputo), herramientas/algoritmos mal elegidos:** Se pueden entrenar modelos que no resuelven el problema de negocio (modelo equivocado), que no escalan en producción (herramientas inadecuadas), que son imposibles de mantener o que dan resultados no confiables. Es decir, se gasta mucho en datos y servidores pero el valor entregado es bajo o negativo.

**Buenos algoritmos y herramientas, infraestructura débil (pocos datos, poco cómputo):** Los modelos no pueden entrenarse bien (pocos datos → overfitting o baja capacidad predictiva), no se pueden reentrenar con la frecuencia necesaria, o no se pueden servir en tiempo real. El diseño puede ser correcto pero la ejecución queda limitada por datos y capacidad de cómputo.

### 3.2 Relacionar "¿Por qué ahora?" con los Pilares

**Pregunta:** Relaciona los tres factores (datos, poder computacional, crecimiento del uso de ML) con los tres pilares (infraestructura, toolkit, algoritmos).

**Respuesta modelo:**

- **Grandes volúmenes de datos** → Se reflejan sobre todo en el pilar de **infraestructura** (almacenamiento, pipelines, calidad de datos), pero también en **algoritmos** (modelos que aprovechan esos datos).

- **Poder computacional asequible** → Sobre todo **infraestructura** (cloud, GPUs, escalabilidad) y también **toolkit** (frameworks que usan ese cómputo de forma eficiente).

- **Crecimiento del uso de ML en finanzas** → Refleja que las organizaciones están invirtiendo en los **tres** pilares: infraestructura para datos y cómputo, toolkit (Python, librerías, APIs) y algoritmos (modelos adecuados a cada problema). Es el resultado de que los dos primeros factores estén disponibles y de la demanda de expertos que combinen los tres.

### 3.3 Priorización en la práctica

**Pregunta:** Si tuvieras que priorizar uno de los tres pilares con presupuesto limitado, ¿cuál elegirías primero y en qué orden los otros dos?

**Respuesta modelo:**

Una priorización razonable es:

1. **Infraestructura (datos + capacidad básica de cómputo)** primero: sin datos de calidad y un mínimo de capacidad para entrenar y desplegar, no tiene sentido invertir mucho en algoritmos complejos ni en herramientas avanzadas. Primero hay que tener el dato y poder procesarlo.

2. **Toolkit correcto** en segundo lugar: elegir lenguaje, librerías y pipelines adecuados permite que los modelos se desarrollen, prueben y pongan en producción de forma eficiente; evita rehacer trabajo o elegir tecnologías que no escalen.

3. **Algoritmos correctos** como tercer pilar en prioridad: una vez hay datos e infraestructura y un toolkit que permita iterar, tiene sentido invertir en elegir y afinar el modelo adecuado al problema de negocio. No siempre el modelo más complejo es el mejor; lo importante es que sea el correcto para el problema.

## Bloque 4: Conceptos de ML y aplicación en finanzas

### 4.1 Tipos de aprendizaje y problemas financieros

**Pregunta:** Para detección de fraude, segmentación de clientes, optimización de trading y predicción de precio de cierre, indica tipo de aprendizaje (supervisado, no supervisado, por refuerzo) y por qué.

**Respuesta modelo:**

- **Detección de fraude:** **Supervisado** (clasificación). Se tienen ejemplos etiquetados (transacción fraudulenta / legítima) y se aprende a predecir la etiqueta a partir de características de la transacción y del usuario.

- **Segmentación de clientes por comportamiento:** **No supervisado** (p. ej. clustering). No hay etiquetas «tipo de cliente»; se buscan grupos naturales en los datos (uso de productos, frecuencia, montos) para descubrir segmentos.

- **Optimización de estrategia de trading (cuándo comprar/vender):** **Por refuerzo**. La decisión es secuencial; se recibe una recompensa (beneficio/riesgo) según las acciones tomadas y se aprende la política que maximiza la recompensa a largo plazo, de forma similar a AlphaGo.

- **Predicción del precio de cierre de un activo:** **Supervisado** (regresión o predicción de serie temporal). La variable objetivo es el precio (o el rendimiento); se usan características pasadas (precios, volumen, indicadores) para predecir un valor numérico.

## Bloque 5: Reflexión crítica y debate

### 5.1 Test de Turing y explicabilidad en finanzas

**Pregunta:** ¿Ves conflicto entre "pasar el Test de Turing" y "ser explicable"? ¿Una máquina puede ser útil en finanzas sin ser explicable?

**Respuesta modelo:**

**Conflicto:** Sí, en parte. El Test de Turing solo exige que la máquina **imite** a un humano en una conversación, no que explique su razonamiento. En finanzas, regulación y riesgo suelen exigir **explicabilidad**: poder justificar por qué se denegó un crédito, se marcó una transacción como fraude o se tomó una decisión de inversión. Un sistema que «pase» el Test podría ser una caja negra; eso choca con requisitos de transparencia y auditoría.

**¿Útil sin ser explicable?** Sí, en contextos donde la decisión no es final o donde se acepta un trade-off: por ejemplo, un modelo que sugiere candidatos a fraude para revisión humana, o que se usa en trading donde la regulación no exige explicar cada operación. La utilidad depende del uso: en crédito al consumo o en cumplimiento normativo, la explicabilidad suele ser requisito; en otras áreas puede ser secundaria.

### 5.2 Dartmouth y los Pilares

**Pregunta:** ¿Cómo relacionas la idea de Dartmouth ("describir con precisión para que una máquina simule") con elegir el algoritmo correcto? ¿Qué pasa cuando el problema no está bien definido o los datos no permiten describir bien el fenómeno?

**Respuesta modelo:**

**Relación con "algoritmo correcto":** Elegir el algoritmo correcto es, en la práctica, **elegir el modelo que mejor "simule"** el fenómeno de negocio con los datos disponibles. Dartmouth sugiere que si el problema se puede describir con precisión (objetivo, variables, restricciones), entonces una máquina puede simularlo; el pilar de algoritmos correctos implica justamente **formalizar** el problema (clasificación, regresión, secuencial, etc.) y elegir el modelo que mejor lo represente.

**Cuando el problema no está bien definido o los datos no describen bien el fenómeno:** Si el problema de negocio es vago (p. ej. "mejorar la experiencia del cliente") o los datos son incompletos, sesgados o no representativos, la "descripción" que la máquina aprende será imperfecta. El modelo puede ser óptimo en el papel pero poco útil en la realidad, o incluso contraproducente. Por eso en la clase se insiste en **problema de negocio** bien entendido y en datos de calidad; sin eso, el mejor algoritmo no compensa.

### 5.3 Secreto de estrategias y ventaja competitiva

**Pregunta:** ¿Hasta qué punto el secreto de las estrategias de ML en finanzas es deseable? ¿Qué beneficios y qué riesgos ves?

**Respuesta modelo:**

**Beneficios del secreto:** Ventaja competitiva: si una firma descubre una señal o estrategia rentable, revelarla haría que otros la exploten y la señal se diluya. El incentivo a innovar se mantiene si hay posibilidad de beneficio privado. Además, evita que competidores copien modelos o datos propietarios.

**Riesgos:** (1) **Estabilidad sistémica:** Si muchas firmas usan estrategias similares (aunque secretas) y no son transparentes, un shock puede provocar comportamientos correlacionados (ventas masivas, liquidez). (2) **Regulación y supervisión:** Los reguladores pueden exigir transparencia o límites a ciertos modelos para proteger al mercado y a los inversores. (3) **Desigualdad de acceso:** Quienes tienen datos y modelos propietarios pueden concentrar rentabilidad y excluir a actores más pequeños. (4) **Sesgos y errores:** El secreto dificulta la revisión por pares y la detección de sesgos o fallos. Un equilibrio razonable es transparencia ante reguladores y auditoría interna, sin revelar al público la estrategia concreta.
