# Traduccion al espanol - `retail_sales_dataset.csv`

Este notebook:

1. Carga el archivo de ventas minoristas.
2. Renombra **todas las columnas** al espanol.
3. Traduce el **contenido textual** (genero y categoria de producto). Los identificadores (`CUST...`), fechas y valores numericos se conservan.
4. Opcionalmente exporta el resultado a un nuevo CSV en esta misma carpeta.

## 1. Librerias y ruta del archivo

```python
from pathlib import Path

import pandas as pd

# Busca el CSV: primero en el directorio de trabajo del kernel, luego ruta tipica del repo
_aqui = Path.cwd()
_candidatos = [
    _aqui / "retail_sales_dataset.csv",
    _aqui / "ML_Finanzas" / "Clases" / "Talleres" / "RFM" / "Insumos" / "retail_sales_dataset.csv",
]
ARCHIVO_ENTRADA = next((p for p in _candidatos if p.is_file()), None)
if ARCHIVO_ENTRADA is None:
    raise FileNotFoundError(
        "No se encontro retail_sales_dataset.csv. "
        "Ejecuta el kernel con carpeta de trabajo en Insumos o en la raiz del repositorio."
    )

ARCHIVO_SALIDA = ARCHIVO_ENTRADA.with_name("datos_de_ventas.csv")
print(f"Entrada:  {ARCHIVO_ENTRADA.resolve()}")
print(f"Salida:   {ARCHIVO_SALIDA.resolve()}")
```

## 2. Carga y vista previa (original)

```python
df = pd.read_csv(ARCHIVO_ENTRADA)
print(df.shape)
df.head()
```

## 3. Nombres de columnas en espanol

```python
RENOMBRAR_COLUMNAS = {
    "Transaction ID": "ID_Transaccion",
    "Date": "Fecha",
    "Customer ID": "ID_Cliente",
    "Gender": "Genero",
    "Age": "Edad",
    "Product Category": "Categoria_Producto",
    "Quantity": "Cantidad",
    "Price per Unit": "Precio_Unitario",
    "Total Amount": "Importe_Total",
}

faltan = set(RENOMBRAR_COLUMNAS) - set(df.columns)
if faltan:
    raise ValueError(f"Faltan columnas esperadas en el CSV: {faltan}")

df_es = df.rename(columns=RENOMBRAR_COLUMNAS)
list(df_es.columns)
```

## 4. Traduccion del contenido (categoricas)

Mapeo explicito para evitar dependencias de internet o APIs. Si en el futuro aparecen nuevas categorias en ingles, amplia los diccionarios.

```python
MAPA_GENERO = {
    "Male": "Masculino",
    "Female": "Femenino",
}

MAPA_CATEGORIA = {
    "Beauty": "Belleza",
    "Clothing": "Ropa",
    "Electronics": "Electronica",
}

df_es["Genero"] = df_es["Genero"].map(MAPA_GENERO).fillna(df_es["Genero"])
df_es["Categoria_Producto"] = df_es["Categoria_Producto"].map(MAPA_CATEGORIA).fillna(
    df_es["Categoria_Producto"]
)

print("Valores unicos - Genero:", df_es["Genero"].unique())
print("Valores unicos - Categoria:", df_es["Categoria_Producto"].unique())
```

## 5. Resultado y exportacion

Se guarda `retail_sales_dataset_es.csv` junto al archivo original (UTF-8 con BOM para abrir bien en Excel).

```python
df_es.head(10)
df_es.info()
df_es.to_csv(ARCHIVO_SALIDA, index=False, encoding="utf-8-sig")
print(f"Guardado: {ARCHIVO_SALIDA}")
```
