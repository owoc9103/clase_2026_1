# **Aplicación Modelo RFM**     


El **modelo RFM (Recencia, Frecuencia y Valor Monetario)** es una de las metodologías más utilizadas en el análisis de comportamiento de clientes, debido a su capacidad para sintetizar patrones de compra y facilitar procesos de segmentación con base en evidencia cuantitativa. Este enfoque parte del supuesto de que los clientes que han comprado recientemente, que compran con mayor frecuencia y que generan un mayor valor económico para la organización tienden a ser más relevantes desde la perspectiva comercial y estratégica.

La **recencia** mide cuántos días han pasado desde la última compra del cliente, permitiendo identificar niveles de desactivación o riesgo de abandono. La **frecuencia** cuantifica el número de transacciones realizadas en un período determinado, mostrando el grado de actividad y constancia en el consumo. El **valor monetario**, por su parte, refleja el monto total gastado por cada cliente, lo que posibilita reconocer a los segmentos que aportan mayor ingreso y aquellos con menor contribución económica.

Al combinar estas tres dimensiones, el modelo RFM permite construir perfiles diferenciados de clientes, tales como consumidores altamente valiosos, clientes leales, compradores ocasionales o clientes inactivos. Esta clasificación es útil para diseñar estrategias de retención, fidelización, personalización de ofertas y reactivación, optimizando los esfuerzos de marketing y maximizando el retorno de las campañas.

La implementación del modelo en un entorno computacional como Jupyter Notebook facilita el cálculo, la visualización y la interpretación de los indicadores RFM, integrando herramientas de manipulación de datos en Python con representaciones gráficas que apoyan la toma de decisiones. A través de este proceso, los estudiantes pueden comprender de manera práctica cómo traducir información transaccional en conocimiento accionable para la gestión de clientes.

Además de la clasificación general basada en las tres dimensiones del modelo RFM, la literatura propone un tipo adicional de segmentación que considera únicamente los puntajes asociados a las variables de **Recencia (R)** y **Frecuencia (F)**. Este enfoque se utiliza cuando el interés principal es evaluar el nivel de actividad reciente y la constancia del cliente en sus transacciones, dejando de lado el componente monetario para simplificar el análisis o cuando el monto no es un factor determinante del comportamiento.

Esta clasificación R–F permite identificar patrones relevantes, tales como clientes altamente activos, clientes frecuentes pero con menor actividad reciente, o usuarios que han tenido pocas interacciones y cuya última transacción ocurrió hace un periodo considerable. Al centrarse en estas dos dimensiones, el modelo facilita una lectura más directa sobre el riesgo de abandono y la intensidad de uso del servicio.

En la siguiente visualización se muestran las categorías resultantes de esta segmentación basada exclusivamente en Recencia y Frecuencia, permitiendo comparar los distintos niveles de actividad y reconocer grupos de clientes que podrían requerir estrategias diferenciadas de seguimiento, retención o reactivación.


segmento.jpg

Lo anterior implica que, por ejemplo, un cliente que obtenga una calificación de **5 en Frecuencia** y **5 en Recencia** puede catalogarse como un **cliente Estrella**. Este tipo de cliente se caracteriza por realizar un número elevado de transacciones y haber interactuado recientemente con el punto de venta, lo que evidencia un alto nivel de actividad y compromiso con el servicio.

Por el contrario, aquellos clientes que presentan una calificación intermedia por ejemplo, **3 en Frecuencia** y **3 en Recencia** reflejan un comportamiento menos dinámico. Este segmento corresponde a usuarios que han reducido tanto la intensidad de sus transacciones como la cercanía temporal de su última visita, lo cual podría indicar el inicio de un proceso de desenganche.

En el extremo inferior se encuentran los **clientes perdidos**, caracterizados por puntajes bajos en ambas dimensiones (recencia y frecuencia ≤ 2). Estos clientes no solo han disminuido significativamente su actividad, sino que además llevan un tiempo prolongado sin realizar transacciones, lo que sugiere un abandono del servicio.

Esta subsegmentación basada en Recencia y Frecuencia es especialmente útil porque permite identificar, dentro de categorías amplias como Diamante, Oro, Plata o Bronce, qué clientes requieren una intervención específica: cuáles mantienen un comportamiento sobresaliente (Estrella), cuáles han disminuido su participación (En Riesgo), cuáles representan un potencial de crecimiento (Promesa) y cuáles se consideran definitivamente perdidos.

Asimismo, esta clasificación facilita reconocer clientes nuevos o con un patrón emergente de consumo. En conjunto, esta información ofrece una base sólida para diseñar estrategias comerciales orientadas a:

* **Potenciar el consumo** de los clientes más rentables, promoviendo su participación en otros productos del portafolio.
* **Implementar acciones de retención** para los segmentos en riesgo.
* **Evaluar alternativas de reactivación** para aquellos clientes que han abandonado la actividad.


# **Importante**

Para cada cliente se asignan puntajes independientes en Recencia, Frecuencia y Monto, utilizando una escala de 1 a 5. En el proceso de ponderación, la variable **Monto** adquiere mayor relevancia, ya que representa el 60 % del peso total en la calificación compuesta.

Como ejemplo, en la imagen previa se presentan los casos de **Lucía** y **Luis**. Aunque ambos obtienen resultados similares en algunas dimensiones, Luis alcanza un puntaje de **5 en Monto**, mientras que Lucía obtiene un puntaje de **2**. Debido al mayor peso asignado a esta variable, la clasificación final favorece a Luis, destacando la importancia de la capacidad de gasto dentro del modelo RFM ponderado.

#**Modelo RFM**

```python
import pandas as pd
import datetime as dt

def ModeloRFM(df, producto, peso_R, peso_F, peso_M):
    # Filtrado por producto
    if producto == "TODOS":
        df2 = df.copy()
    else:
        df2 = df[df["producto_nomb_producto"] == producto].reset_index(drop=True)

    # Convertir id_tiempo a fecha
    df2['Fecha'] = pd.to_datetime(df2['id_tiempo'], format='%Y%m%d', errors='coerce')

    # Fecha de snapshot para cálculo de recencia
    snapshot_date = df2['Fecha'].max() + dt.timedelta(days=1)

    # Consolidación por cliente_documento
    df_rfm = df2.groupby('cliente_documento').agg({
        'Fecha': lambda x: (snapshot_date - x.max()).days,
        'cant_formularios': 'sum',
        'vlr_transaccion': 'sum',
        'cliente_genero': 'first',         # Mantener el género
        'cliente_edad': 'first'            # Para calcular generación
    }).reset_index()

    # Renombrar columnas
    df_rfm.rename(columns={
        'Fecha': 'Recencia',
        'cant_formularios': 'Frecuencia',
        'vlr_transaccion': 'Monto'
    }, inplace=True)

    # Calcular generación basada en edad
    def clasificar_generacion(edad):
        if edad < 18:
            return "Menor de edad"
        elif 18 <= edad <= 27:
            return "Generación Z"
        elif 28 <= edad <= 43:
            return "Millennial"
        elif 44 <= edad <= 59:
            return "Generación X"
        elif 60 <= edad <= 78:
            return "Baby Boomer"
        else:
            return "Generación Silenciosa"

    df_rfm['Generacion'] = df_rfm['cliente_edad'].apply(clasificar_generacion)

    # Ranking RFM
    df_rfm['RANK_R'] = df_rfm['Recencia'].rank(pct=True)
    df_rfm['RANK_F'] = df_rfm['Frecuencia'].rank(pct=True)
    df_rfm['RANK_M'] = df_rfm['Monto'].rank(pct=True)

    # Funciones de scoring
    def Rscore(x):
        if x <= 0.2: return 5
        elif x <= 0.4: return 4
        elif x <= 0.6: return 3
        elif x <= 0.8: return 2
        else: return 1

    def FMScore(x):
        if x <= 0.2: return 1
        elif x <= 0.4: return 2
        elif x <= 0.6: return 3
        elif x <= 0.8: return 4
        else: return 5

    df_rfm['R'] = df_rfm['RANK_R'].apply(Rscore)
    df_rfm['F'] = df_rfm['RANK_F'].apply(FMScore)
    df_rfm['M'] = df_rfm['RANK_M'].apply(FMScore)

    # RFM ponderado
    df_rfm['RFM_nivel'] = df_rfm['R']*peso_R + df_rfm['F']*peso_F + df_rfm['M']*peso_M

    # Segmentación general
    def segment_me(bd):
        if bd['RFM_nivel'] >= 4.2:
            return 'Diamante'
        elif bd['RFM_nivel'] >= 3.1:
            return 'Oro'
        elif bd['RFM_nivel'] >= 1.9:
            return 'Plata'
        else:
            return 'Bronce'

    df_rfm['General_Segment'] = df_rfm.apply(segment_me, axis=1)

    # Segmentación detallada
    def new_segment(bd):
        if (bd['R'] <= 2) and (bd['F'] <= 2):
            return "Perdidos"
        elif (1<=bd['R']<=2) and (3<=bd['F']<=4):
            return "En Riesgo"
        elif (1<=bd['R']<=2) and (bd['F']==5):
            return "Proximos a Perder"
        elif (bd['R']==3) and (1<=bd['F']<=2):
            return "En Duda"
        elif (bd['R']==3) and (bd['F']==3):
            return "Necesita Atencion"
        elif (3<=bd['R']<=4) and (4<=bd['F']<=5):
            return "Leales"
        elif (bd['R']==4) and (bd['F']==1):
            return "Promesa"
        elif (bd['R']==5) and (bd['F']==1):
            return "Nuevos"
        elif (4<=bd['R']<=5) and (2<=bd['F']<=3):
            return "Leales Potenciales"
        elif (bd['R']==5) and (4<=bd['F']<=5):
            return "Estrella"
        else:
            return "Vacio"

    df_rfm['Segment'] = df_rfm.apply(new_segment, axis=1)
    df_rfm['Producto'] = producto

    return df_rfm
```

La función definina anteriormente como "ModeloRFM", realiza el cálculo del modelo RFM, basado en la información de las transacciones de los clientes que realizan en cada uno de los productos. La función tiene como entrada 5 parametros. EL primero es el data frame depurado con la información de la transaccion de todos los productos.

Un segundo parámetro es el producto, esto es importante dado que el comportamiento del consumo del cliente puede variar de producto a producto, es decir, puede que un cliente sea Diamante en chance pero puede que sea Bronce en Astro Millonario dado sus preferencias de consumo. Sin embargo, este parametro puede recibir cualquier producto para verlo de manera individual o puede recibir el valor de "TODOS" para que realice el cálculo del RFM en general.

Los últimos tres parámetros corresponden a la ponderación que se le dará a las variables de Frecuencia, Recencia y Monto. Se tiene por regla de negocio definir los valores de 10% para Frecuencia, 30% para Recencia y lo que más se valora es el Monto el cual se pondera por el 60%. Pero queda abierto por si se desea realizar otra configuración.

El output de la función anterior es un data frame que oculta el id del cliente (por seguridad) y muestra el valor consolidado de la Frecuencia, la Recencia y el Monto. Adicionalmente muestra el rango de las variables y su puntaje de RFM, cone l cual se cálcula qué tipo de cliente es (General_Segment) y el estado en que se encuentra (Segment).

# **Conclusiones generales**

Patrones de comportamiento de compra.
El análisis sugiere que, para el producto Giros, los clientes presentan niveles relativamente altos de frecuencia de compra, aunque estos no se traducen necesariamente en montos elevados. Esto indica un patrón de transacciones frecuentes pero de bajo valor unitario.

Oportunidades de fortalecimiento relacional según RFM.
La clasificación obtenida a partir del modelo RFM permite identificar segmentos con potencial de crecimiento, especialmente aquellos ubicados en categorías como Plata y Bronce. Esto abre la posibilidad de diseñar estrategias focalizadas que incentiven su migración hacia segmentos de mayor valor, como Oro o Diamante, mediante acciones de retención, ofertas diferenciadas o programas de fidelización.

Posibilidades de ventas cruzadas.
El análisis por producto sugiere que existen oportunidades para implementar estrategias de cross-selling. Estas podrían orientarse hacia incrementar la diversificación de productos dentro de cada cliente, fortaleciendo el número de transacciones y los montos asociados, especialmente en líneas con menor demanda relativa.
