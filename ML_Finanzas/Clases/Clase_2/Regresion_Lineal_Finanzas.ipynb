# Regresión Lineal: Fundamentos y Aplicaciones en Finanzas

## Introducción Conceptual

La **regresión lineal** es una de las técnicas más fundamentales y ampliamente utilizadas en el análisis financiero y la modelación predictiva. En esencia, nos permite entender y cuantificar la relación entre variables financieras mediante el ajuste de una línea recta que mejor describe esta relación.

### ¿Qué es la Regresión Lineal?

La regresión lineal es un método estadístico que busca encontrar la **mejor línea recta** que describe la relación entre una variable dependiente (lo que queremos predecir) y una o más variables independientes (los factores que influyen). 

**Concepto Clave**: La regresión lineal asume que existe una relación lineal entre las variables, es decir, que un cambio unitario en la variable independiente produce un cambio constante en la variable dependiente.

**Ecuación básica:**

$$y = \beta_0 + \beta_1 x + \epsilon$$

Donde:
- **$y$** = Variable dependiente (lo que queremos predecir o explicar). En finanzas, podría ser el precio de un activo, el retorno de una inversión, o el valor de un portafolio.
- **$x$** = Variable independiente (el factor explicativo). Representa la variable que creemos que influye en $y$.
- **$\beta_0$** = Intercepto (valor de $y$ cuando $x = 0$). Representa el valor base de la variable dependiente cuando la variable independiente es cero.
- **$\beta_1$** = Pendiente o coeficiente de regresión. Indica cuánto cambia $y$ en promedio por cada unidad de cambio en $x$. Este es el parámetro más importante desde la perspectiva financiera, ya que cuantifica la sensibilidad de la variable dependiente a cambios en la variable independiente.
- **$\epsilon$** = Término de error (diferencia entre el valor real y el predicho). Captura toda la variabilidad en $y$ que no puede ser explicada por $x$. En finanzas, este error puede representar factores no observables, ruido de mercado, o eventos aleatorios.

### ¿Por qué es Importante en Finanzas?

En el mundo financiero, la regresión lineal nos ayuda a:

1. **Análisis de Riesgo**: Entender cómo los cambios en factores de mercado afectan el precio de un activo
2. **Valuación de Activos**: Estimar el valor de un activo basado en características observables
3. **Modelos de Factores**: Explicar los retornos de un portafolio usando factores de riesgo (como el modelo CAPM)
4. **Pronósticos**: Predecir valores futuros basados en tendencias históricas
5. **Análisis de Sensibilidad**: Medir cómo cambios en variables económicas afectan resultados financieros

## Importación de Librerías

Primero, importamos las librerías necesarias para nuestro análisis:

````python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import LinearRegression
from scipy import stats
from scipy.stats import shapiro, jarque_bera

# Configuración para mejores gráficos
plt.style.use('seaborn-v0_8-darkgrid')
sns.set_palette("husl")
plt.rcParams['figure.figsize'] = (10, 6)
plt.rcParams['font.size'] = 12

print("Librerías importadas exitosamente")
print("Incluye: numpy, matplotlib, seaborn, scikit-learn, scipy")
````

## Ejemplo Financiero: Relación entre Tasa de Interés y Precio de Bonos

Para este ejemplo, simularemos datos que representan la relación entre:
- **Tasa de interés del mercado** (variable independiente)
- **Precio de un bono corporativo** (variable dependiente)

En la realidad, cuando las tasas de interés suben, los precios de los bonos tienden a bajar (relación inversa). Para simplificar, simularemos una relación positiva donde el precio del bono aumenta con la tasa de interés, lo cual podría representar un escenario donde bonos de mayor riesgo ofrecen mayores rendimientos.

### Generación de Datos Simulados

Vamos a crear datos que simulen esta relación financiera:

````python
# Generar datos simulados
# Simulamos tasas de interés entre 2% y 8%
np.random.seed(42)  # Para reproducibilidad
tasas_interes = np.random.uniform(2, 8, 100)  # 100 observaciones

# Simulamos precios de bonos con una relación lineal + ruido aleatorio
# Precio base = 1000, relación positiva con tasa de interés + variabilidad
precios_bonos = 1000 + 50 * tasas_interes + np.random.normal(0, 30, 100)

print(f"Tasas de interés: rango de {tasas_interes.min():.2f}% a {tasas_interes.max():.2f}%")
print(f"Precios de bonos: rango de ${precios_bonos.min():.2f} a ${precios_bonos.max():.2f}")
````

### Visualización Exploratoria de los Datos

Antes de construir el modelo, es crucial visualizar los datos para entender la relación:

````python
# Visualizar la relación entre las variables
plt.figure(figsize=(10, 6))
plt.scatter(tasas_interes, precios_bonos, alpha=0.6, s=50)
plt.xlabel('Tasa de Interés (%)', fontsize=12)
plt.ylabel('Precio del Bono ($)', fontsize=12)
plt.title('Relación entre Tasa de Interés y Precio de Bonos', fontsize=14, fontweight='bold')
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
````

**Observación**: Podemos ver que existe una relación aproximadamente lineal entre las tasas de interés y los precios de los bonos. Los puntos se distribuyen alrededor de una línea recta, aunque con cierta variabilidad (ruido).

## Paso 1: Elegir una Clase de Modelo

En este caso, elegimos **regresión lineal** porque:

1. Observamos una relación aproximadamente lineal en los datos
2. Es interpretable: los coeficientes tienen significado financiero directo
3. Es computacionalmente eficiente
4. Funciona bien como modelo base antes de explorar modelos más complejos

### Importar el Modelo

Scikit-learn nos proporciona una implementación eficiente y fácil de usar:

````python
from sklearn.linear_model import LinearRegression

print("Modelo de Regresión Lineal importado exitosamente")
````

## Paso 2: Elegir Hiperparámetros del Modelo

Los **hiperparámetros** son configuraciones del modelo que debemos establecer antes de entrenarlo. Para la regresión lineal, los principales hiperparámetros son:

### Hiperparámetros Clave:

- **`fit_intercept=True`**: Indica si el modelo debe calcular el intercepto ($\beta_0$). En finanzas, generalmente queremos incluir el intercepto porque representa el valor base del activo cuando la variable explicativa es cero.

- **`normalize=False`**: Indica si normalizar las características. En nuestro caso simple, no es necesario.

- **`n_jobs`**: Número de trabajos paralelos (útil cuando tenemos múltiples características).

### Configuración del Modelo:

````python
# Crear instancia del modelo con hiperparámetros
modelo = LinearRegression(fit_intercept=True)

print("Modelo configurado con:")
print(f"- Incluir intercepto: {modelo.fit_intercept}")
print(f"- Normalizar: {modelo.normalize if hasattr(modelo, 'normalize') else 'No aplica'}")
````

**Nota Conceptual**: El intercepto ($\beta_0$) en nuestro contexto financiero representa el precio teórico del bono cuando la tasa de interés es 0%. Aunque esto no tiene sentido práctico directo, permite que el modelo tenga más flexibilidad para ajustarse a los datos.

## Paso 3: Organizar los Datos

Este es un paso crítico. Scikit-learn espera los datos en un formato específico:

### Estructura de Datos Requerida:

- **Matriz de características (X)**: Debe ser un array 2D, donde cada fila es una observación y cada columna es una característica (variable independiente).
  - Forma: `(n_muestras, n_características)`
  - En nuestro caso: `(100, 1)` - 100 observaciones, 1 característica (tasa de interés)

- **Vector objetivo (y)**: Debe ser un array 1D con los valores que queremos predecir.
  - Forma: `(n_muestras,)`
  - En nuestro caso: `(100,)` - 100 precios de bonos

### Transformación de los Datos:

````python
# Reorganizar los datos en el formato requerido
# X debe ser una matriz 2D (n_muestras, n_características)
X = tasas_interes.reshape(-1, 1)  # reshape(-1, 1) convierte el array 1D en 2D

# y es el vector objetivo (1D)
y = precios_bonos

print(f"Forma de X (características): {X.shape}")
print(f"Forma de y (objetivo): {y.shape}")
print(f"\nPrimeras 5 filas de X:\n{X[:5]}")
print(f"\nPrimeros 5 valores de y:\n{y[:5]}")
````

**Explicación del `reshape(-1, 1)`**:
- `-1` significa "calcula automáticamente el número de filas"
- `1` significa "1 columna"
- Resultado: convierte `[2.5, 3.1, 4.2, ...]` en `[[2.5], [3.1], [4.2], ...]`

Esto es necesario porque scikit-learn espera que cada característica esté en su propia columna, incluso si solo tenemos una característica.

## Paso 4: Ajustar el Modelo a los Datos

El proceso de **ajustar** (o **entrenar**) el modelo consiste en encontrar los valores óptimos de los coeficientes ($\beta_0$ y $\beta_1$) que minimizan la diferencia entre los valores predichos y los valores reales.

### ¿Cómo Funciona el Ajuste? El Método de Mínimos Cuadrados Ordinarios (OLS)

El algoritmo utiliza el método de **Mínimos Cuadrados Ordinarios (OLS - Ordinary Least Squares)**, que es el método estándar para estimar los parámetros de una regresión lineal.

**Objetivo del OLS**: Encontrar los valores de $\beta_0$ y $\beta_1$ que minimizan la **Suma de Cuadrados de los Residuos (SSR)**:

$$\text{SSR} = \sum_{i=1}^{n} (y_i - \hat{y}_i)^2 = \sum_{i=1}^{n} \epsilon_i^2$$

Donde:
- $y_i$ = Valor real observado
- $\hat{y}_i = \beta_0 + \beta_1 x_i$ = Valor predicho por el modelo
- $\epsilon_i = y_i - \hat{y}_i$ = Residuo (error) para la observación $i$

**Proceso del OLS**:

1. **Función Objetivo**: Minimizar la suma de los cuadrados de los errores. Al elevar los errores al cuadrado, penalizamos más los errores grandes y garantizamos que todos los términos sean positivos.

2. **Derivación Matemática**: El OLS utiliza cálculo diferencial para encontrar los valores óptimos:
   - Se calculan las derivadas parciales de SSR con respecto a $\beta_0$ y $\beta_1$
   - Se igualan a cero para encontrar los puntos críticos
   - Esto produce un sistema de ecuaciones lineales que se puede resolver analíticamente

3. **Solución Cerrada**: A diferencia de muchos algoritmos de machine learning que requieren optimización iterativa, el OLS tiene una solución matemática exacta:

$$\beta_1 = \frac{\sum_{i=1}^{n}(x_i - \bar{x})(y_i - \bar{y})}{\sum_{i=1}^{n}(x_i - \bar{x})^2} = \frac{\text{Cov}(x,y)}{\text{Var}(x)}$$

$$\beta_0 = \bar{y} - \beta_1 \bar{x}$$

Donde $\bar{x}$ y $\bar{y}$ son las medias de las variables independiente y dependiente respectivamente.

**Interpretación Financiera del OLS**:
- El coeficiente $\beta_1$ es esencialmente la **covarianza** entre $x$ e $y$ dividida por la **varianza** de $x$. Esto significa que estamos midiendo qué tan fuertemente se mueven juntas las variables, ajustado por la variabilidad de la variable independiente.
- El intercepto $\beta_0$ asegura que la línea pase por el punto $(\bar{x}, \bar{y})$, es decir, por el centroide de los datos.

**Ventajas del OLS**:
- Es **eficiente**: Proporciona los estimadores con menor varianza entre todos los estimadores lineales insesgados (Teorema de Gauss-Markov)
- Es **insesgado**: En promedio, los coeficientes estimados son iguales a los valores verdaderos
- Tiene **solución única**: No hay problemas de convergencia como en métodos iterativos

### Entrenamiento del Modelo:

````python
# Ajustar el modelo a los datos
modelo.fit(X, y)

print("Modelo entrenado exitosamente!")
print(f"\nCoeficientes del modelo:")
print(f"- Pendiente (β₁): {modelo.coef_[0]:.2f}")
print(f"- Intercepto (β₀): {modelo.intercept_:.2f}")
````

````python
# Importar métricas de evaluación
from sklearn.metrics import r2_score, mean_squared_error, mean_absolute_error

# Predecir valores para los datos de entrenamiento usando el modelo ajustado
y_pred = modelo.predict(X)

# Calcular métricas de evaluación
r2 = r2_score(y, y_pred)
mse = mean_squared_error(y, y_pred)
rmse = np.sqrt(mse)
mae = mean_absolute_error(y, y_pred)

# Mostrar métricas
print("=" * 80)
print("EVALUACIÓN DEL MODELO DE REGRESIÓN LINEAL")
print("=" * 80)
print("\nMétricas de Evaluación:")
print(f"\n{'Métrica':<40} {'Valor':<20}")
print("-" * 60)
print(f"{'Coeficiente de Determinación (R²)':<40} {r2:<20.4f} ({r2*100:.2f}%)")
print(f"{'Error Cuadrático Medio (MSE)':<40} {mse:<20.2f}")
print(f"{'Raíz del Error Cuadrático Medio (RMSE)':<40} ${rmse:<19.2f}")
print(f"{'Error Absoluto Medio (MAE)':<40} ${mae:<19.2f}")
print("=" * 80)

# Interpretación de R²
print("\nInterpretación del R²:")
if r2 > 0.7:
    calidad = "Excelente"
elif r2 > 0.5:
    calidad = "Buena"
elif r2 > 0.3:
    calidad = "Moderada"
else:
    calidad = "Baja"
    
print(f"El modelo tiene un ajuste {calidad.lower()} (R² = {r2:.2%})")
print(f"Esto significa que el modelo explica el {r2*100:.2f}% de la variabilidad")
print(f"en los precios de los bonos usando la tasa de interés como variable explicativa.")
print(f"\nEn promedio, nuestras predicciones tienen un error de aproximadamente ${rmse:.2f}")
print(f"lo cual representa un error del {rmse/np.mean(y)*100:.2f}% respecto al precio promedio.")
````

````python
# Visualización de Residuos
# Los residuos son las diferencias entre los valores reales y los predichos
residuos = y - y_pred

# Crear gráficos de análisis de residuos
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# Gráfico 1: Residuos vs Valores Predichos
axes[0, 0].scatter(y_pred, residuos, alpha=0.6, s=50)
axes[0, 0].axhline(y=0, color='r', linestyle='--', linewidth=2)
axes[0, 0].set_xlabel('Valores Predichos ($)', fontsize=11)
axes[0, 0].set_ylabel('Residuos ($)', fontsize=11)
axes[0, 0].set_title('Residuos vs Valores Predichos', fontsize=12, fontweight='bold')
axes[0, 0].grid(True, alpha=0.3)

# Gráfico 2: Histograma de Residuos
axes[0, 1].hist(residuos, bins=15, edgecolor='black', alpha=0.7, color='skyblue')
axes[0, 1].axvline(x=0, color='r', linestyle='--', linewidth=2)
axes[0, 1].set_xlabel('Residuos ($)', fontsize=11)
axes[0, 1].set_ylabel('Frecuencia', fontsize=11)
axes[0, 1].set_title('Distribución de Residuos', fontsize=12, fontweight='bold')
axes[0, 1].grid(True, alpha=0.3)

# Gráfico 3: Q-Q Plot para verificar normalidad
from scipy import stats
stats.probplot(residuos, dist="norm", plot=axes[1, 0])
axes[1, 0].set_title('Q-Q Plot (Normalidad de Residuos)', fontsize=12, fontweight='bold')
axes[1, 0].grid(True, alpha=0.3)

# Gráfico 4: Residuos vs Variable Independiente (Tasa de Interés)
axes[1, 1].scatter(tasas_interes, residuos, alpha=0.6, s=50)
axes[1, 1].axhline(y=0, color='r', linestyle='--', linewidth=2)
axes[1, 1].set_xlabel('Tasa de Interés (%)', fontsize=11)
axes[1, 1].set_ylabel('Residuos ($)', fontsize=11)
axes[1, 1].set_title('Residuos vs Tasa de Interés', fontsize=12, fontweight='bold')
axes[1, 1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

# Estadísticas de residuos
print("\n" + "=" * 80)
print("ANÁLISIS DE RESIDUOS")
print("=" * 80)
print(f"\nMedia de residuos: ${residuos.mean():.4f} (debe ser cercana a 0)")
print(f"Desviación estándar de residuos: ${residuos.std():.2f}")
print(f"Mediana de residuos: ${np.median(residuos):.4f}")

# Verificar normalidad de residuos con test de Shapiro-Wilk
from scipy.stats import shapiro
stat_shapiro, p_value_shapiro = shapiro(residuos)
print(f"\nTest de Shapiro-Wilk para normalidad:")
print(f"  Estadístico: {stat_shapiro:.4f}")
print(f"  p-valor: {p_value_shapiro:.4f}")
if p_value_shapiro > 0.05:
    print(f"  ✓ Los residuos parecen seguir una distribución normal (p > 0.05)")
else:
    print(f"  ⚠ Los residuos NO parecen seguir una distribución normal (p ≤ 0.05)")

# Verificar homocedasticidad (varianza constante)
# Si los residuos se distribuyen aleatoriamente alrededor de 0, hay homocedasticidad
print(f"\nAnálisis de Homocedasticidad:")
print(f"  Si los residuos se distribuyen aleatoriamente alrededor de 0 en el gráfico")
print(f"  'Residuos vs Valores Predichos', hay homocedasticidad (varianza constante).")
````

### Interpretación Financiera de los Coeficientes:

**Pendiente ($\beta_1$)**: Indica cuánto cambia el precio del bono por cada cambio de 1% en la tasa de interés.
- Si $\beta_1 = 50$, significa que por cada aumento del 1% en la tasa de interés, el precio del bono aumenta en promedio $50.

**Intercepto ($\beta_0$)**: Representa el precio teórico del bono cuando la tasa de interés es 0%.
- En nuestro ejemplo, $\beta_0$ sería el precio base del bono.

````python
# Crear un rango de valores para visualizar la línea de regresión
tasas_interes_grafico = np.linspace(tasas_interes.min(), tasas_interes.max(), 100)
X_grafico = tasas_interes_grafico.reshape(-1, 1)

# Predecir valores usando el modelo
precios_predichos = modelo.predict(X_grafico)

# Visualizar datos y línea de regresión
plt.figure(figsize=(12, 7))
plt.scatter(tasas_interes, precios_bonos, alpha=0.6, s=50, label='Datos observados')
plt.plot(tasas_interes_grafico, precios_predichos, 'r-', linewidth=2, label='Línea de regresión')
plt.xlabel('Tasa de Interés (%)', fontsize=12)
plt.ylabel('Precio del Bono ($)', fontsize=12)
plt.title('Regresión Lineal: Tasa de Interés vs Precio de Bonos', fontsize=14, fontweight='bold')
plt.legend(fontsize=11)
plt.grid(True, alpha=0.3)

# Agregar ecuación de la recta en el gráfico
ecuacion = f'y = {modelo.intercept_:.2f} + {modelo.coef_[0]:.2f}x'
plt.text(0.05, 0.95, f'Ecuación: {ecuacion}', transform=plt.gca().transAxes,
         fontsize=11, verticalalignment='top', bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.5))

plt.tight_layout()
plt.show()
````

**Análisis Visual**:

- Algunos puntos están por encima de la línea (subestimados) y otros por debajo (sobrestimados), lo cual es normal debido al ruido en los datos

## Evaluación del Modelo

Es importante evaluar qué tan bien funciona nuestro modelo. Las métricas más comunes son:

### 1. Coeficiente de Determinación (R²)

El **R²** (R-cuadrado) mide qué proporción de la variabilidad en los datos es explicada por el modelo.

- **R² = 1.0**: El modelo explica el 100% de la variabilidad (perfecto)
- **R² = 0.0**: El modelo no explica nada de la variabilidad
- **R² > 0.7**: Generalmente considerado un buen ajuste

### 2. Error Cuadrático Medio (MSE)

Mide el promedio de los errores al cuadrado entre valores predichos y reales.

### 3. Raíz del Error Cuadrático Medio (RMSE)

Es la raíz cuadrada del MSE, en las mismas unidades que la variable objetivo.

## Selección de Variables en Regresión Múltiple

Cuando trabajamos con **regresión múltiple** (múltiples variables independientes), surge una pregunta fundamental: **¿Qué variables debemos incluir en el modelo?**

### ¿Por qué es Importante la Selección de Variables?

En finanzas, a menudo tenemos acceso a muchas variables potencialmente relevantes:
- Variables macroeconómicas (inflación, PIB, tasas de interés)
- Métricas de la empresa (ingresos, deuda, capitalización)
- Indicadores de mercado (volatilidad, volumen, momentum)
- Factores de riesgo (beta, tamaño, valor)

**Problemas de incluir demasiadas variables**:
1. **Sobreajuste (Overfitting)**: El modelo se ajusta demasiado a los datos de entrenamiento y funciona mal en datos nuevos
2. **Multicolinealidad**: Variables correlacionadas causan inestabilidad en los coeficientes
3. **Interpretación difícil**: Demasiadas variables hacen el modelo difícil de entender y explicar
4. **Costo computacional**: Más variables requieren más datos y más tiempo de cómputo

**Problemas de incluir muy pocas variables**:
1. **Sesgo**: Omitir variables importantes puede sesgar los coeficientes estimados
2. **Bajo poder explicativo**: El modelo no captura toda la variabilidad relevante

### Objetivo de la Selección de Variables

Encontrar el **subconjunto óptimo** de variables que:
- Maximiza el poder predictivo del modelo
- Minimiza el sobreajuste
- Facilita la interpretación
- Es parsimonioso (principio de parsimonia: el modelo más simple que explica bien los datos)

### Métodos de Selección de Variables

Existen tres métodos principales de selección paso a paso:

### Método 1: Forward Selection (Selección Hacia Adelante)

**Concepto**: Comenzamos con un modelo vacío (sin variables) y vamos agregando variables una por una.

**Algoritmo**:
1. **Paso inicial**: Empezar con el modelo que solo tiene el intercepto: $y = \beta_0 + \epsilon$
2. **Para cada variable no incluida**: 
   - Agregar la variable al modelo
   - Evaluar el modelo (usando un criterio como R², AIC, BIC, o p-valor)
   - Seleccionar la variable que produce la **mayor mejora** en el criterio
3. **Decisión**: Si la mejora es estadísticamente significativa, agregar la variable y repetir. Si no, detener.
4. **Iteración**: Repetir hasta que ninguna variable adicional mejore significativamente el modelo

**Criterios de Evaluación Comunes**:
- **R² ajustado**: Aumenta cuando agregamos variables relevantes
- **AIC (Akaike Information Criterion)**: Penaliza modelos complejos; menor es mejor
- **BIC (Bayesian Information Criterion)**: Similar a AIC pero con penalización más fuerte; menor es mejor
- **p-valor del F-test**: Prueba si la variable agregada es significativa

**Ventajas**:
- ✅ Computacionalmente eficiente (evalúa menos modelos que backward)
- ✅ Útil cuando hay muchas variables candidatas
- ✅ Comienza con modelo simple y va agregando complejidad

**Desventajas**:
- ❌ Puede quedar atrapado en un óptimo local
- ❌ Una vez que una variable entra, nunca sale (puede incluir variables que luego se vuelven redundantes)
- ❌ No considera interacciones entre variables desde el inicio

**Ejemplo en Finanzas**: 
Queremos predecir el retorno de una acción. Comenzamos sin variables, luego agregamos "beta del mercado" (mejora significativa), luego "tamaño de la empresa" (mejora), luego "ratio precio/valor en libros" (no mejora significativamente → detenemos).

### Método 2: Backward Elimination (Eliminación Hacia Atrás)

**Concepto**: Comenzamos con un modelo completo (todas las variables) y vamos eliminando variables una por una.

**Algoritmo**:
1. **Paso inicial**: Empezar con el modelo completo que incluye todas las variables candidatas: $y = \beta_0 + \beta_1 x_1 + ... + \beta_k x_k + \epsilon$
2. **Para cada variable incluida**:
   - Eliminar la variable del modelo
   - Evaluar el modelo resultante
   - Identificar la variable cuya eliminación produce la **menor degradación** (o incluso mejora) del modelo
3. **Decisión**: Si eliminar la variable no empeora significativamente el modelo (o lo mejora), eliminarla y repetir. Si todas las variables son significativas, detener.
4. **Iteración**: Repetir hasta que todas las variables restantes sean estadísticamente significativas

**Criterios de Evaluación**:
- **p-valor**: Eliminar variables con p-valor > umbral (típicamente 0.05 o 0.10)
- **AIC/BIC**: Eliminar variables que reducen AIC/BIC al ser eliminadas
- **R² ajustado**: Eliminar variables que no reducen R² ajustado significativamente

**Ventajas**:
- ✅ Considera todas las variables desde el inicio
- ✅ Puede detectar interacciones entre variables mejor que forward
- ✅ Útil cuando tenemos un conjunto razonable de variables candidatas

**Desventajas**:
- ❌ Computacionalmente más costoso si empezamos con muchas variables
- ❌ Requiere que el modelo completo sea estimable (suficientes observaciones)
- ❌ Puede eliminar variables que son importantes en combinación con otras

**Ejemplo en Finanzas**:
Queremos explicar el precio de un bono usando 10 variables macroeconómicas. Comenzamos con todas, eliminamos "tasa de desempleo" (no significativa), luego "crecimiento del PIB" (no significativa), y continuamos hasta que todas las variables restantes sean significativas.

### Método 3: Stepwise Selection (Selección Paso a Paso)

**Concepto**: Combina forward y backward. En cada paso, podemos **agregar** una variable (como forward) o **eliminar** una variable (como backward).

**Algoritmo**:
1. **Paso inicial**: Comenzar con un modelo (puede ser vacío o con algunas variables iniciales)
2. **En cada iteración**:
   - **Fase Forward**: Evaluar agregar cada variable no incluida. Si alguna mejora significativamente el modelo, agregarla.
   - **Fase Backward**: Evaluar eliminar cada variable incluida. Si eliminar alguna mejora el modelo (o no lo empeora significativamente), eliminarla.
3. **Criterio de parada**: Detener cuando:
   - No se puede agregar ninguna variable que mejore el modelo
   - No se puede eliminar ninguna variable sin empeorar significativamente el modelo
   - Se alcanza un número máximo de iteraciones

**Ventajas**:
- ✅ Más flexible que forward o backward solos
- ✅ Puede corregir decisiones anteriores (agregar y luego eliminar si es necesario)
- ✅ Generalmente encuentra mejores modelos que forward o backward por separado
- ✅ Considera tanto agregar como eliminar en cada paso

**Desventajas**:
- ❌ Computacionalmente más costoso
- ❌ Puede ser más difícil de interpretar el proceso
- ❌ Riesgo de sobreajuste si no se controla adecuadamente

**Variantes**:
- **Bidirectional Elimination**: Similar pero permite múltiples agregaciones/eliminaciones por iteración
- **Stepwise con validación cruzada**: Usa validación cruzada para evaluar modelos en lugar de solo criterios estadísticos

**Ejemplo en Finanzas**:
Modelando el retorno de un portafolio. Agregamos "beta" (forward), luego "tamaño" (forward), luego eliminamos "beta" porque con "tamaño" ya no es significativa (backward), luego agregamos "momentum" (forward), y así sucesivamente hasta encontrar el modelo óptimo.

### Comparación de los Tres Métodos

| Característica | Forward Selection | Backward Elimination | Stepwise Selection |
|----------------|-------------------|----------------------|---------------------|
| **Punto de inicio** | Modelo vacío | Modelo completo | Modelo inicial (vacío o parcial) |
| **Operación principal** | Agregar variables | Eliminar variables | Agregar y eliminar |
| **Costo computacional** | Bajo | Medio-Alto | Medio-Alto |
| **Flexibilidad** | Baja | Media | Alta |
| **Mejor cuando** | Muchas variables candidatas | Pocas variables, modelo completo estimable | Balance entre ambos |
| **Riesgo de sobreajuste** | Medio | Bajo-Medio | Medio-Alto |
| **Capacidad de corregir** | No puede eliminar | No puede agregar | Puede agregar y eliminar |

### Criterios de Evaluación para Selección de Variables

**1. R² Ajustado (Adjusted R²)**
- Ajusta R² penalizando por el número de variables
- Fórmula: $R^2_{adj} = 1 - \frac{(1-R^2)(n-1)}{n-p-1}$ donde $p$ es el número de variables
- **Mayor es mejor**: Queremos maximizar R² ajustado

**2. AIC (Akaike Information Criterion)**
- Balancea bondad de ajuste y complejidad del modelo
- Fórmula: $AIC = 2p - 2\ln(L)$ donde $L$ es la verosimilitud
- **Menor es mejor**: Penaliza modelos con muchas variables

**3. BIC (Bayesian Information Criterion)**
- Similar a AIC pero con penalización más fuerte
- Fórmula: $BIC = p\ln(n) - 2\ln(L)$
- **Menor es mejor**: Tiende a seleccionar modelos más simples que AIC

**4. p-valor del F-test**
- Prueba si agregar/eliminar una variable mejora significativamente el modelo
- **Umbral típico**: 0.05 o 0.10
- Si p-valor < umbral: la variable es significativa

**5. Validación Cruzada**
- Evalúa el rendimiento predictivo en datos no vistos
- **Métrica común**: RMSE o R² en conjunto de validación
- **Mejor para predicción**: Más relevante que criterios estadísticos si el objetivo es predicción

### Implementación de Forward Selection

Vamos a implementar forward selection desde cero para entender el proceso:

````python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
import itertools

def calcular_aic(r2, n, p):
    """Calcula AIC para un modelo de regresión"""
    # AIC = n * ln(RSS/n) + 2p
    # Simplificado: AIC ≈ n * ln(1-R²) + 2p (aproximación)
    rss = (1 - r2) * n  # Residual Sum of Squares aproximado
    if rss <= 0:
        return np.inf
    aic = n * np.log(rss / n) + 2 * (p + 1)  # +1 por el intercepto
    return aic

def forward_selection(X, y, max_vars=None, threshold_pvalue=0.05):
    """
    Implementa Forward Selection
    
    Parámetros:
    -----------
    X : DataFrame con variables candidatas
    y : Serie con variable dependiente
    max_vars : Número máximo de variables a incluir (None = sin límite)
    threshold_pvalue : Umbral de p-valor para incluir variable
    """
    variables_incluidas = []
    variables_candidatas = list(X.columns)
    mejor_aic = np.inf
    historial = []
    
    print("=" * 70)
    print("FORWARD SELECTION")
    print("=" * 70)
    
    # Modelo inicial (solo intercepto)
    modelo_inicial = LinearRegression()
    modelo_inicial.fit(np.ones((len(y), 1)), y)
    y_pred_inicial = modelo_inicial.predict(np.ones((len(y), 1)))
    r2_inicial = r2_score(y, y_pred_inicial)
    aic_inicial = calcular_aic(r2_inicial, len(y), 0)
    
    print(f"\nPaso 0: Modelo inicial (solo intercepto)")
    print(f"  R²: {r2_inicial:.4f}")
    print(f"  AIC: {aic_inicial:.2f}")
    print(f"  Variables incluidas: ninguna")
    
    mejor_aic = aic_inicial
    historial.append({
        'paso': 0,
        'variables': [],
        'r2': r2_inicial,
        'aic': aic_inicial
    })
    
    paso = 1
    while len(variables_candidatas) > 0:
        if max_vars and len(variables_incluidas) >= max_vars:
            break
            
        mejor_mejora = None
        mejor_variable = None
        mejor_r2 = None
        mejor_aic_nuevo = None
        
        # Probar agregar cada variable candidata
        for var in variables_candidatas:
            vars_temp = variables_incluidas + [var]
            X_temp = X[vars_temp]
            
            modelo = LinearRegression()
            modelo.fit(X_temp, y)
            y_pred = modelo.predict(X_temp)
            r2 = r2_score(y, y_pred)
            aic_nuevo = calcular_aic(r2, len(y), len(vars_temp))
            
            mejora_aic = mejor_aic - aic_nuevo  # Reducción en AIC (mayor es mejor)
            
            if mejor_mejora is None or mejora_aic > mejor_mejora:
                mejor_mejora = mejora_aic
                mejor_variable = var
                mejor_r2 = r2
                mejor_aic_nuevo = aic_nuevo
        
        # Decisión: ¿agregar la variable?
        if mejor_mejora > 0:  # Si mejora el AIC
            variables_incluidas.append(mejor_variable)
            variables_candidatas.remove(mejor_variable)
            mejor_aic = mejor_aic_nuevo
            
            print(f"\nPaso {paso}: Agregar '{mejor_variable}'")
            print(f"  R²: {mejor_r2:.4f}")
            print(f"  AIC: {mejor_aic_nuevo:.2f} (mejora: {mejor_mejora:.2f})")
            print(f"  Variables incluidas: {variables_incluidas}")
            
            historial.append({
                'paso': paso,
                'variable_agregada': mejor_variable,
                'variables': variables_incluidas.copy(),
                'r2': mejor_r2,
                'aic': mejor_aic_nuevo,
                'mejora_aic': mejor_mejora
            })
            paso += 1
        else:
            print(f"\nPaso {paso}: Ninguna variable adicional mejora el modelo")
            print(f"  Deteniendo forward selection")
            break
    
    print("\n" + "=" * 70)
    print("RESUMEN FINAL")
    print("=" * 70)
    print(f"Variables seleccionadas: {variables_incluidas}")
    print(f"R² final: {mejor_r2:.4f}")
    print(f"AIC final: {mejor_aic:.2f}")
    
    return variables_incluidas, historial

# Preparar datos
X_vars = datos.drop('precio_bono', axis=1)
y_var = datos['precio_bono']

# Ejecutar Forward Selection
vars_seleccionadas_forward, historial_forward = forward_selection(X_vars, y_var, max_vars=6)
````

### Implementación de Backward Elimination

````python
def backward_elimination(X, y, min_vars=1, threshold_pvalue=0.10):
    """
    Implementa Backward Elimination
    
    Parámetros:
    -----------
    X : DataFrame con variables candidatas
    y : Serie con variable dependiente
    min_vars : Número mínimo de variables a mantener
    threshold_pvalue : Umbral de p-valor para mantener variable
    """
    variables_incluidas = list(X.columns)
    historial = []
    
    print("=" * 70)
    print("BACKWARD ELIMINATION")
    print("=" * 70)
    
    # Modelo inicial (todas las variables)
    modelo_inicial = LinearRegression()
    modelo_inicial.fit(X[variables_incluidas], y)
    y_pred_inicial = modelo_inicial.predict(X[variables_incluidas])
    r2_inicial = r2_score(y, y_pred_inicial)
    aic_inicial = calcular_aic(r2_inicial, len(y), len(variables_incluidas))
    
    print(f"\nPaso 0: Modelo inicial (todas las variables)")
    print(f"  R²: {r2_inicial:.4f}")
    print(f"  AIC: {aic_inicial:.2f}")
    print(f"  Variables incluidas: {variables_incluidas}")
    
    mejor_aic = aic_inicial
    historial.append({
        'paso': 0,
        'variables': variables_incluidas.copy(),
        'r2': r2_inicial,
        'aic': aic_inicial
    })
    
    paso = 1
    while len(variables_incluidas) > min_vars:
        mejor_mejora = None
        mejor_variable = None
        mejor_r2 = None
        mejor_aic_nuevo = None
        
        # Probar eliminar cada variable incluida
        for var in variables_incluidas:
            vars_temp = [v for v in variables_incluidas if v != var]
            if len(vars_temp) == 0:
                continue
                
            X_temp = X[vars_temp]
            
            modelo = LinearRegression()
            modelo.fit(X_temp, y)
            y_pred = modelo.predict(X_temp)
            r2 = r2_score(y, y_pred)
            aic_nuevo = calcular_aic(r2, len(y), len(vars_temp))
            
            mejora_aic = mejor_aic - aic_nuevo  # Si es negativo, eliminar mejora el modelo
            
            if mejor_mejora is None or mejora_aic < mejor_mejora:
                mejor_mejora = mejora_aic
                mejor_variable = var
                mejor_r2 = r2
                mejor_aic_nuevo = aic_nuevo
        
        # Decisión: ¿eliminar la variable?
        if mejor_mejora < 0:  # Si eliminar mejora el AIC (mejora_aic negativo)
            variables_incluidas.remove(mejor_variable)
            mejor_aic = mejor_aic_nuevo
            
            print(f"\nPaso {paso}: Eliminar '{mejor_variable}'")
            print(f"  R²: {mejor_r2:.4f}")
            print(f"  AIC: {mejor_aic_nuevo:.2f} (mejora: {abs(mejor_mejora):.2f})")
            print(f"  Variables incluidas: {variables_incluidas}")
            
            historial.append({
                'paso': paso,
                'variable_eliminada': mejor_variable,
                'variables': variables_incluidas.copy(),
                'r2': mejor_r2,
                'aic': mejor_aic_nuevo,
                'mejora_aic': abs(mejor_mejora)
            })
            paso += 1
        else:
            print(f"\nPaso {paso}: Eliminar cualquier variable empeora el modelo")
            print(f"  Deteniendo backward elimination")
            break
    
    print("\n" + "=" * 70)
    print("RESUMEN FINAL")
    print("=" * 70)
    print(f"Variables seleccionadas: {variables_incluidas}")
    print(f"R² final: {mejor_r2:.4f}")
    print(f"AIC final: {mejor_aic:.2f}")
    
    return variables_incluidas, historial

# Ejecutar Backward Elimination
vars_seleccionadas_backward, historial_backward = backward_elimination(X_vars, y_var, min_vars=1)
````

### Implementación de Stepwise Selection

````python
def stepwise_selection(X, y, max_iter=20):
    """
    Implementa Stepwise Selection (combina forward y backward)
    
    Parámetros:
    -----------
    X : DataFrame con variables candidatas
    y : Serie con variable dependiente
    max_iter : Número máximo de iteraciones
    """
    variables_incluidas = []
    variables_candidatas = list(X.columns)
    historial = []
    
    print("=" * 70)
    print("STEPWISE SELECTION")
    print("=" * 70)
    
    # Modelo inicial (solo intercepto)
    modelo_inicial = LinearRegression()
    modelo_inicial.fit(np.ones((len(y), 1)), y)
    y_pred_inicial = modelo_inicial.predict(np.ones((len(y), 1)))
    r2_inicial = r2_score(y, y_pred_inicial)
    mejor_aic = calcular_aic(r2_inicial, len(y), 0)
    
    print(f"\nPaso 0: Modelo inicial (solo intercepto)")
    print(f"  R²: {r2_inicial:.4f}")
    print(f"  AIC: {mejor_aic:.2f}")
    print(f"  Variables incluidas: ninguna")
    
    historial.append({
        'paso': 0,
        'accion': 'inicio',
        'variables': [],
        'r2': r2_inicial,
        'aic': mejor_aic
    })
    
    paso = 1
    for iteracion in range(max_iter):
        cambio = False
        
        # FASE FORWARD: Intentar agregar una variable
        if len(variables_candidatas) > 0:
            mejor_mejora_forward = None
            mejor_variable_forward = None
            mejor_r2_forward = None
            mejor_aic_forward = None
            
            for var in variables_candidatas:
                vars_temp = variables_incluidas + [var]
                X_temp = X[vars_temp]
                
                modelo = LinearRegression()
                modelo.fit(X_temp, y)
                y_pred = modelo.predict(X_temp)
                r2 = r2_score(y, y_pred)
                aic_nuevo = calcular_aic(r2, len(y), len(vars_temp))
                
                mejora_aic = mejor_aic - aic_nuevo
                
                if mejor_mejora_forward is None or mejora_aic > mejor_mejora_forward:
                    mejor_mejora_forward = mejora_aic
                    mejor_variable_forward = var
                    mejor_r2_forward = r2
                    mejor_aic_forward = aic_nuevo
            
            # Si agregar mejora, agregar la variable
            if mejor_mejora_forward > 0:
                variables_incluidas.append(mejor_variable_forward)
                variables_candidatas.remove(mejor_variable_forward)
                mejor_aic = mejor_aic_forward
                cambio = True
                
                print(f"\nPaso {paso}: FORWARD - Agregar '{mejor_variable_forward}'")
                print(f"  R²: {mejor_r2_forward:.4f}")
                print(f"  AIC: {mejor_aic_forward:.2f} (mejora: {mejor_mejora_forward:.2f})")
                print(f"  Variables incluidas: {variables_incluidas}")
                
                historial.append({
                    'paso': paso,
                    'accion': 'forward',
                    'variable': mejor_variable_forward,
                    'variables': variables_incluidas.copy(),
                    'r2': mejor_r2_forward,
                    'aic': mejor_aic_forward
                })
                paso += 1
        
        # FASE BACKWARD: Intentar eliminar una variable
        if len(variables_incluidas) > 0:
            mejor_mejora_backward = None
            mejor_variable_backward = None
            mejor_r2_backward = None
            mejor_aic_backward = None
            
            for var in variables_incluidas:
                vars_temp = [v for v in variables_incluidas if v != var]
                if len(vars_temp) == 0:
                    continue
                    
                X_temp = X[vars_temp]
                
                modelo = LinearRegression()
                modelo.fit(X_temp, y)
                y_pred = modelo.predict(X_temp)
                r2 = r2_score(y, y_pred)
                aic_nuevo = calcular_aic(r2, len(y), len(vars_temp))
                
                mejora_aic = mejor_aic - aic_nuevo  # Negativo si eliminar mejora
                
                if mejor_mejora_backward is None or mejora_aic < mejor_mejora_backward:
                    mejor_mejora_backward = mejora_aic
                    mejor_variable_backward = var
                    mejor_r2_backward = r2
                    mejor_aic_backward = aic_nuevo
            
            # Si eliminar mejora, eliminar la variable
            if mejor_mejora_backward < 0:  # Negativo significa que eliminar mejora
                variables_incluidas.remove(mejor_variable_backward)
                variables_candidatas.append(mejor_variable_backward)
                mejor_aic = mejor_aic_backward
                cambio = True
                
                print(f"\nPaso {paso}: BACKWARD - Eliminar '{mejor_variable_backward}'")
                print(f"  R²: {mejor_r2_backward:.4f}")
                print(f"  AIC: {mejor_aic_backward:.2f} (mejora: {abs(mejor_mejora_backward):.2f})")
                print(f"  Variables incluidas: {variables_incluidas}")
                
                historial.append({
                    'paso': paso,
                    'accion': 'backward',
                    'variable': mejor_variable_backward,
                    'variables': variables_incluidas.copy(),
                    'r2': mejor_r2_backward,
                    'aic': mejor_aic_backward
                })
                paso += 1
        
        # Si no hubo cambios, detener
        if not cambio:
            print(f"\nPaso {paso}: No se pueden hacer más mejoras")
            print(f"  Deteniendo stepwise selection")
            break
    
    print("\n" + "=" * 70)
    print("RESUMEN FINAL")
    print("=" * 70)
    print(f"Variables seleccionadas: {variables_incluidas}")
    print(f"R² final: {mejor_r2_backward if 'mejor_r2_backward' in locals() else mejor_r2_forward:.4f}")
    print(f"AIC final: {mejor_aic:.2f}")
    
    return variables_incluidas, historial

# Ejecutar Stepwise Selection
vars_seleccionadas_stepwise, historial_stepwise = stepwise_selection(X_vars, y_var, max_iter=15)
````

## Supuestos de la Regresión Lineal

Para que la regresión lineal proporcione estimadores confiables y válidos, es fundamental que se cumplan ciertos **supuestos estadísticos**. Estos supuestos son la base teórica que garantiza que el método de Mínimos Cuadrados Ordinarios (OLS) produzca los mejores estimadores lineales insesgados (BLUE - Best Linear Unbiased Estimators).

### ¿Por qué son Importantes los Supuestos?

En finanzas, violar estos supuestos puede llevar a:
- **Estimaciones sesgadas** de los coeficientes
- **Errores estándar incorrectos**, afectando la confiabilidad de las predicciones
- **Inferencias estadísticas inválidas** sobre la significancia de las variables
- **Predicciones poco confiables** para decisiones de inversión

Vamos a examinar cada supuesto en detalle, cómo verificarlo, y qué hacer si se viola.

### Supuesto 1: Linealidad

**El Supuesto**: La relación entre la variable dependiente $y$ y la variable independiente $x$ es **lineal** en los parámetros.

**Formulación Matemática**: 
$$E[y|x] = \beta_0 + \beta_1 x$$

Esto significa que el valor esperado de $y$ dado $x$ es una función lineal de $x$.

**¿Qué Significa en Finanzas?**
- Un cambio de una unidad en $x$ produce un cambio constante en $y$, independientemente del valor actual de $x$
- Por ejemplo, si $\beta_1 = 50$, un aumento del 1% en la tasa de interés siempre aumenta el precio del bono en $50, sin importar si la tasa actual es 2% o 8%

**Cómo Verificarlo**:
1. **Gráfico de dispersión**: Visualizar la relación entre $x$ e $y$. Debería verse aproximadamente como una línea recta.
2. **Gráfico de residuos vs valores predichos**: Los residuos deberían distribuirse aleatoriamente alrededor de cero sin patrones curvilíneos.
3. **Gráfico de residuos vs variable independiente**: Similar al anterior, pero contra $x$ directamente.

**Qué Hacer si se Viola**:
- **Transformaciones**: Aplicar transformaciones logarítmicas, exponenciales o polinómicas
  - Log-log: $\log(y) = \beta_0 + \beta_1 \log(x)$ (útil para elasticidades constantes)
  - Semi-log: $y = \beta_0 + \beta_1 \log(x)$ o $\log(y) = \beta_0 + \beta_1 x$
- **Modelos no lineales**: Usar regresión polinómica, modelos exponenciales, o machine learning no lineal
- **Segmentación**: Dividir los datos en segmentos donde la relación sea aproximadamente lineal

**Ejemplo en Finanzas**: La relación entre precio y rendimiento de un bono es **no lineal** (convexidad), por lo que la regresión lineal simple puede no ser apropiada para rangos amplios de tasas de interés.

### Supuesto 2: Independencia de las Observaciones

**El Supuesto**: Las observaciones son **independientes** entre sí. Esto significa que el valor de $y$ para una observación no está influenciado por el valor de $y$ para otra observación.

**Formulación Matemática**: 
$$\text{Cov}(\epsilon_i, \epsilon_j) = 0 \quad \text{para } i \neq j$$

Los errores de diferentes observaciones no están correlacionados.

**¿Qué Significa en Finanzas?**
- Cada observación (por ejemplo, cada bono o cada período de tiempo) es independiente de las demás
- El precio de un bono en el día $t$ no está influenciado por el precio del mismo bono en el día $t-1$ (esto sería un problema de autocorrelación en series temporales)

**Cómo Verificarlo**:
1. **Conocimiento del proceso de datos**: Entender cómo se generaron los datos
2. **Gráfico de residuos vs orden de observación**: Si hay autocorrelación, veremos patrones (tendencias, ciclos)
3. **Test de Durbin-Watson**: Prueba estadística para detectar autocorrelación (valores cercanos a 2 indican independencia)

**Qué Hacer si se Viola**:
- **Modelos de series temporales**: Si los datos son temporales, usar modelos ARIMA, GARCH, o regresión con errores autocorrelacionados
- **Errores estándar robustos**: Usar errores estándar de Newey-West que son robustos a la autocorrelación
- **Datos de panel**: Si hay dependencia espacial o temporal, considerar modelos de efectos fijos o aleatorios

**Ejemplo en Finanzas**: En series temporales financieras (precios diarios de acciones), la independencia rara vez se cumple debido a la autocorrelación y efectos de memoria del mercado.

### Supuesto 3: Homocedasticidad (Varianza Constante)

**El Supuesto**: La varianza del término de error $\epsilon$ es **constante** para todos los valores de $x$. Esto se conoce como homocedasticidad.

**Formulación Matemática**: 
$$\text{Var}(\epsilon_i | x_i) = \sigma^2 \quad \text{para todo } i$$

La varianza de los errores es la misma independientemente del valor de $x$.

**¿Qué Significa en Finanzas?**
- La incertidumbre (variabilidad) en nuestras predicciones es la misma para todos los niveles de la variable independiente
- Por ejemplo, si estamos prediciendo precios de bonos, el error de predicción debería ser similar tanto para tasas de interés bajas (2%) como altas (8%)

**Heterocedasticidad** (violación del supuesto): Cuando la varianza de los errores **no es constante**. En finanzas, esto es muy común porque:
- La volatilidad del mercado puede cambiar con el nivel de precios
- Los activos de mayor valor pueden tener mayor variabilidad
- Períodos de crisis tienen mayor incertidumbre que períodos estables

**Cómo Verificarlo**:
1. **Gráfico de residuos vs valores predichos**: Buscar patrones de "cono" o "embudo" donde la dispersión de residuos aumenta o disminuye con los valores predichos
2. **Gráfico de residuos vs variable independiente**: Similar al anterior
3. **Test de Breusch-Pagan**: Prueba estadística formal para detectar heterocedasticidad
4. **Test de White**: Otra prueba común para heterocedasticidad

**Qué Hacer si se Viola (Heterocedasticidad)**:
- **Errores estándar robustos**: Usar errores estándar de White o Huber-White que son consistentes incluso con heterocedasticidad
- **Transformación de variables**: Aplicar transformaciones logarítmicas que pueden estabilizar la varianza
- **Mínimos Cuadrados Ponderados (WLS)**: Ponderar las observaciones inversamente proporcional a su varianza
- **Modelos GARCH**: Para modelar explícitamente la varianza cambiante en series temporales financieras

**Ejemplo en Finanzas**: La volatilidad de los retornos de acciones es mayor durante crisis financieras (heterocedasticidad condicional), lo que viola este supuesto en modelos simples de regresión.

### Supuesto 4: Normalidad de los Residuos

**El Supuesto**: Los residuos $\epsilon_i$ siguen una **distribución normal** con media cero y varianza constante.

**Formulación Matemática**: 
$$\epsilon_i \sim N(0, \sigma^2)$$

**¿Qué Significa en Finanzas?**
- Los errores del modelo se distribuyen normalmente alrededor de cero
- Esto es importante principalmente para **inferencia estadística** (pruebas de hipótesis, intervalos de confianza)
- Para **predicción**, este supuesto es menos crítico, especialmente con muestras grandes (debido al Teorema del Límite Central)

**Por qué es Importante**:
- Permite hacer inferencias sobre los coeficientes (pruebas t, intervalos de confianza)
- Garantiza que los estimadores de MCO sean los más eficientes
- Facilita la construcción de intervalos de predicción

**Cómo Verificarlo**:
1. **Histograma de residuos**: Debería tener forma de campana (distribución normal)
2. **Q-Q Plot (Quantile-Quantile Plot)**: Los puntos deberían caer aproximadamente en una línea recta si los residuos son normales
3. **Test de Shapiro-Wilk**: Prueba estadística formal para normalidad
4. **Test de Jarque-Bera**: Prueba que combina asimetría y curtosis

**Qué Hacer si se Viola**:
- **Muestras grandes**: Con muchas observaciones (>30-50), el Teorema del Límite Central garantiza que las inferencias sean válidas incluso sin normalidad exacta
- **Transformaciones**: Transformar la variable dependiente (log, raíz cuadrada) puede ayudar a normalizar los residuos
- **Métodos robustos**: Usar métodos que no requieren normalidad (bootstrap, errores estándar robustos)
- **Modelos alternativos**: Considerar distribuciones no normales si es apropiado para el contexto financiero

**Ejemplo en Finanzas**: Los retornos financieros a menudo tienen "colas pesadas" (más valores extremos que una distribución normal), lo que puede violar este supuesto. Sin embargo, para predicción esto puede no ser crítico.

## Aplicaciones Prácticas en Finanzas

La regresión lineal tiene múltiples aplicaciones en el mundo financiero:

### 1. Modelo CAPM (Capital Asset Pricing Model)

El CAPM utiliza regresión lineal para explicar los retornos de un activo:

$$R_i = R_f + \beta_i (R_m - R_f) + \epsilon_i$$

Donde:
- $R_i$ = Retorno del activo
- $R_f$ = Tasa libre de riesgo
- $\beta_i$ = Beta del activo (medida de riesgo sistemático)
- $R_m$ = Retorno del mercado

### 2. Análisis de Sensibilidad

Medir cómo cambios en variables económicas (inflación, PIB, tasas de interés) afectan el valor de activos o portafolios.

### 3. Valuación de Activos

Estimar el valor de un activo basado en características observables (multiples valuation).

### 4. Análisis de Factores de Riesgo

Identificar y cuantificar factores que explican la variabilidad en retornos de activos.

### 5. Pronósticos Financieros

Predecir valores futuros de variables financieras basándose en tendencias históricas.

### Supuesto 6: No Multicolinealidad Perfecta (Regresión Múltiple)

**El Supuesto**: En regresión múltiple (con varias variables independientes), ninguna variable independiente puede ser una combinación lineal perfecta de otras variables.

**Formulación Matemática**: 
No existe un conjunto de constantes $c_1, c_2, ..., c_k$ (no todas cero) tal que:
$$c_1 x_1 + c_2 x_2 + ... + c_k x_k = 0$$

**¿Qué Significa en Finanzas?**
- Las variables independientes deben ser linealmente independientes
- No podemos tener, por ejemplo, tanto "ingresos totales" como "ingresos netos" si uno es una función exacta del otro
- En la práctica, la multicolinealidad **alta** (aunque no perfecta) también es problemática

**Multicolinealidad Alta**: Cuando las variables están muy correlacionadas entre sí (aunque no perfectamente). Esto causa:
- **Varianza inflada** de los coeficientes estimados
- **Inestabilidad** de las estimaciones (pequeños cambios en los datos producen grandes cambios en los coeficientes)
- **Dificultad de interpretación**: Es difícil separar el efecto individual de cada variable

**Cómo Verificarlo**:
1. **Matriz de correlación**: Calcular correlaciones entre variables independientes (valores > 0.8-0.9 indican problema)
2. **Factor de Inflación de Varianza (VIF)**: Valores VIF > 10 indican multicolinealidad problemática
3. **Tolerancia**: El inverso del VIF, valores < 0.1 indican problema

**Qué Hacer si se Viola**:
- **Eliminar variables redundantes**: Quitar una de las variables altamente correlacionadas
- **Transformaciones**: Crear nuevas variables que combinen información (por ejemplo, ratios en lugar de valores absolutos)
- **Regularización**: Usar Ridge o Lasso que manejan multicolinealidad mejorando la estabilidad
- **Análisis de componentes principales (PCA)**: Reducir la dimensionalidad creando componentes no correlacionados

**Ejemplo en Finanzas**: Si incluimos tanto "capitalización de mercado" como "número de acciones" y "precio por acción" en un modelo, habrá multicolinealidad perfecta porque capitalización = acciones × precio.

````python
# Verificación de Supuestos

# 1. Verificar Linealidad: Gráfico de residuos vs valores predichos
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# Gráfico 1: Residuos vs Valores Predichos (verificar linealidad y homocedasticidad)
axes[0, 0].scatter(y_pred, residuos, alpha=0.6)
axes[0, 0].axhline(y=0, color='r', linestyle='--', linewidth=2)
axes[0, 0].set_xlabel('Valores Predichos ($)', fontsize=11)
axes[0, 0].set_ylabel('Residuos ($)', fontsize=11)
axes[0, 0].set_title('1. Linealidad y Homocedasticidad\n(Residuos vs Predichos)', fontsize=12, fontweight='bold')
axes[0, 0].grid(True, alpha=0.3)

# Agregar línea de tendencia para detectar patrones
from scipy import stats
z = np.polyfit(y_pred, residuos, 1)
p = np.poly1d(z)
axes[0, 0].plot(y_pred, p(y_pred), "g--", alpha=0.5, linewidth=1, label='Tendencia')
axes[0, 0].legend()

# Gráfico 2: Residuos vs Variable Independiente
axes[0, 1].scatter(tasas_interes, residuos, alpha=0.6)
axes[0, 1].axhline(y=0, color='r', linestyle='--', linewidth=2)
axes[0, 1].set_xlabel('Tasa de Interés (%)', fontsize=11)
axes[0, 1].set_ylabel('Residuos ($)', fontsize=11)
axes[0, 1].set_title('2. Linealidad\n(Residuos vs Variable Independiente)', fontsize=12, fontweight='bold')
axes[0, 1].grid(True, alpha=0.3)

# Gráfico 3: Q-Q Plot para verificar normalidad
from scipy.stats import probplot
probplot(residuos, dist="norm", plot=axes[1, 0])
axes[1, 0].set_title('3. Normalidad de Residuos\n(Q-Q Plot)', fontsize=12, fontweight='bold')
axes[1, 0].grid(True, alpha=0.3)

# Gráfico 4: Histograma de residuos con curva normal
axes[1, 1].hist(residuos, bins=15, density=True, edgecolor='black', alpha=0.7, label='Residuos')
# Superponer curva normal teórica
mu, sigma = residuos.mean(), residuos.std()
x_norm = np.linspace(residuos.min(), residuos.max(), 100)
y_norm = stats.norm.pdf(x_norm, mu, sigma)
axes[1, 1].plot(x_norm, y_norm, 'r-', linewidth=2, label='Distribución Normal')
axes[1, 1].set_xlabel('Residuos ($)', fontsize=11)
axes[1, 1].set_ylabel('Densidad', fontsize=11)
axes[1, 1].set_title('4. Normalidad de Residuos\n(Histograma)', fontsize=12, fontweight='bold')
axes[1, 1].legend()
axes[1, 1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

# Pruebas estadísticas formales
from scipy.stats import shapiro, jarque_bera

print("=" * 60)
print("VERIFICACIÓN DE SUPUESTOS - PRUEBAS ESTADÍSTICAS")
print("=" * 60)

# Test de Shapiro-Wilk para normalidad
stat_sw, p_value_sw = shapiro(residuos)
print(f"\n1. Test de Normalidad (Shapiro-Wilk):")
print(f"   Estadístico: {stat_sw:.4f}")
print(f"   p-valor: {p_value_sw:.4f}")
if p_value_sw > 0.05:
    print(f"   ✓ No rechazamos normalidad (p > 0.05)")
else:
    print(f"   ✗ Rechazamos normalidad (p < 0.05)")

# Test de Jarque-Bera para normalidad
stat_jb, p_value_jb = jarque_bera(residuos)
print(f"\n2. Test de Normalidad (Jarque-Bera):")
print(f"   Estadístico: {stat_jb:.4f}")
print(f"   p-valor: {p_value_jb:.4f}")
if p_value_jb > 0.05:
    print(f"   ✓ No rechazamos normalidad (p > 0.05)")
else:
    print(f"   ✗ Rechazamos normalidad (p < 0.05)")

# Verificar homocedasticidad (Breusch-Pagan simplificado)
# Dividir datos en grupos y comparar varianzas
n = len(residuos)
grupo1 = residuos[:n//2]
grupo2 = residuos[n//2:]
var1, var2 = grupo1.var(), grupo2.var()
print(f"\n3. Verificación de Homocedasticidad:")
print(f"   Varianza grupo 1: {var1:.2f}")
print(f"   Varianza grupo 2: {var2:.2f}")
print(f"   Ratio de varianzas: {max(var1, var2)/min(var1, var2):.2f}")
if max(var1, var2)/min(var1, var2) < 2:
    print(f"   ✓ Varianzas similares (homocedasticidad probable)")
else:
    print(f"   ✗ Varianzas muy diferentes (heterocedasticidad probable)")

# Verificar independencia (correlación serial simple)
corr_lag1 = np.corrcoef(residuos[:-1], residuos[1:])[0, 1]
print(f"\n4. Verificación de Independencia:")
print(f"   Correlación serial (lag 1): {corr_lag1:.4f}")
if abs(corr_lag1) < 0.2:
    print(f"   ✓ Baja autocorrelación (independencia probable)")
else:
    print(f"   ✗ Alta autocorrelación (dependencia probable)")

print("\n" + "=" * 60)
print("NOTA: Estas son verificaciones básicas. Para análisis rigurosos,")
print("se recomienda usar librerías especializadas como statsmodels.")
print("=" * 60)
````

## Limitaciones y Consideraciones

Es importante entender las limitaciones de la regresión lineal:

### 1. Asunción de Linealidad

La regresión lineal asume que la relación entre variables es lineal. En finanzas, muchas relaciones son no lineales (por ejemplo, opciones con convexidad).

### 2. Outliers

Los valores atípicos pueden tener un impacto desproporcionado en el modelo (crisis financieras, eventos extremos).

### 3. Multicolinealidad

Cuando hay múltiples variables independientes, si están altamente correlacionadas, puede causar problemas de interpretación.

### 4. Heterocedasticidad

La varianza de los errores puede no ser constante (común en datos financieros con volatilidad cambiante).

### 5. Autocorrelación

En series temporales financieras, los errores pueden estar correlacionados entre períodos.

### 6. Causalidad vs Correlación

La regresión muestra correlación, no necesariamente causalidad. En finanzas, esto es crucial para entender relaciones reales.

## Diferencia entre Estadística y Machine Learning

Es importante notar una diferencia clave:

### Enfoque Estadístico:
- Se enfoca en **inferencia** y **significancia estadística**
- Analiza normalidad de residuos, autocorrelación, heterocedasticidad
- Evalúa la confiabilidad de los parámetros estimados
- Busca entender las relaciones causales

### Enfoque de Machine Learning:
- Se enfoca en **predicción** y **generalización**
- Evalúa el rendimiento en datos no vistos
- Optimiza métricas de predicción (R², RMSE, etc.)
- Busca modelos que funcionen bien en nuevos datos

**Scikit-learn** sigue el enfoque de Machine Learning, por lo que no proporciona automáticamente análisis estadísticos detallados como pruebas de hipótesis o intervalos de confianza. Para análisis estadísticos más profundos, se recomienda usar librerías como `statsmodels`.
